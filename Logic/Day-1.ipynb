```python
#1. Take two numbers from the user and print their sum, difference, product, and division. 

num1 = int(input("Enter a first number :"))
num2 = int(input("Enter a second number :"))

print(f"Sum : {num1+num2}")
print(f"Difference : {abs(num1-num2)}")
print(f"Product : {num1*num2}")
print(f"Division : {num1/num2}")
```

```python
#2. Take a number as input and check whether it's even or odd. 

num = int(input("Enter a number :"))

if num % 2 == 0:
    print("Entered number is even")
else:
    print("Entered number is odd")
```

```python
#3. Given three numbers, find the largest one (without using max()). 

num1 = int(input("Enter a first number :"))
num2 = int(input("Enter a second number :"))
num3 = int(input("Enter a third number :"))

largest = num1
if num2 > largest:
    largest = num2
if num3 > largest:
    largest = num3

print(f"Largest number is {largest}")
```

```python
#4. Take age as input from the user and print whether they are a "minor" or "adult" (18+). 

age = int(input("Enter a age :"))

if age >= 18:
    print("Adult")
else:
    print("Minor")
```

```python
#5. Take a number and print its square and cube. 

num = int(input("Enter a number :"))

print(f"Square of {num} is {num**2}")
print(f"Cube of {num} is {num**3}")
```

```python
#6. Take a temperature in Celsius and convert it to Fahrenheit. (Formula: F = C × 9/5 + 32) 

Celsius = float(input("Enter a temperature (Celsius) :"))

Fahrenheit = (Celsius * 1.8 + 32)

print(f"{Celsius} Celsius = {Fahrenheit} Fahrenheit")
```

```python
#7. Print the multiplication table of a number from 1 to 10 using a loop.

num = int(input("Enter a number :"))

for i in range(1,11):
    print(f"{num} X {i} = {num*i}")
```

```python
#8. Take a number and print all numbers from 1 up to that number using a while loop. 

num = int(input("Enter a number :"))

i = 1
while i <= num:
    print(i)
    i += 1
```
