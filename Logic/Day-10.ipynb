```python
#1
"""
Take a string: s = "Mississippi". Count how many times the letter "s" appears 
— but using only .count() (no loops, no manual indexing) 
"""

s = "Mississippi"

print(s.count("s"))
```

```python
#2
"""
Take a string: s = " Python Is Awesome " (irregular spacing). Use a method to 
collapse all the extra spaces into single spaces between words, and strip the 
leading/trailing spaces — print the clean result. (Hint: think about .split() and 
.join() working together — .split() without arguments has a special behavior 
with whitespace.) 
"""

s = "  Python   Is   Awesome  "

words = s.split()
result = " ".join(words)

print(result)
```

```python
#3
"""
Take a string: s = "level". Check if it's a palindrome using only string methods 
and slicing (no loops) — but this time also handle mixed case, like "Level" 
should still count as a palindrome. 
"""

s = "level"

print(s.lower() == s[::-1].lower())
```

```python
#4
"""
Take a string: s = "2024-08-15" (a date in YYYY-MM-DD format). Use .split("-") 
to break it into year, month, day — then print it reformatted as DD/MM/YYYY 
using an f-string. 
"""

s = "2024-08-15"

sp = s.split("-")

year = sp[0]
month = sp[1]
day = sp[2]

result = f"{day}/{month}/{year}"
print(result)
```

```python
#5
"""
Take a string: s = "The rain in Spain". Use .replace() to replace only the first 
occurrence of "in" with "XYZ" (hint: .replace() has an optional third argument 
for count) — print the result and explain what happens if you don't limit the 
count. 
"""

s = "The rain in Spain"

print(s.replace("in","XYZ",1))
```

```python
#7
"""
Take a string: s = "PYTHON". Use an f-string with a formatting specifier to print 
it left-aligned in a field of width 10 with * as the fill character (hint: look up f
string alignment specifiers like <, >, ^). 
"""

s = "PYTHON"

print(f"{s:*<10}")
```

```python
#9
"""
Take a string: text = "münchen" (contains a unicode character ü). Print 
len(text), then encode it with UTF-8 and print len(text.encode("utf-8")) — 
compare the two lengths and explain in a comment why they're different. 
"""

text = "münchen"

print(len(text))

encoded = text.encode("utf-8")
print(len(encoded))
```
