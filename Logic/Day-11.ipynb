```python
#1
"""
Take two numbers using input(), convert them to int, and print their sum, 
difference, product, and division — then print the type() of the division result 
(notice it's different from int). 
"""

num1 = input("Enter a first number :")
num2 = input("Enter a second number :")

print(f"Sum : {int(num1)+int(num2)}")
print(f"Difference : {abs(int(num1)-int(num2))}")
print(f"Product : {int(num1)*int(num2)}")
print(f"Division : {int(num1)/int(num2)}")

print(type(int(num1)/int(num2)))
```

```python
#2
"""
Take a float number: x = 0.1 + 0.2. Print it directly — you'll notice it doesn't 
show exactly 0.3 (floating-point precision issue). Now use the decimal module 
(from decimal import Decimal) to compute Decimal("0.1") + Decimal("0.2") 
and print that instead — compare both outputs. 
"""

x = 0.1 + 0.2
print(x)

from decimal import Decimal

y = Decimal("0.1") + Decimal("0.2")
print(y)
```

```python
#3
"""
Take a fraction using the fractions module: create Fraction(1, 3) and 
Fraction(1, 6), add them together, and print the result — observe that Python 
keeps it as an exact fraction instead of a decimal. 
"""

from fractions import Fraction

a = Fraction(1,3)
b = Fraction(1,6)

print(a+b)
```

```python
#4
"""
Create a complex number z = 3 + 4j. Print its real part (.real), imaginary part 
(.imag), and its magnitude using abs(z) (this uses the Pythagorean theorem 
internally: √(real² + imag²)). 
"""

z = 3 + 4j

print(z.real)
print(z.imag)
print(abs(z))
```

```python
#5
"""
Using the math module, take a number from the user and print its square root 
(math.sqrt), its ceiling (math.ceil), and its floor (math.floor) — use a number 
with decimals, like 7.3, to see all three clearly. 
"""

import math

num = float(input("Enter a number :"))

print(math.sqrt(num))
print(math.ceil(num))
print(math.floor(num))
```

```python
#6
"""
Using the random module, generate and print a random integer between 1 
and 100 (random.randint), and a random float between 0 and 1 
(random.random). 
"""

import random

print(random.randint(1,100))

print(random.random())
```

```python
#7
"""
Take a number as a string, like num_str = "42". Convert it to int, then format it 
into a sentence using an f-string: "The number 42 has 2 digits" (use len() on 
the original string to count digits, not the int). 
"""

num_str = "42"

result = f"The number {int(num_str)} has {len(num_str)} digits"

print(result)
```

```python
#8
"""
Using math.pi, print the value of pi rounded to 4 decimal places using f-string 
formatting (.4f), and also print math.e (Euler's number) the same way. 
"""

import math

print(f"{math.pi:.4f}")
print(f"{math.e:.4f}")
```

```python
#9
"""
Take a number n = 17. Use the math module to check whether the square root 
of n is a whole number (hint: compare math.sqrt(n) to int(math.sqrt(n)) — if 
they're equal, it's a perfect square). Print True/False. 
"""

import math

n = 17

print(math.sqrt(n) == int(math.sqrt(n)))
```

```python
#10
"""
Combine strings and numbers: take a price = 499.99 and a discount_percent = 
20. Calculate the discounted price (price - (price * discount_percent / 100)), 
and print a formatted sentence like "Original: $499.99, Discount: 20%, Final: 
$399.99" — use f-string .2f formatting for all money values. 
"""

price = 499.99
discount_percent = 20

discounted = (price - (price * discount_percent / 100))

print(f"Original: ${price:.2f}, Discount: {discount_percent}%, Final: ${discounted:.2f}")
```
