```python
#1
"""
Take a number n = 7. Use math to check if it's a perfect square (reuse Day 11 
Q9 logic), and print the result as a full sentence using f-string: "7 is a perfect 
square: False". 
"""

import math

n = 7
result = math.sqrt(n) == int(math.sqrt(n))

print(f"{n} is a perfect square: {result}")
```

```python
#2
"""
Take Fraction(3, 4) and convert it to a float using float() — then convert that 
float back into a Fraction using Fraction(the_float). Print all three values and 
observe whether the round-trip gives back exactly 3/4 or something messier 
(hint: this connects to floating-point precision from Day 11 Q2). 
"""

from fractions import Fraction

fr = Fraction(3,4)
print(fr)

fl = float(Fraction(3,4))
print(fl)

con = Fraction(fl)
print(con)
```

```python
#3
"""
Take two complex numbers z1 = 2 + 3j and z2 = 1 - 1j. Print their sum, 
difference, and product — Python handles complex arithmetic automatically, 
just use +, -, *. 
"""

z1 = 2 + 3j
z2 = 1 - 1j

print((z1.real)+(z2.imag))
print((z1.real)-(z2.imag))
print((z1.real)*(z2.imag))
```

```python
#4
"""
Take price = "1299.999" (a string). Convert it to float, round it to 2 decimal 
places using f-string .2f, and print a sentence combining the original string and 
the rounded value: "Raw: 1299.999 → Rounded: $1300.00". 
"""

price = "1299.999"

rounded = float(price)

print(f"Raw: {price} → Rounded: ${rounded:.2f}")
```

```python
#5
"""
Use random.randint() to simulate a single dice roll (1 to 6). Print the result as a 
sentence using f-string, and also check using % whether the roll is even or odd, 
printing that in the same sentence. 
"""

import random

rint = random.randint(1, 6)
is_even = rint % 2 == 0

print(f"You rolled a {rint}. Is it even? {is_even}")
```

```python
#8
"""
Take radius = 5. Using math.pi, calculate the area of a circle (π × r²) and the 
circumference (2 × π × r) — print both formatted to 2 decimal places in one 
combined f-string sentence. 
"""

import math

radius = 5

area = math.pi * (radius**2)
circumference = 2 * math.pi * radius

print(f"Area: {area:.2f}, Circumference: {circumference:.2f}") 
```

```python
#9
"""
Take a string temp_str = "36.6" representing body temperature in Celsius. 
Convert it to float, calculate Fahrenheit (C × 9/5 + 32), and print a sentence 
with both values formatted to 1 decimal place: "36.6°C is 97.9°F". 
"""

temp_str = "36.6"

Fahrenheit = (float(temp_str) * 9/5) + 32

print(f"{float(temp_str):.1f}°C is {Fahrenheit:.1f}°F")
```
