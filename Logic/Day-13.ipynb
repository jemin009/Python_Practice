```python
#3
"""
Create a set: a = {1, 2, 3, 4} and b = {3, 4, 5, 6}. Print their union (|), 
intersection (&), and difference (a - b) — observe how sets automatically 
handle duplicates/overlaps. 
"""

a = {1, 2, 3, 4}
b = {3, 4, 5, 6}

print(a.union(b))
print(a.intersection(b))
print(a.difference(b))
```

```python
#4
"""
Create a dictionary: person = {"name": "Riya", "age": 25, "city": "Pune"}. Print 
just the .keys(), just the .values(), and just the .items() separately — observe 
the data type each one returns using type(). 
"""

person = {"name": "Riya", "age": 25, "city": "Pune"}

print(person.keys())
print(person.values())
print(person.items())

print(type(person.keys()))
print(type(person.values()))
print(type(person.items()))
```

```python
#5
"""
Take a list: fruits = ["apple", "banana", "cherry"]. Use .append() to add 
"mango", then use .insert() to add "kiwi" at index 1, then print the final list. 
"""

fruits = ["apple", "banana", "cherry"]

fruits.append("mango")
fruits.insert(1,"kiwi")

print(fruits)
```

```python
#6
"""
Create a list with duplicates: nums = [1, 2, 2, 3, 3, 3, 4]. Convert it to a set to 
remove duplicates, then convert that set back into a list — print all three 
(original list, set, final list) and their len(). 
"""

nums = [1, 2, 2, 3, 3, 3, 4]

convert = set(nums)
back = list(convert)

print(nums)
print(convert)
print(back)

print(len(nums))
print(len(convert))
print(len(back))
```

```python
#7
"""
Take a dictionary: scores = {"Amit": 85, "Riya": 90, "Sam": 75}. Use .get() to 
safely look up "Amit"'s score, and also use .get() to look up a name that 
doesn't exist (like "John") with a default value of 0 — print both results (this 
avoids the KeyError you'd get with scores["John"]). 
"""

scores = {"Amit": 85, "Riya": 90, "Sam": 75}

print(scores.get("Amit"))
print(scores.get("John",0))
```
