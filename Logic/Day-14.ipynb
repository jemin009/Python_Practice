```python
#1
"""
Take a list of 10 numbers from the user (using a loop, one at a time — like Day 
4 Q2's while pattern). Using a loop, count how many are even and how many 
are odd, and print both counts.
"""

number = []
even = []
odd = []
even_count = 0
odd_count = 0

for i in range(1,11):
    num = int(input(f"Enter a number ({i}): "))

    if num % 2 == 0:
        even.append(num)
        even_count += 1
    elif num % 2 != 0:
        odd.append(num)
        odd_count += 1

print(f"Even number : {even} Count : {even_count}")
print(f"Odd number : {odd} Count : {odd_count}")
```

```python
#5
"""
Take a number n from the user. Use a loop to check if it's prime (reuse your 
old prime logic) — but this time, wrap the whole thing so it keeps asking 
"Check another number? (yes/no)" and repeats until the user says no (use a 
while loop around the whole check). 
"""

num = int(input("Enter a number: "))

is_prime = num > 1

for i in range(2, num):
    if num % i == 0:
        is_prime = False
        break

print(f"{num} is prime: {is_prime}")
```

```python
#6
"""
Take a list of numbers: nums = [5, 12, 8, 21, 3, 17]. Loop through and build a 
new list containing only numbers that are prime (reuse prime-check logic 
inside the loop). 
"""

nums = [5, 12, 8, 21, 3, 17]

new = []

for num in nums:
    is_prime = num > 1
    
    for i in range(2, num):
        if num % i == 0:
            is_prime = False
            break
    
    if is_prime:
        new.append(num)

print(new)
```

```python
#7
"""
Take a sentence from the user. Loop through each character and build a 
dictionary counting how many times each vowel (a, e, i, o, u) appears — skip 
non-vowels using if. 
"""

sentence = input("Enter a sentence: ").lower()

new = {}
vowels = ['a', 'e', 'i', 'o', 'u']

for ch in sentence:
    if ch in vowels:
        if ch in new:
            new[ch] += 1
        else:
            new[ch] = 1

print(new)
```

```python
#9
"""
Simulate rolling a dice (random.randint(1,6)) 10 times using a loop, storing 
each result in a list. After the loop, print the list, and separately count how 
many times you rolled a 6.
"""

import random

list1 = []
count = 0

for i in range(10):
    roll = random.randint(1, 6)
    list1.append(roll)
    
    if roll == 6:
        count += 1

print(list1)
print(count)
```

```python
#10
"""
Take a list of mixed data: data = [15, "hello", 3.14, True, "world", 42]. Loop 
through it and use type() inside an if/elif chain to print each item labeled by its 
type: "15 is an int", "hello is a str", etc. 
"""

data = [15, "hello", 3.14, True, "world", 42]

for i in data:
    if type(i) == int:
        print(f"{i} is an int")
    elif type(i) == str:
        print(f"{i} is a str")
    elif type(i) == float:
        print(f"{i} is a float")
    elif type(i) == bool:
        print(f"{i} is a bool")
```
