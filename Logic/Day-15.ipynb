```python
#1
"""
Take a list: nums = [4, 15, 8, 23, 42, 16, 9]. Loop through and build two 
separate lists — one for numbers divisible by 4, another for numbers divisible 
by 3 (a number could end up in both, or neither). 
"""

nums = [4, 15, 8, 23, 42, 16, 9]

four = []
three = []

for i in nums:
    if i % 4 == 0:
        four.append(i)
    if i % 3 == 0:
        three.append(i)

print(f"Divisible by 3 {three}")
print(f"Divisible by 4 {four}")
```

```python
#2
"""
Take a sentence from the user. Loop through the words and find the longest 
word (without using max()) — print it along with its length. 
"""

sentence = input("Enter a sentence :")

longest = ""
for i in sentence.split():
    if len(i) > len(longest):
        longest = i

print(f"{longest} : {len(longest)}")
```

```python
#3
"""
Take a dictionary: inventory = {"apples": 50, "bananas": 0, "cherries": 30, 
"grapes": 0}. Loop through and print only the items that are out of stock 
(value == 0), formatted as: "bananas is out of stock". 
"""

inventory = {"apples": 50, "bananas": 0, "cherries": 30, "grapes": 0}

for key,value in inventory.items():
    if value == 0:
        print(f"{key} is out of stock")
```

```python
#4
"""
Take a number n from the user. Use a while loop to print the Collatz sequence 
starting from n: if the number is even, divide by 2; if odd, multiply by 3 and 
add 1 — repeat until you reach 1, printing each step. (Example: starting at 6 → 
6, 3, 10, 5, 16, 8, 4, 2, 1) 
"""

n = int(input("Enter a number: "))

while n != 1:
    print(n)

    if n % 2 == 0:
        n = n // 2
    else:
        n = n * 3 + 1

print(n)
```

```python
#5
"""
Take a list of words: words = ["apple", "Banana", "cherry", "Date", 
"elderberry"]. Loop through and build a new list containing only the words 
that start with a lowercase letter (hint: check word[0].islower()). 
"""

words = ["apple", "Banana", "cherry", "Date", "elderberry"]

new = []
for i in words:
    if i[0].islower():
        new.append(i)

print(new)
```

```python
#6
"""
Take two numbers from the user representing a range (start, end). Loop 
through that range and build a set of all numbers divisible by either 3 or 5 
(classic FizzBuzz-adjacent logic) — print the set and its length.
"""

num1 = int(input("Enter a starting number :"))
num2 = int(input("Enter a ending number :"))

new = set()
for i in range(num1,num2):
    if i % 3 == 0 or i % 5 == 0:
        new.add(i)

print(f"{new} : {len(new)}")
```

```python
#7
"""
Take a string: password = "MyP@ss123". Loop through each character and 
check (using flags/booleans) whether the password has at least one digit, at 
least one uppercase letter, and at least one special character (non
alphanumeric) — print all three results.
"""

password = "MyP@ss123"

has_digit = False
has_upper = False
has_special = False

for i in password:
    if i.isdigit():
        has_digit = True
    if i.isupper():
        has_upper = True
    if not i.isalnum():
        has_special = True

print(f"Has digit: {has_digit}")
print(f"Has uppercase: {has_upper}")
print(f"Has special character: {has_special}")
```

```python
#8
"""
Take a list of numbers: nums = [23, 45, 12, 67, 34, 89, 21]. Using a loop, find 
both the largest and second largest number in a single pass (reuse Day 4 Q7 
logic, but now properly using loops instead of hardcoded comparisons). 
"""

nums = [23, 45, 12, 67, 34, 89, 21]

first_largest = nums[0]
second_largest = nums[0]

for i in nums:
    if i > first_largest:
        second_largest = first_largest
        first_largest = i

    elif i > second_largest and i < first_largest:
        second_largest = i

print(f"First largest numbers is {first_largest}")
print(f"Second largest numbers is {second_largest}")
        
```

```python
#9
"""
Take a dictionary of student scores: scores = {"Amit": 85, "Riya": 45, "Sam": 
92, "Neha": 38, "Kiran": 67}. Loop through and build a new dictionary that 
maps each name to their grade (get_grade logic from Day 5 Q4 — reuse the 
A/B/C/D thresholds). 
"""

scores = {"Amit": 85, "Riya": 45, "Sam": 92, "Neha": 38, "Kiran": 67}

new = {}

def get_grade(marks):
    if marks >= 90:
        return "A"
    elif marks >= 75 and marks <= 89:
        return "B"
    elif marks >= 60 and marks <= 74:
        return "C"
    else:
        return "D"
    
for key,value in scores.items():
    new[key] = get_grade(value)

print(new)
```

```python
#10
"""
Take a number n from the user. Use nested loops to print a multiplication 
table grid for numbers 1 to n (like a times-table chart) — for n=3, output 
should look like: 
1  2  3 
2  4  6 
3  6  9
"""

num = int(input("Enter a number: "))

for i in range(1, num+1):
    for j in range(1, num+1):
        print(i*j, end=" ")
    print()
```
