```python
#1
"""
Take a list: nums = [3, 6, 9, 11, 14, 18, 20]. Loop through and build a dictionary 
that categorizes each number as "even" or "odd" as the key, with a list of 
matching numbers as the value — e.g., {"even": [6, 14, 18, 20], "odd": [3, 9, 
11]}. 
"""

nums = [3, 6, 9, 11, 14, 18, 20]

new = {}
even = []
odd = []

for i in nums:
    if i % 2 == 0:
        even.append(i)

    if i % 2 != 0:
        odd.append(i)

new["even"] = even
new["odd"] = odd

print(new)
```

```python
#2
"""
Take a string: text = "hello world". Loop through each character and build a 
dictionary counting every character's frequency (not just vowels this time) — 
skip spaces. 
"""

text = "hello world"

new = {}

for i in text.replace(" ",""):
    if i in new:
        new[i] += 1
    else:
        new[i] = 1

print(new)
```

```python
#3
"""
Take a list of numbers: nums = [10, 25, 3, 47, 8, 33]. Using a loop, find the 
number closest to 20 (smallest absolute difference) — print that number and 
the difference. 
"""

nums = [10, 25, 3, 47, 8, 33]

closest_num = nums[0]
smallest_diff = abs(nums[0] - 20)

for i in nums:
    if abs(i - 20) < smallest_diff:
        smallest_diff = abs(i - 20)
        closest_num = i

print(f"Closest number: {closest_num}, Difference: {smallest_diff}")
```

```python
#4
"""
Take a sentence from the user. Using a loop and a dictionary, count how many 
words have length 3, how many have length 4, how many have length 5+, 
grouped as {"short": count, "medium": count, "long": count} (short=≤3, 
medium=4, long=5+).
"""

sentence = input("Enter a sentence :")

new = {"Short": 0, "Medium": 0, "Long": 0}

for i in sentence.split():
    if len(i) <= 3:
        new["Short"] += 1
    elif len(i) == 4:
        new["Medium"] += 1
    else:
        new["Long"] += 1

print(new)
```

```python
#5
"""
Write a function is_armstrong(n) that checks if a number is an Armstrong 
number (reuse Day 4 Q9 concept, but now properly using a loop inside the 
function to sum the digit cubes — no hardcoding for 3-digit numbers only, 
make it work for any digit count using len(str(n))). 
"""

def is_armstrong(n):
    digits = str(n)
    power = len(digits)
    total = 0
    
    for digit in digits:
        total += int(digit) ** power
    
    return total == n

num = int(input("Enter a number: "))
print(is_armstrong(num))
```

```python
#6
"""
Take two lists: list1 = [1, 2, 3, 4, 5] and list2 = [10, 20, 30, 40, 50]. Using a loop 
and zip() (new: zip() pairs up elements from two lists), print each pair 
multiplied together: 1×10=10, 2×20=40, etc. 
"""

list1 = [1, 2, 3, 4, 5]
list2 = [10, 20, 30, 40, 50]

for a, b in zip(list1, list2):
    print(f"{a}*{b} = {a*b}")
```

```python
#7
"""
Take a list of numbers: nums = [5, -3, 8, -12, 0, 7, -1]. Using a loop, separate 
them into three lists: positives, negatives, and zeros — print all three. 
"""

nums = [5, -3, 8, -12, 0, 7, -1]

positive = []
negative = []
zero = []

for i in nums:
    if i > 0:
        positive.append(i)
    elif i < 0:
        negative.append(i)
    else:
        zero.append(i)

print(f"Positive list : {positive}")
print(f"Negative list : {negative}")
print(f"Zero list : {zero}")
```

```python
#8
"""
Write a function count_words(sentence) that takes a sentence and returns a 
dictionary of word frequencies (how many times each word appears) — test it 
with a sentence that has repeated words, like "the cat sat on the mat the cat 
ran".
"""

def count_words(sentence):
    new = {}
    
    for word in sentence.split():
        if word in new:
            new[word] += 1
        else:
            new[word] = 1
    
    return new

sentence = input("Enter a sentence: ")
result = count_words(sentence)
print(result)
```

```python
#9
"""
Take a number n. Using a loop, generate the first n triangular numbers (1, 3, 
6, 10, 15, ... where each number is the sum of all integers up to that position) 
and store them in a list — print the list. 
"""

n = int(input("Enter n: "))

triangular = []
total = 0

for i in range(1, n+1):
    total += i
    triangular.append(total)

print(triangular)
```

```python
#10
"""
 Take a list of tuples representing (name, score) pairs: students = [("Amit", 85), 
("Riya", 45), ("Sam", 92)]. Using a loop with tuple unpacking (for name, score 
in students:), print each student's name and whether they passed (score ≥ 40).
"""

students = [("Amit", 85), ("Riya", 45), ("Sam", 92)]

for name, score in students:
    if score >= 40:
        print(f"{name}: Pass")
    else:
        print(f"{name}: Fail")
```
