```python
#1
"""
Take a list of tuples: sales = [("Jan", 5000), ("Feb", 7200), ("Mar", 3100), 
("Apr", 8900)]. Using tuple unpacking in a loop, find and print the month with 
the highest sales (without using max()).
"""

sales = [("Jan", 5000), ("Feb", 7200), ("Mar", 3100), ("Apr", 8900)]

highest_sale = sales[0][1]
highest_month = sales[0][0]

for i in sales:
    if i[1] > highest_sale:
        highest_sale = i[1]
        highest_month = i[0]

print(f"Highest month : {highest_month}\nHighest sale : {highest_sale}")
```

```python
#2
"""
Write a function is_leap_year(year) that returns True/False based on leap year 
rules: divisible by 4, EXCEPT if divisible by 100 (not a leap year), UNLESS also 
divisible by 400 (then it IS a leap year). Test it on 2000, 1900, 2024, 2023.
"""

def is_leap_year(year):
    if year % 4 != 0:
        return False
    elif year % 100 != 0:
        return True
    elif year % 400 == 0:
        return True 
    else:
        return False

for test_year in [2000, 1900, 2024, 2023]:
    print(f"{test_year}: {is_leap_year(test_year)}")
```

```python
#3
"""
Take two lists: names = ["Amit", "Riya", "Sam"] and scores = [85, 45, 92]. Using 
zip(), build a dictionary mapping each name to their score, then loop through 
that dictionary to print who passed/failed (≥40).
"""

names = ["Amit", "Riya", "Sam"]
scores = [85, 45, 92]

new = {}
for name, score in zip(names, scores):
    new[name] = score

result = {}
for key, value in new.items():
    if value >= 40:
        result[key] = "Pass"
    else:
        result[key] = "Fail"

print(new)
print(result)
```

```python
#4
"""
Take a list of numbers: nums = [4, 8, 15, 16, 23, 42]. Write a function 
running_total(nums) that returns a new list where each element is the 
cumulative sum up to that point (e.g., [4, 12, 27, 43, 66, 108]) — reuse Day 16 
Q9's triangular number logic pattern. 
"""

nums = [4, 8, 15, 16, 23, 42]

def running_total(nums):
    total = 0
    new = []
    
    for i in nums:
        total += i
        new.append(total)
    return new

print(running_total(nums))
```

```python
#5
"""
Take a sentence from the user. Using a loop, build a dictionary where each 
unique word (case-insensitive) maps to a list of the positions (indices) where 
it appears in the sentence — e.g., "the cat and the dog" → {"the": [0, 3], "cat": 
[1], "and": [2], "dog": [4]}.
"""

sentence = "the cat and the dog"

words = sentence.lower().split()

new = {}
for i in range(len(words)):
    word = words[i]
    
    if word in new:
        new[word].append(i)
    else:
        new[word] = [i]

print(new)
```

```python
#6
"""
Write a function gcd(a, b) that finds the Greatest Common Divisor of two 
numbers using a loop (check every number from 1 up to the smaller of the 
two, keep track of the largest one that divides both evenly — no math.gcd() 
allowed, build it manually).
"""

def gcd(a, b):
    if a < b:
        smaller = a
    else:
        smaller = b
    
    result = 1
    for i in range(1, smaller + 1):
        if a % i == 0 and b % i == 0:
            result = i
    
    return result

print(gcd(12, 18))
```

```python
#7
"""
Take a list of dictionaries: orders = [{"item": "pen", "qty": 3, "price": 10}, 
{"item": "book", "qty": 2, "price": 150}, {"item": "bag", "qty": 1, "price": 500}]. 
Loop through and calculate the total cost of each order (qty × price), and print 
a summary sentence for each, plus a grand total at the end. 
"""

orders = [{"item": "pen", "qty": 3, "price": 10},
          {"item": "book", "qty": 2, "price": 150},
          {"item": "bag", "qty": 1, "price": 500}]

total = 0
for order in orders:
    cost = order['qty'] * order['price']
    total += cost
    print(f"{order['item']}: {order['qty']} x {order['price']} = {cost}")

print(f"Grand total: {total}")
```

```python
#8
"""
Take a number n. Using nested loops, print all Pythagorean triples (a, b, c 
where a²+b²=c²) where a, b, c are all ≤ n (hint: three nested loops, checking 
the condition inside the innermost one). Try n=20 and see what triples you 
find. 
"""

n = 20

for a in range(1, n+1):
    for b in range(1, n+1):
        for c in range(1, n+1):
            if a**2 + b**2 == c**2:
                print(f"{a}, {b}, {c}")
```

```python
#9
"""
Write a function flatten(list_of_lists) that takes a list containing other lists (like 
[[1,2,3], [4,5], [6,7,8,9]]) and returns a single flat list combining all elements 
([1,2,3,4,5,6,7,8,9]) — use nested loops (outer loop through each sublist, inner 
loop through each element).
"""

list_of_lists = [[1,2,3], [4,5], [6,7,8,9]]

def flatten(list_of_lists):
    new = []

    for i in list_of_lists:
        new.extend(i)

print(new)
```

```python
#10
"""
Take a dictionary of employees: employees = {"Amit": {"dept": "IT", "salary": 
50000}, "Riya": {"dept": "HR", "salary": 45000}, "Sam": {"dept": "IT", "salary": 
60000}} (a dictionary of dictionaries). Loop through and print each employee's 
department and salary, and separately calculate the total salary for IT 
department only. 
"""

employees = {"Amit": {"dept": "IT", "salary": 50000},
              "Riya": {"dept": "HR", "salary": 45000},
                "Sam": {"dept": "IT", "salary": 60000}}

new = {}
total = 0
for key,value in employees.items():
    new[key] = value['dept'], value['salary']

    if value["dept"] == 'IT':
        total += value['salary']

print(f"Employee : {new}\nIT department total salary : {total}")
```
