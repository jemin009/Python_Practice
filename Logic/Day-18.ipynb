```python
#1
"""
Take a list: nums = [12, 7, 19, 3, 25, 8, 14]. Using a loop, find the range (max - 
min) of the list without using max()/min() — track both the largest and 
smallest in a single pass.
"""

nums = [12, 7, 19, 3, 25, 8, 14]

smallest = nums[0]
largest = nums[0]

for i in nums: 
    if i < smallest:
        smallest = i
    elif i > largest:
        largest = i

range_value = largest - smallest

print(f"Smallest number : {smallest}\nLargest number : {largest}\nRange : {range_value}")
```

```python
#2
"""
Write a function reverse_words(sentence) that reverses the order of words in 
a sentence (not the letters) using a loop (no slicing [::-1] this time — build the 
reversed list manually by inserting at the front, or looping backward through 
indices). 
"""

def reverse_words(sentence):
    words = sentence.split()

    new = []
    for word in words:
        new.insert(0, word)
        
    return " ".join(new)

sentence = "the cat and the dog"
print(reverse_words(sentence))
```

```python
#3
"""
Take a list of numbers: nums = [1, 4, 2, 8, 5, 7]. Using nested loops, check if 
the list contains any two numbers that add up to 10 (print the pair if found, 
e.g., "2 + 8 = 10").
"""

nums = [1, 4, 2, 8, 5, 7]

for i in range(len(nums)):

    for j in range(len(nums)):

        if i != j and nums[i] + nums[j] == 10:

            print(f"{nums[i]} + {nums[j]} = {nums[i]+nums[j]}")
```

```python
#4
"""
Write a function is_anagram(word1, word2) that checks if two words are 
anagrams of each other (contain the exact same letters, possibly in different 
order — hint: sort both strings' characters and compare, or count character 
frequency using dictionaries).
"""

def is_anagram(word1, word2):
    return sorted(word1) == sorted(word2)

print(is_anagram("listen", "silent"))
print(is_anagram("hello", "world"))
```

```python
#5
"""
Take a dictionary: inventory = {"apples": 50, "bananas": 20, "cherries": 5, 
"dates": 0}. Using a loop, categorize each item into "high" (≥30), "medium" 
(10-29), "low" (1-9), or "out" (0) stock levels — build and print a new 
dictionary grouping item names under each category. 
"""

inventory = {"apples": 50, "bananas": 20, "cherries": 5, "dates": 0}

new = {"High": [], "Medium": [], "Low": [], "Out": []}

for item, qty in inventory.items():
    if qty >= 30:
        new["High"].append(item)
    elif qty >= 10:
        new["Medium"].append(item)
    elif qty >= 1:
        new["Low"].append(item)
    else:
        new["Out"].append(item)

print(new)
```

```python
#6
"""
Take a number n. Using a loop, generate the first n numbers of the Fibonacci 
sequence (reuse Day 4 Q8 logic) but this time also calculate and print the sum 
of all those Fibonacci numbers.
"""

n = int(input("Enter n: "))

a, b = 0, 1
fib_list = []
total = 0

for i in range(n):
    fib_list.append(a)
    total += a
    
    a, b = b, a + b

print(fib_list)
print(f"Sum: {total}")
```

```python
#7
"""
Take a string: text = "programming". Using a loop, find and print all repeated 
characters (characters that appear more than once) along with their counts — 
e.g., {"r": 2, "g": 2, "m": 2}.
"""

text = "programming"

new = {}
for i in text:
    count = text.count(i)

    if count >= 2:
        new[i] = count

print(new)
```

```python
#8
"""
Take a list of student dictionaries: students = [{"name": "Amit", "marks": [85, 
90, 78]}, {"name": "Riya", "marks": [45, 60, 55]}]. Loop through and calculate 
each student's average marks, printing a sentence like "Amit's average: 
84.33". 
"""

students = [{"name": "Amit", "marks": [85, 90, 78]},
            {"name": "Riya", "marks": [45, 60, 55]}]

for student in students:
    total = 0
    count = 0
    
    for mark in student['marks']:
        total += mark
        count += 1
    
    average = total / count
    print(f"{student['name']}'s average: {average:.2f}")
```

```python
#9
"""
Write a function caesar_cipher(text, shift) that shifts each letter in text 
forward by shift positions in the alphabet (e.g., "abc" with shift 1 becomes 
"bcd") — handle wraparound ("z" with shift 1 becomes "a") using the % 
operator, and only shift letters (leave spaces/punctuation unchanged). 
"""

def caesar_cipher(text, shift):
    result = ""
    for ch in text:
        if ch.isalpha():
            if ch.islower():
                new_char = chr((ord(ch) - ord('a') + shift) % 26 + ord('a'))
            else:
                new_char = chr((ord(ch) - ord('A') + shift) % 26 + ord('A'))
            result += new_char
        else:
            result += ch
    return result

print(caesar_cipher("abc", 1))
print(caesar_cipher("xyz", 3))
```

```python
#10
"""
 Take two dictionaries: dict1 = {"a": 1, "b": 2, "c": 3} and dict2 = {"b": 20, "c": 
30, "d": 40}. Using a loop, merge them into a new dictionary — if a key exists 
in both, add their values together (e.g., "b" should become 2+20=22), 
otherwise just keep the single value.
"""

dict1 = {"a": 1, "b": 2, "c": 3}
dict2 = {"b": 20, "c": 30, "d": 40}

new = {}

for key, value in dict1.items():
    new[key] = value

for key, value in dict2.items():
    if key in new:
        new[key] += value
    else:
        new[key] = value

print(new)
```
