```python
#1
"""
Take a list of numbers: nums = [7, 2, 9, 4, 11, 6]. Using a loop, check if the list 
is sorted in ascending order (compare each element to the next one — no 
sorted() allowed) — print True/False. 
"""

nums = [7, 2, 9, 4, 11, 6]

is_sorted = True

for i in range(len(nums) - 1):
    if nums[i] > nums[i+1]:
        is_sorted = False
        break

print(is_sorted)
```
