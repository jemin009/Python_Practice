```python
#1
"""
Take a list: nums = [8, 3, 5, 1, 9, 2]. Using a loop, implement bubble sort to 
sort it in ascending order (reuse Day 3 Q9 / Day 4 Q3 logic). 
"""

nums = [8, 3, 5, 1, 9, 2]

for i in range(len(nums)):

    for j in range(len(nums) - 1):
        
        if nums[j] > nums[j+1]:
            nums[j], nums[j+1] = nums[j+1], nums[j]

print(nums)
```
