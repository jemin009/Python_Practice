```python
#1
"""
Take a list: nums = [4, 8, 15, 16, 23, 42]. Loop through and print each number 
multiplied by 2. 
"""

nums = [4, 8, 15, 16, 23, 42]

new = []
for i in nums:
    new.append(i*2)

print(new)
```

```python
#2
"""
Take a list of names: names = ["Amit", "Riya", "Sam", "Neha"]. Loop through 
and print a greeting for each: "Hello, Amit!", etc. 
"""

names = ["Amit", "Riya", "Sam", "Neha"]

for i in names:
    print(f"Hello, {i}!")
```

```python
#3
"""
Take a number n from the user. Use a loop to print all numbers from 1 to n 
that are divisible by 5.
"""

n = int(input("Enter a number :"))

for i in range(1,n+1):
    if i % 5 == 0:
        print(i)
```

```python
#4
"""
Take a dictionary: prices = {"apple": 50, "banana": 20, "mango": 80}. Loop 
through and print each item with its price formatted as: "apple costs ₹50".
"""

prices = {"apple": 50, "banana": 20, "mango": 80}

for fruits, price in prices.items():
    print(f"{fruits} costs ₹{price}")
```

```python
#5
"""
Write a function add(a, b) that simply returns the sum of two numbers. Call it 
with 3 different pairs of numbers and print each result.
"""

def add(a, b):
    return a + b

print(add(12,15))
print(add(89,56))
print(add(45,67))
```

```python
#6
"""
Take a list of numbers: nums = [3, 7, 2, 9, 4]. Loop through and count how 
many numbers are greater than 5.
"""

nums = [3, 7, 2, 9, 4]

count = 0
for i in nums:
    if i > 5:
        count += 1

print(count)
```

```python
#7
"""
Take a string: word = "python". Loop through each character and print it on its 
own line, along with its index position (e.g., "0: p", "1: y").
"""

word = "python"

for i in range(len(word)):
    print(f"{i}: {word[i]}")
```

```python
#8
"""
Take a list: nums = [10, 20, 30]. Write a function list_sum(nums) that returns 
the total sum using a loop (no sum() built-in) — print the result.
"""

nums = [10, 20, 30]

def list_sum(nums):
    total = 0
    for i in nums:
        total += i

    return total

list_sum(nums)
```

```python
#9
"""
Take a number n from the user. Use a while loop to print a simple countdown 
from n to 1, then print "Liftoff!" at the end.
"""

n = int(input("Enter a number: "))

while n >= 1:
    print(n)
    n -= 1

print("Liftoff!")
```
