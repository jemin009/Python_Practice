```python
#1. Take a list of 5 numbers from the user and find their sum and average. 

list1 = list(map(int, input("Enter numbers separated by space: ").split()))

total = 0
count = 0
for i in list1:
    total += i
    count += 1

print(f"Sum of numbers is {total}")
print(f"Average of numbers is {total/count}")
```

```python
#2. Take a list of numbers and find the largest and smallest number in it (without using max()/min()). 

list1 = list(map(int,input("Enter a numbers separated by space :").split()))

maxx = list1[0]
minn = list1[0]

for i in list1:
    if i > maxx:
        maxx = i
    if i < minn:
        minn = i

print(f"Max is {maxx}")
print(f"Min is {minn}")
```

```python
#3. Take a sentence from the user and count how many vowels it contains.

sentence = input("Write one sentence: ")

vowels = ['a','e','i','o','u','A','E','I','O','U']

count = 0
for i in sentence:
    if i in vowels:
        count += 1

print(count)
    
```

```python
#4. Take a list of numbers and print only the even numbers from it. 

list1 = list(map(int,input("Enter a numbers separated with space :").split()))

new = []
for i in list1:
    if i % 2 == 0:
        new.append(i)

print(new)
```

```python
#5. Take a string and check if it's a palindrome, ignoring spaces and case. 
# (Example: "Was it a car or a cat I saw" type patterns like "nurses run") 

string = "Was it a car or a cat I saw"

if string.replace(" ","").lower() == string[::-1].replace(" ","").lower():
    print(f"Yes, '{string}' is palindrome")
else:
    print(f"No, '{string}' is not palindrome")
```

```python
#6. Take a list of numbers and remove duplicates from it (without using set()).

#list1 = list(map(int,input("Enter a numbers separated with space :").split))

list1 = [89,56,78,56,23,12]

new = []
for i in list1:
    if i not in new:
        new.append(i)

print(new)
```

```python
#7. Take a word from the user and count how many times each character appears in it. 
# (Example: "hello" → h:1, e:1, l:2, o:1) 

word = "hello"

count = {}
for i in word:
    if i in count:
        count[i] += 1
    else:
        count[i] = 1

print(count)
```

```python
#8. Take two lists and print only the common elements between them. 

list1 = [12,78,56,89,23]
list2 = [78,12,55,45,56]

new = []
for i in list1:
    if i in list2:
        new.append(i)

print(new)
```

```python
#9. Take a list of numbers and sort it in ascending order (without using .sort() or sorted() — write your own logic). 

list1 = [12,78,89,45,23,56,120,112,123,2,8]

new = []

for i in range(len(list1)):
    new.append(min(list1))
    list1.remove(min(list1))

print(new)
```

```python
#10.  Take a sentence and print it with the words in reverse order. 
# (Example: "I love Python" → "Python love I") 

sentence = "I love Python"

reverse = sentence.split()[::-1]
result = " ".join(reverse)
print(result)
```
