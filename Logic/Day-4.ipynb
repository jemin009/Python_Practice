```python
#1.
"""
(Repeat — Prime Number, new approach)  
Take a number and check if it's prime —  
but this time, only loop up to the square root of the number 

(hint: import math and use math.sqrt(num)).  
Think about why checking beyond the square root is unnecessary.
"""

import math

num = 7
is_prime = True

for i in range(2, int(math.sqrt(num)) + 1):
    if num % i == 0:
        is_prime = False
        break

print(is_prime)
```

```python
#2
"""
(Repeat — List input, new approach)  
Take a list of numbers from the user, but this time use a while loop instead of 
.split() —  

keep asking "Enter a number (or type 'done' to stop):" and build the list one 
number at a time. 
"""

num = []

while True:   
    value = input("Enter a number (or type 'done' to stop): ")

    if value.lower() == "done":
        break

    num.append(int(value))

print(num)
```

```python
#3
"""
(Repeat — Sorting, new approach)  
Write Bubble Sort from scratch (no min(), no .sort()) to sort a list in 
descending order this time (biggest to smallest). 
"""

list1 = [23,56,67,34,21,23,29,2,4,5,6,9]

for i in range(len(list1)):
    for j in range(len(list1) - 1):
        if list1[j] < list1[j+1]:
            list1[j], list1[j+1] = list1[j+1], list1[j]

print(list1)
```

```python
#4
"""
Write a function called is_even(num) that returns True if a number is even, 
False otherwise. Call it on 5 different numbers and print the results. 
"""

def is_even(num):
    if num % 2 == 0:
        return True
    else:
        return False

print(is_even(12))
print(is_even(7))
print(is_even(40))
print(is_even(120))
print(is_even(729))
```

```python
#5
"""
Write a function called factorial(num) that returns the factorial of a number 
(reuse your Day 2 logic, but wrap it in a function this time). 
"""

def factoarial(num):
    fact = 1
    for i in range(1,num+1):
        fact *= i
    return fact

print(factoarial(5))
```

```python
#6
"""
Write a function called count_vowels(sentence) that takes a sentence and 
returns the vowel count (reuse Day 3 logic, but as a function). 
"""

def count_vowels(string):
    vowels = ['a','e','i','o','u','A','E','I','O','U']

    count = 0
    for i in string:
        if i in vowels:
            count += 1

    return count

print(count_vowels("Jemin"))
```

```python
#7
"""
Take a list of numbers and find the second largest number in it (without 
sorting the whole list, without using max()). 
"""

list1 = [23,56,67,34,21,23,29,2,4,5,6,9]

f_large = float('-inf')
s_large = float('-inf')

for i in list1:
    if i > f_large:
        s_large = f_large 
        f_large = i
        
    elif i > s_large:
        s_large = i

print(f_large)
print(s_large)
```

```python
#8
"""
Write a function called is_palindrome(word) that returns True/False for 
whether a word is a palindrome. 
"""


def is_palindrome(string):
    if string.replace(" ","").lower() == string[::-1].replace(" ","").lower():
        return True
    else:
        return False


print(is_palindrome("Was it a car or a cat I saw")) 
```

```python
#9
"""
Take a number and check whether it's an Armstrong number. (Example: 153 = 
1³ + 5³ + 3³ = 1 + 125 + 27 = 153 ) 
"""

num = input("Enter a number :")

if (int(num[0])**3) + (int(num[1])**3) + (int(num[2])**3) == int(num):
    print(f"Yes, {num} is Armstrong number.")
else:
    print(f"No, {num} is Armstrong number.")
```

```python
#10
"""
Write a function called find_common(list1, list2) that returns a list of common 
elements between two lists (reuse Day 3 logic, but as a function). 
"""

def find_common(list1,list2):
    new = []
    for i in list1:
        if i in list2:
            new.append(i)
    return new

find_common([12,78,56,89,23],[78,12,55,45,56])
        
```
