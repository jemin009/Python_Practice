```python
#1
"""
Create a dictionary where 5 student names are keys and their marks are 
values. Print the name of the student with the highest marks (without using 
max()). 
"""

mark = { 
"Rahul":92, 
"Neha":51, 
"Amit":98, 
"Riya":69, 
"Jay":76, 
"Pooja":35 
}

high_mark = 0
high_stu = None

for key,value in mark.items():
    if value > high_mark:
        high_mark = value
        high_stu = key

print(high_stu,":",high_mark)
```

```python
#2. Find the average marks from the dictionary above.

mark = { 
"Rahul":92, 
"Neha":51, 
"Amit":98, 
"Riya":69, 
"Jay":76, 
"Pooja":35 
}

total = 0
count = 0

for value in mark.values():
    total += value
    count += 1
avg = total/count

print(avg)
```

```python
#4 
"""
Write a function get_grade(marks) that returns a grade based on marks: 90+ → 
"A", 75-89 → "B", 60-74 → "C", <60 → "D". Apply it to each student in the students 
list above and print name: grade. 
"""

students = [ 
    {"name": "Amit", "age": 20, "marks": 85}, 
    {"name": "Riya", "age": 22, "marks": 90}, 
    {"name": "Sam", "age": 21, "marks": 75} 
] 

def get_grade(marks):
    if marks >= 90:
        return "A"
    elif marks >= 75 and marks <= 89:
        return "B"
    elif marks >= 60 and marks <= 74:
        return "C"
    else:
        return "D"

for row in students:    
    print(row["name"],":",get_grade(row["marks"]))
```

```python
#5
"""
Write a function count_by_key(list_of_dicts, key) that takes a list of dictionaries 
and tells how many times each unique value appears for that key. (Example: 
count_by_key(students, "age") → how many students exist at each age) 
"""

students = [
    {"name": "Amit", "age": 20, "marks": 85},
    {"name": "Riya", "age": 22, "marks": 90},
    {"name": "Sam", "age": 21, "marks": 75},
    {"name": "Neha", "age": 20, "marks": 60}
]

def count_by_key(list_of_dicts, key):
    result = {}
    
    for row in list_of_dicts:
        value = row[key]

        if value in result:
            result[value] += 1
        else:
            result[value] = 1
    return result

print(count_by_key(students, "age"))
```
