```python
#1 
"""
Take your name and age using input(), and print both together in a single 
print() statement using concatenation. 
"""

name = input("Enter a name :")
age = input("Enter a age :")

print(name + " : " + age)
```

```python
#2
"""
Take a number using input() (it will come in as a string) — print its type(), then 
convert it to int and print the type() again. 
"""

number = input("Enter a number :")

print(type(number))

number = int(number)

print(type(number))
```

```python
#3
"""
Take a string: word = "python". Print the first character and last character 
using positive indexing.    
"""

word = "python"

print(word[0])

print(word[5])
```

```python
#4
"""
Same string word = "python" — print the last character using negative 
indexing (without using len()). 
"""

word = "python"

print(word[-1])
```

```python
#5. Take a string: text = "programming". Print its length using len(). 

text = "programming"

print(len(text))
```

```python
#6. Same string text = "programming" — use slicing to print only the "gram" part.

text = "programming"

print(text[3:7])
```

```python
#7 
"""
Take a string: s = "abcdefgh". Use slicing with a step to print only alternate 
characters (output should look like "acegh"-ish — think about what step value 
gives this). 
"""

s = "abcdefgh"

print(s[::2])
```

```python
#8
"""
Create two strings: first = "Hello" and second = "World". Concatenate them 
into a single string with a space in between, using the + operator (add the 
space manually). 
"""

first = "Hello" 
second = "World"

print(first + " " + second)
```

```python
#9
"""
Take a string: symbol = "-". Use the repetition operator to repeat it 20 times 
and print it as one line (like a separator line). 
"""

symbol = "-"

print(symbol * 20)
```

```python
#10
"""
Take a string: sentence = "I love Python programming". Use the membership 
operator (in) to check whether "Python" is present in the sentence, and print 
the result. 
"""

sentence = "I love Python programming"

print("Python" in sentence)
```
