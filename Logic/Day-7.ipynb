```python
#4
"""
Take a string: s = "12345678". Using slicing, print only the first half and then 
the second half as two separate print statements (don't hardcode the 
numbers — use len(s) to calculate the midpoint). 
"""

s = "12345678"

first = len(s) // 2
last = len(s)

print(s[:first])
print(s[first:last])
```

```python
#10
"""
Take a string: s = "banana". Count manually (using indexing, NOT a loop) how 
many times the character at index 1 ('a') appears — check indices 1, 3, and 5 
individually using the membership/equality idea, and print True/False for each 
one you check (this previews why loops will be useful later, but for now do it 
manually). 
"""

s = "banana"

count = 0
for i in s:
    if i == 'a':
        count += 1

print(count)

print(s[1] == 'a')
print(s[3] == 'a')
print(s[5] == 'a')
```
