```python
#3.
"""
Take a string: s = " Hello World ". Use a string method to remove the leading 
and trailing spaces, then print the result (also print its length before and after, 
to confirm the spaces were removed). 
"""

s = " Hello World "

print(f"Before length : {len(s)}")

s_stripped = s.strip()
print(s_stripped)

print(f"After length : {len(s_stripped)}")
```

```python
#4.
"""
Take a string: s = "python programming". Use string methods to print it in: (a) 
all uppercase, (b) all lowercase, (c) title case ("Python Programming"). 
"""

s = "python programming"

print(s.upper())
print(s.lower())
print(s.title())
```

```python
#5.
"""
Take a string: s = "Hello123". Use checking methods (.isalpha(), .isdigit(), 
.isalnum()) to test the whole string and print all three results — then explain in 
a comment why each gives the result it does. 
"""

s = "Hello123"

print(s.isalpha()) # return True if value has only alphabets
print(s.isdigit()) # return True if value has only digit
print(s.isalnum()) # return True if values has alphabets and digit both 
```

```python
#6.
"""
Take a string: sentence = "I like apples and apples are healthy". Use a string 
method to count how many times "apples" appears, and another method to 
replace "apples" with "mangoes" — print both results. 
"""

sentence = "I like apples and apples are healthy"

print(sentence.count("apples"))
print(sentence.replace("apples","mangoes"))
```

```python
#7.
"""
Take name and marks from the user, and print a sentence like "Jemin scored 
85 marks" using f-string formatting (not + concatenation this time). 
"""

name = input("Enter a name :")
marks = int(input("Enter a marks :"))

print(f"{name} scored {marks} marks")
```

```python
#8.
"""
Take a string: s = "python-is-fun". Use a string method to split it by "-" into a 
list, then use another method to join it back together with a space instead. 
"""

s = "python-is-fun"

sp = s.split("-")

print(" ".join(sp))
```

```python
#9.
"""
Take the string text = "café". Print its encoded form using UTF-8 (hint: 
.encode("utf-8")), then decode it back to the original string, and print both — 
observe what an encoded string actually looks like (bytes). 
"""

text = "café"

encoded = text.encode("utf-8")
print(encoded)

decoded = encoded.decode("utf-8")
print(decoded)
```

```python
#10.
"""
Take two variables: a = "hello" and b = "hello" (typed identically, separately). 
Use the is operator to check whether a and b point to the same object in 
memory — print the result and research/explain in a comment what you 
observe (this relates to string interning). 
"""

a = "hello"
b = "hello"

print(a == b)
print(a is b)
```
