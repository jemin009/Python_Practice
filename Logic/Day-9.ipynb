```python
#3.
"""
Take a string: s = "racecar". Using string methods only (no manual 
indexing/slicing), check if it starts with "race" and ends with "car" — print 
both results. 
"""

s = "racecar"

print(s.startswith("race"))
print(s.endswith("car"))
```

```python
#4.
"""
Take a sentence: sentence = "the Quick Brown fox". Use string methods to 
count how many words are capitalized (start with an uppercase letter) — hint: 
you'll need .split() combined with a checking method (but still no loops — 
check each word individually by indexing into the resulting list). 
"""

sentence = "the Quick Brown fox"
words = sentence.split()

print(words[0][0].isupper())   # check first letter of "the"
print(words[1][0].isupper())   # check first letter of "Quick"
print(words[2][0].isupper())   # check first letter of "Brown"
print(words[3][0].isupper())   # check first letter of "fox"
```

```python
#5.
"""
Take a number: price = 1234.5678. Use string formatting (f-string) to print it 
rounded to exactly 2 decimal places, formatted like "Price: $1234.57". 
"""

price = 1234.5678

print(f"Price: ${price:.2f}")
```

```python
#6.
"""
Take a string: s = "Hello, World! 123". Use checking methods to test .isalpha(), 
.isalnum(), and .isspace() on just the substring " " (a single space extracted 
from s using indexing) — print all three results and explain why each behaves 
that way. 
"""

s = "Hello, World! 123"

space_char = s[6]

print(space_char.isalpha())
print(space_char.isalnum())
print(space_char.isspace())
```
