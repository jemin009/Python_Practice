```python
#1 
"""
Take three numbers. 
Print the largest value using only arithmetic/comparison operators. 
(Do not use max())
"""

num1 = int(input("Enter a first number :"))
num2 = int(input("Enter a second number :"))
num3 = int(input("Enter a third number :"))

def highest(n1,n2,n3):
    if n1 > n2 and n1 > n3:
        return n1
    elif n2 > n1 and n2 > n3:
        return n2
    else:
        return n3

largest = highest(num1,num2,num3)

print(f"Largest values is {largest}")
```

```python
#2 
"""
Take three numbers. 
Print the middle value (neither largest nor smallest).
"""

num1 = int(input("Enter a first number :"))
num2 = int(input("Enter a second number :"))
num3 = int(input("Enter a third number :"))

def mid(n1, n2, n3):
    if (n1 > n2 > n3) or (n3 > n2 > n1):
        return n2
    elif (n2 > n1 > n3) or (n3 > n1 > n2):
        return n1
    else:
        return n3

middle = mid(num1,num2,num3)

print(f"Middle value is {middle}")
```

```python
#3 
"""
Take three subject marks. 

Print: 
 Total  
 Average  
 Highest mark  
 Lowest mark  
 Difference between highest and lowest  

(No max() or min().)
"""

mark1 = int(input("Enter first subject mark :"))
mark2 = int(input("Enter second subject mark :"))
mark3 = int(input("Enter third subject mark :"))

def highest(m1,m2,m3):
    if m1 > m2 and m1 > m3:
        return m1
    elif m2 > m1 and m2 > m3:
        return m2
    else: 
        return m3

def lowest(m1,m2,m3):
    if m1 < m2 and m1 < m3:
        return m1
    elif m2 < m1 and m2 < m3:
        return m2
    else:
        return m3

high = highest(mark1,mark2,mark3)
low = lowest(mark1,mark2,mark3)

print(f"Total : {mark1+mark2+mark3}")
print(f"Average : {(mark1+mark2+mark3)/3}")
print(f"Highest mark : {high}")
print(f"Lowest mark : {low}")
print(f"Difference : {high-low}")
```

```python
#4 
"""
Take length and breadth. 

Print: 
 Area  
 Perimeter  
 Difference between area and perimeter  
 Which is greater (True/False) 
"""

length = float(input("Enter a length :"))
breadth = float(input("Enter a breadth :"))

def great(n1,n2):
    if n1 > n2:
        return True
    else:
        return False

greater = great((length*breadth),(2 * (length + breadth)))

print(f"Area : {length*breadth}")
print(f"Perimeter : {2 * (length + breadth)}")
print(f"Difference = {abs((length*breadth) - (2 * (length + breadth)))}")
print(f"Greater : {greater}")
```

```python
#5 
"""
Take two numbers. 

Print: 
 Sum  
 Product  
 Whether the product is greater than the sum.  
 Difference between them. 
"""

num1 = int(input("Enter a first number :"))
num2 = int(input("Enter a second number :"))

def g(n1,n2):
    if n1 > n2:
        return True
    else:
        return False

greater = g((num1*num2),(num1+num2))

print(f"Sum : {num1+num2}")
print(f"Product : {num1*num2}")
print(f"Greater : {greater}")
print(f"Difference : {abs((num1*num2)-(num1+num2))}")
```

```python
#6
"""
Take principal, rate, and time. 

Calculate: 
 Simple Interest  
 Total Amount  

Print whether: 
 Interest > 5000  
 Amount > 2 × Principal  
"""

principal = int(input("Enter a principal :"))
rate = float(input("Enter a rate :"))
time = int(input("Enter a time :"))

def SI(n1,n2,n3):
    si = (n1*n2*n3) / 100
    return si
Simple_Interest = SI(principal,rate,time)

def TA(n1,n2):
    ta = n1 + n2
    return ta
total_amount = TA(principal,Simple_Interest)

def I(n1):
    if n1 > 5000:
        return True
    else:
        return False
Interest = I(Simple_Interest)

def A(n1):
    if n1 > 2*principal:
        return True
    else:
        return False
Amount = A(total_amount)

print(f"Simple Interest : {Simple_Interest}")
print(f"Total Amount : {total_amount}")

print(f"Interest : {Interest}")
print(f"Amount : {Amount}")
```

```python
#8 
"""
Take four numbers.
 
Print: 
 Largest  
 Smallest  
 Average  
 Difference between largest and smallest  

(Without max() and min())
"""

num1 = int(input("Enter a first number :"))
num2 = int(input("Enter a second number :"))
num3 = int(input("Enter a third number :"))
num4 = int(input("Enter a fourth number :"))

def Large(n1,n2,n3,n4):
    if n1 > n2 and n1 > n3 and n1 > n4:
        return n1
    elif n2 > n1 and n2 > n3 and n2 > n4:
        return n2
    elif n3 > n1 and n3 > n2 and n3 > n4:
        return n3
    else:
        return n4
largest = Large(num1,num2,num3,num4)

def Small(n1,n2,n3,n4):
    if n1 < n2 and n1 < n3 and n1 < n4:
        return n1
    elif n2 < n1 and n2 < n3 and n2 < n4:
        return n2
    elif n3 < n1 and n3 < n2 and n3 < n4:
        return n3
    else:
        return n4
smallest = Small(num1,num2,num3,num4)

average = (num1+num2+num3+num4) / 4

difference = abs(largest-smallest)

print(f"Largest : {largest}")
print(f"Smallest : {smallest}")
print(f"Average : {average}")
print(f"Difference : {difference}")
```

```python
#9 
"""
Take one number. 

Print: 
 Square  
 Cube  
 Half  
 Double  
 Triple  

Then print whether: 
Cube > Square
"""

num = int(input("Enter a number :"))

Square = num**2
Cube = num**3
Half = num/2
Double = num*2
Triple = num*3

def check(n1,n2):
    if n1>n2:
        return True
    else: 
        return False
Check = check(Cube,Square)

print(f"Square : {Square}")
print(f"Cube : {Cube}")
print(f"Half : {Half}")
print(f"Double : {Double}")
print(f"Triple : {Triple}")
print(f"Check : {Check}")
```
