```python
#1 
"""
Take three numbers. 

Print: 
• Largest 
• Smallest 
• Middle 
• Sum of Largest and Smallest 
• Whether Middle > Average of Largest and Smallest
"""

num1 = int(input("Enter a first number :"))
num2 = int(input("Enter a second number :"))
num3 = int(input("Enter a third number :"))

def large(n1,n2,n3):
    if n1 > n2 and n1 > n3:
        return n1
    elif n2 > n1 and n2 > n3:
        return n2
    else:
        return n3

largest = large(num1,num2,num3)

def small(n1,n2,n3):
    if n1 < n2 and n1 < n3:
        return n1
    elif n2 < n1 and n2 < n3:
        return n2
    else:
        return n3

smallest = small(num1,num2,num3)

def mid(n1,n2,n3):
    if n1 > n2 > n3 or n3 > n2 > n1:
        return n2
    elif n2 > n1 > n3 or n3 > n1 > n2:
        return n1
    else:
        return n3

middle = mid(num1,num2,num3)
average = (smallest + largest) / 2

print(f"Largest : {largest}")
print(f"Smallest : {smallest}")
print(f"Middle : {middle}")
print(f"Sum of Largest and Smallest : {largest+smallest}")
print(f"Whether Middle > Average of Largest and Smallest : {middle > average}")
```

```python
#2 
"""
Take principal, rate, and time. 

Calculate: 
• Simple Interest 
• Total Amount 
• Interest Percentage of Principal 

Print: 
Interest % > 25 
Amount > 1.5 × Principal 
"""

principal = int(input("Enter a principal :"))
rate = float(input("Enter a rate :"))
time = int(input("Enter a time :"))

def Interest(n1,n2,n3):
    simple_interest = (n1*n2*n3) / 100
    return simple_interest

simple_interest = Interest(principal,rate,time)

def Amount(n1,n2):
    total_amount = n1+n2
    return total_amount

total_amount = Amount(simple_interest,principal)

def percentage(n1,n2):
    interest_percentage = (n1/n2) * 100
    return interest_percentage

Interest_percentage = percentage(simple_interest,principal)

def Int(n1):
    if n1>25:
        return True
    else: 
        return False

con1 = Int(Interest_percentage)

def Amo(n1,n2):
    if n1>1.5*n2:
        return True
    else: 
        return False

con2 = Amo(total_amount,principal)

print(f"Simple Interest : {simple_interest}")
print(f"Total Amount  : {total_amount}")
print(f"Interest Percentage : {Interest_percentage}")
print(f"Interest % > 25 : {con1}")
print(f"Amount > 1.5 × Principal : {con2}")
```

```python
#3 
"""
Take four subject marks. 

Print: 
• Highest 
• Lowest 
• Average 
• Difference between Highest and Average 
• Difference between Lowest and Average 

Finally print: 
Highest Difference > Lowest Difference 
"""

mark1 = int(input("Enter a first subject mark :"))
mark2 = int(input("Enter a second subject mark :"))
mark3 = int(input("Enter a third subject mark :"))
mark4 = int(input("Enter a four subject mark :"))

def high(n1,n2,n3,n4):
    if n1 > n2 and n1 > n3 and n1 > n4:
        return n1
    elif n2 > n1 and n2 > n3 and n2 > n4:
        return n2
    elif n3 > n1 and n3 > n2 and n3 > n4:
        return n3
    else:
        return n4

highest = high(mark1,mark2,mark3,mark4)

def low(n1,n2,n3,n4):
    if n1 < n2 and n1 < n3 and n1 < n4:
        return n1
    elif n2 < n1 and n2 < n3 and n2 < n4:
        return n2
    elif n3 < n1 and n3 < n2 and n3 < n4:
        return n3
    else:
        return n4

lowest = low(mark1,mark2,mark3,mark4)

def avg(n1,n2,n3,n4):
    average = (n1+n2+n3+n4) / 4
    return average

average = avg(mark1,mark2,mark3,mark4)

def dif(n1,n2):
    difference = n1 - n2
    return difference

difference1 = dif(highest,average) 

def dif2(n1,n2):
    difference = n1 - n2
    return difference

difference2 = dif2(lowest,average) 

print(f"Highest : {highest}")
print(f"Lowest : {lowest}")
print(f"Average : {average}")
print(f"Difference (Highest - Average) : {difference1}")
print(f"Difference (Lowest - Average) : {difference2}")
print(f"Finally : {difference1>difference2}")
```

```python
#4 
"""
Take length, breadth and height.

Calculate: 
• Volume 
• Total Surface Area 
• Difference 

Print: 
Volume > Surface Area 
"""

Length = float(input("Enter a length :"))
Breadth = float(input("Enter a breadth :"))
Height = float(input("Enter a height :"))

def vol(n1,n2,n3):
    volume = n1*n2*n3
    return volume

volume = vol(Length,Breadth,Height)

def sur(n1,n2,n3):
    surface = 2 * ((n1*n2)+(n2*n3)+(n3*n1))
    return surface

surface = sur(Length,Breadth,Height)

def dif(n1,n2):
    difference = abs(n1-n2)
    return difference

difference = dif(volume,surface)

def ans(n1,n2):
    if n1 > n2:
        return True
    else: 
        return False

answer = ans(volume,surface)

print(f"Volume : {volume}")
print(f"Total Surface Area : {surface}")
print(f"Difference : {difference}")
print(f"Volume > Surface Area : {answer}")
```

```python
#5 
"""
Take three integers. 

Calculate: 
x = a + b 
y = b + c 
z = a + c 

Print: 
Largest value among x, y and z 
Smallest value among x, y and z 
Difference

Without max() and min().
"""

a = int(input("Enter a first number :"))
b = int(input("Enter a second number :"))
c = int(input("Enter a third number :"))

x = a + b 
y = b + c 
z = a + c 

def large(n1,n2,n3):
    if n1>n2 and n1>n3:
        return n1
    elif n2>n1 and n2>n3:
        return n2
    else:
        return n3

largest = large(x,y,z)

def small(n1,n2,n3):
    if n1<n2 and n1<n3:
        return n1
    elif n2<n1 and n2<n3:
        return n2
    else:
        return n3

smallest = small(x,y,z)

def dif(n1,n2):
    difference = abs(n1-n2)
    return difference

difference = dif(largest,smallest)

print(f"x = a + b : {x}")
print(f"y = b + c : {y}")
print(f"z = a + c : {z}")

print(f"Largest : {largest}")
print(f"Smallest : {smallest}")
print(f"Difference : {difference}")
```

```python
#6 
"""
Take two numbers 

Print: 
• Sum 
• Product 
• Square of Sum 
• Square of Product 

Finally print: 
Square of Product > Square of Sum 
"""

num1 = int(input("Enter a first number :"))
num2 = int(input("Enter a second number :"))

def tot(n1,n2):
    total = n1+n2
    return total

total = tot(num1,num2)

def pro(n1,n2):
    product = n1*n2
    return product

product = pro(num1,num2)

def ss(n1):
    square_sum = n1**2
    return square_sum

square1 = ss(total)

def sp(n1):
    square_product = n1**2
    return square_product

square2 = sp(product)

def ans(n1,n2):
    if n1>n2:
        return True
    else:
        return False

answer = ans(square2,square1)

print(f"Sum : {total}")
print(f"Product : {product}")
print(f"Square of Sum : {square1}")
print(f"Square of Product : {square2}")
```

```python
#7 
"""
Take four integers 

Calculate: 
A = first + second 
B = third + fourth 
C = first × fourth 

Print: 
A 
B 
C 

Finally print: 
(A > B) and (C > A)
"""

num1 = int(input("Enter a first number :"))
num2 = int(input("Enter a second number :"))
num3 = int(input("Enter a third number :"))
num4 = int(input("Enter a four number :"))

A = num1 + num2 
B = num3 + num4 
C = num1 * num4

def final(n1,n2,n3):
    if n1>n2 and n3>n1:
        return True
    else:
        return False
final = final(A,B,C)
    
print(f"{A}\n{B}\n{C}")
print(f"Final : {final}")
```

```python
#8 
"""
Take three numbers. 

Calculate: 
Average 
Now calculate the absolute difference of every number from the average. 
Print the number that is closest to the average. 

(Without min()) 

"""
num1 = int(input("Enter a first number :"))
num2 = int(input("Enter a second number :"))
num3 = int(input("Enter a third number :"))

def avg(n1,n2,n3):
    average = (n1+n2+n3) / 3
    return average

average = avg(num1,num2,num3)

diff1 = abs(num1 - average)
diff2 = abs(num2 - average)
diff3 = abs(num3 - average)

def closest(d1, d2, d3, n1, n2, n3):
    if d1 <= d2 and d1 <= d3:
        return n1
    elif d2 <= d1 and d2 <= d3:
        return n2
    else:
        return n3

closest_number = closest(diff1, diff2, diff3, num1, num2, num3)

print(f"Average : {average}")
print(f"Difference of {num1} : {diff1}")
print(f"Difference of {num2} : {diff2}")
print(f"Difference of {num3} : {diff3}")
print(f"Closest Number : {closest_number}")
```

```python
#9
"""
Take radius. 

Calculate: 
• Area of Circle 
• Circumference 

Print: 
Area 
Circumference 
Difference 

Finally print: 
Area > Circumference 
Use π = 3.14
"""

radius = float(input("Enter a radius :"))

def area(n1):
    area = 3.14*(n1**2)
    return area
area = area(radius)

def Circum(n1):
    circumference = 2 * 3.14 * n1
    return circumference
circumference = Circum(radius)

def dif(n1,n2):
    difference = abs(n1-n2)
    return difference
Difference = dif(area,circumference)

print(f"Area of Circle : {area}")
print(f"Circumference : {circumference}")
print(f"Difference : {Difference}")
print(f"Finally : {area > circumference}")
```

```python
#10 
"""
Take three integers. 

Calculate: 
x = a² + b² 
y = b² + c² 
z = a² + c² 
 
Print: 
x 
y 
z 
 
Finally print: 
Largest among x, y and z 
Smallest among x, y and z 
Difference 

(Without max() and min()) 
"""

a = int(input("Enter a first number :"))
b = int(input("Enter a second number :"))
c = int(input("Enter a third number :"))

x = (a**2) + (b**2) 
y = (b**2) + (c**2) 
z = (a**2) + (c**2) 

print(f"{x}\n{y}\n{z}")

def large(n1,n2,n3):
    if n1>n2 and n1>n3:
        return n1
    elif n2>n1 and n2>n3:
        return n2
    else:
        return n3

largest = large(x,y,z)

def small(n1,n2,n3):
    if n1<n2 and n1<n3:
        return n1
    elif n2<n1 and n2<n3:
        return n2
    else:
        return n3

smallest = small(x,y,z)

def dif(n1,n2):
    difference = abs(n1-n2)
    return difference
difference = dif(largest,smallest)

print(f"Largest among x, y and z is {largest}")
print(f"Smallest among x, y and z is {smallest}")
print(f"Difference is {difference}")
```
