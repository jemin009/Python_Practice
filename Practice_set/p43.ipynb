```python
#1 
"""
A company stores employee salaries.

Find: 
• Average salary 
• Employee whose salary is closest to the average 
• Difference between that salary and the average
"""

employees = { 
    "Rahul": 54000, 
    "Neha": 62000, 
    "Amit": 47000, 
    "Riya": 69000, 
    "Jay": 51000 
} 

total_sal = 0
total_count = 0

for key,value in employees.items():
    total_sal += value
    total_count += 1
avg = total_sal/total_count

print(f"Average salary : {avg}")

closest_employee = None
closest_salary = None
closest_difference = float("inf")

for employee, salary in employees.items():
    difference = abs(salary - avg)

    if difference < closest_difference:
        closest_difference = difference
        closest_employee = employee
        closest_salary = salary

print(f"Closest Employee : {closest_employee}")
print(f"Salary : {closest_salary}")
print(f"Difference : {closest_difference}")
```

```python
#2 
"""
Create a new dictionary whose key is product name and value is stock value (price × quantity). 

Include only products: 
• whose stock value is above the average stock value 
• whose quantity is below the average quantity
"""

products = { 
"Laptop":[82000,7], 
"Mobile":[41000,16], 
"Tablet":[28000,9], 
"TV":[98000,4], 
"Monitor":[51000,11], 
"Speaker":[17000,19] 
} 

new = {}

total_stock = 0
total_qty = 0
total_count = 0

for key,value in products.items():
    total_stock += value[0] * value[1]
    total_qty += value[1]
    total_count += 1
v_avg = total_stock/total_count
q_avg = total_qty/total_count

for key,value in products.items():
    if value[0] * value[1] > v_avg and value[1] < q_avg:
        new[key] = value[0] * value[1]

print(new)
```

```python
#3 
"""
Create a dictionary: 
Key → Student 
Value → Total Marks 

Include only students whose total marks are greater than the class average total.
"""

marks = { 
"Rahul":[82,75,91], 
"Neha":[94,89,87], 
"Amit":[61,73,69], 
"Riya":[98,96,99], 
"Jay":[78,84,80] 
} 

stu_total = {}
total_marks = 0
total_count = 0

for key,value in marks.items():
    stu_total[key] = value[0] + value[1] + value[2]
    total_marks += value[0] + value[1] + value[2]
    total_count += 1
avg = total_marks/total_count

new = {}
for key,value in stu_total.items():
    if value > avg:
        new[key] = value

print(new)
```

```python
#4
"""
Create a new list containing only numbers: 
• First digit > Last digit 
• Sum of digits is even 
• Number is greater than the average of the list
"""

numbers = [132,451,284,913,760,528,341]

total = 0
count = 0

for i in numbers:
    total += i
    count += 1

avg = total / count

new = []

for i in numbers:
    txt = str(i)

    digit_sum = int(txt[0]) + int(txt[1]) + int(txt[2])

    if int(txt[0]) > int(txt[-1]) and digit_sum % 2 == 0 and i > avg:
        new.append(i)

print(new)
```

```python
#5 
"""
Count customers who: 
• placed more than one order 
• total order value is greater than the average customer total order value 
"""

orders = [ 
["Rahul",13000], 
["Rahul",15000], 
["Amit",9000], 
["Neha",21000], 
["Jay",11000], 
["Jay",9000], 
["Riya",12000], 
["Riya",8000], 
["Karan",7000] 
] 

cus_value = {}
order_count = {}
total_value = 0
total_count = 0

for row in orders:
    customer = row[0]
    values = row[1]

    if customer in cus_value:
        cus_value[customer] += values
        order_count[customer] += 1
    else:
        cus_value[customer] = values
        order_count[customer] = 1
        total_count += 1

total_customer_value = 0

for value in cus_value.values():
    total_customer_value += value

avg = total_customer_value / len(cus_value)

count = 0

for customer, value in cus_value.items():
    if value > avg and order_count[customer] > 1:
        count += 1

print(count)
```

```python
#6 
"""
Create a dictionary: 
Key → Department 
Value → Average Salary 

Include only departments: 
• average salary > company average salary 
• department employee count == average department employee count
"""

employees = [ 
["Rahul","IT",68000], 
["Neha","IT",72000], 
["Amit","HR",49000], 
["Riya","HR",52000], 
["Jay","Sales",65000], 
["Karan","Sales",70000], 
["Pooja","Finance",91000], 
["Raj","Finance",56000] 
] 

dep_sal = {}
dep_count = {}

total_sal = 0
total_emp = 0

# Department total salary and employee count
for row in employees:
    department = row[1]
    salary = row[2]

    if department in dep_sal:
        dep_sal[department] += salary
        dep_count[department] += 1
    else:
        dep_sal[department] = salary
        dep_count[department] = 1

    total_sal += salary
    total_emp += 1

# Company average salary
company_avg = total_sal / total_emp

# Average department employee count
total_dep_emp = 0

for count in dep_count.values():
    total_dep_emp += count

avg_dep_emp = total_dep_emp / len(dep_count)

# Create new dictionary
new = {}

for department in dep_sal:

    department_avg = dep_sal[department] / dep_count[department]

    if department_avg > company_avg and dep_count[department] == avg_dep_emp:
        new[department] = department_avg

print(new)
```

```python
#7 
"""
Find the number whose 
(square - cube/number) 

is farthest from the average of all such values. 
"""

numbers = [41,52,63,74,85,96,107]

values = {}

total = 0
count = 0

for num in numbers:
    result = (num ** 2) - ((num ** 3) / num)
    values[num] = result
    total += result
    count += 1

avg = total / count

far_num = None
far_diff = -1

for num, value in values.items():
    diff = abs(value - avg)

    if diff > far_diff:
        far_diff = diff
        far_num = num

print("Number :", far_num)
print("Value :", values[far_num])
print("Difference :", far_diff)
```

```python
#9 
"""
Find the product whose 
(price ÷ quantity) 
is closest to the average 

and print: 
Product Name 
Price per Quantity 
Difference from Average
"""

inventory = { 
"Laptop":[76000,5], 
"Mobile":[42000,12], 
"Tablet":[29000,8], 
"TV":[95000,3], 
"Monitor":[51000,10] 
}

pro_value = {}
value_total = 0
value_count = 0 

for key, value in inventory.items():
    pro_value[key] = value[0] / value[1]
    value_total += value[0] / value[1]
    value_count += 1
avg = value_total/value_count

close_pro = None
close_price = 0
close_value = float("inf")

for key, value in pro_value.items():
    if abs(value - avg) < close_value:
        close_value = abs(value - avg)
        close_price = value

print(f"Product name : {close_pro}")
print(f"Price per Quantity : {close_price}")
print(f"Difference from average : {close_value}")
```
