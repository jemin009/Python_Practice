```python
#1 
"""
A company stores employee project hours. 

Find: 
• Average project hours 
• Employee whose hours are farthest from the average 
• Difference from the average
"""

employees = { 
"Rahul": 168, 
"Neha": 142, 
"Amit": 191, 
"Riya": 157, 
"Jay": 176 
}

total_hours = 0
total_count = 0

for key,value in employees.items():
    total_hours += value
    total_count += 1
avg = total_hours/total_count

far_emp = None
far_hour = 0

for key,value in employees.items():
    if abs(value-avg) > far_hour:
        far_hour = abs(value-avg)
        far_emp = key
        diff = abs(value-avg)

print(f"Average project hours : {avg}")
print(f"Employees whose hours are farthest from the average : {far_emp}")
print(f"Difference from the average : {diff}")
```

```python
#2 
"""
Create a new dictionary. 
Key → Product Name 
Value → Discounted Price (Price × 0.90) 

Include only products: 
• Discounted price > average discounted price 
• Quantity is an odd number
"""

products = { 
"Laptop":[82000,5], 
"Mobile":[39000,8], 
"Tablet":[31000,7], 
"TV":[99000,3], 
"Monitor":[46000,10], 
"Speaker":[21000,9] 
} 

pro_value = {}
pro_qty = {}
value_total = 0
value_count = 0

for key,value in products.items():
    pro_value[key] = value[0] * 0.9 
    pro_qty[key] = value[1]
    value_total += value[0] * 0.9 
    value_count += 1
avg = value_total/value_count

new = {}
for key,value in pro_value.items():
    if value > avg and pro_qty[key] % 2 != 0:
        new[key] = value
        
print(new)
```

```python
#3
"""
Create a dictionary: 
Key → Student 
Value → Highest mark among three subjects 

Include only students whose highest mark is above the average highest mark.
"""

marks = { 
"Rahul":[84,76,90], 
"Neha":[96,88,91], 
"Amit":[66,71,69], 
"Riya":[99,95,97], 
"Jay":[81,83,85] 
}

stu_mark = {}
total_mark = 0
total_count = 0

for key,value in marks.items():
    if value[0] > value[1] and value[0] > value[2]:
        stu_mark[key] = value[0]
        total_mark += value[0]
    elif value[1] > value[0] and value[1] > value[2]:
        stu_mark[key] = value[1]
        total_mark += value[1]
    else:
        stu_mark[key] = value[2]
        total_mark += value[2]
    total_count += 1
avg = total_mark/total_count

new = {}
for key,value in stu_mark.items():
    if value > avg:
        new[key] = value
print(new)
```

```python
#4 
"""
Create a new list containing only numbers: 
• Last digit is greater than first digit 
• Product of digits is odd 
• Number is less than the average of the list
"""

numbers = [173,482,951,337,615,729,864] 

total_num = 0
total_count = 0

for i in numbers:
    total_num  += i
    total_count += 1
avg = total_num/total_count

new = []
for i in numbers:
    txt = str(i)

    if (int(txt[0]) * int(txt[1]) * int(txt[2])) % 2 != 0 and i < avg and (int(txt[-1]) > int(txt[0])):
        new.append(i)

print(new)
```

```python
#6 
"""
Create a dictionary. 
Key → Department 
Value → Employee Count 

Include only departments: 
• Average salary is below company average salary 
• Employee count is greater than average department employee count
"""

employees = [ 
["Rahul","IT",68000], 
["Neha","IT",71000], 
["Amit","HR",48000], 
["Riya","HR",53000], 
["Jay","Sales",65000], 
["Karan","Sales",62000], 
["Pooja","Finance",89000], 
["Raj","Finance",60000], 
["Rohan","IT",69000] 
]

dep_sal = {}
dep_count = {}

sal_total = 0
emp_count = 0

total_emp = 0
count_dep = 0

for row in employees:
    employee = row[0]
    department = row[1]
    salary = row[2]
 
    if department in dep_sal:
        dep_sal[department] += salary
        dep_count[department] += 1
    else:
        dep_sal[department] = salary
        dep_count[department] = 1

        count_dep += 1
        
    total_emp += 1
    sal_total += salary
    emp_count += 1
 
e_avg = total_emp/count_dep
c_avg = sal_total/emp_count

new = {}
for key, value in dep_sal.items():
    d_avg = dep_sal[key] / dep_count[key]

    if d_avg < c_avg and dep_count[key] > e_avg:
        new[key] = dep_count[key]

print(new)
```

```python
#7
"""
Find the number whose 
(square + cube) 
is closest to the average of all calculated values. 

Print: 
• Number 
• Calculated Value 
• Difference from Average 
"""

numbers = [18,22,31,27,35,40,29] 

plus = 0
total_count = 0

for i in numbers:
    plus += (i**2 + i**3)
    total_count += 1
avg = plus/total_count

close_var = float("inf")
close_num = 0
cal_value = 0

for i in numbers:
    if abs((i**2 + i**3) - avg) < close_var:
        close_var = abs((i**2 + i**3) - avg)
        close_num = i
        cal_value = (i**2 + i**3)

print(f"Number : {close_num}")
print(f"Calculated Value : {cal_value}")
print(f"Difference from Average : {close_var}")
```

```python
#8 
"""
Count students who scored: 
• above subject average in exactly two subjects 
• below subject average in exactly one subject
"""

students = [ 
["Rahul",85,76,91], 
["Neha",95,92,88], 
["Amit",67,72,70], 
["Riya",98,97,99], 
["Jay",81,84,86] 
]

m1_total = 0
m2_total = 0
m3_total = 0
total_count = 0

for row in students:
    student = row[0]
    m1 = row[1]
    m2 = row[2]
    m3 = row[3]

    m1_total += m1
    m2_total += m2
    m3_total += m3
    total_count += 1

m1_avg = m1_total/total_count
m2_avg = m2_total/total_count
m3_avg = m3_total/total_count

count = 0
for row in students:
    student = row[0]
    m1 = row[1]
    m2 = row[2]
    m3 = row[3]

    above = 0
    below = 0

    if m1 > m1_avg:
        above += 1
    elif m1 < m1_avg:
        below += 1

    if m2 > m2_avg:
        above += 1
    elif m2 < m2_avg:
        below += 1

    if m3 > m3_avg:
        above += 1
    elif m3 < m3_avg:
        below += 1

    if above == 2 and below == 1:
        count += 1

print(count)
```

```python
#9 
"""
Find the product whose 
(price × quantity) 
is closest to the average stock value. 

Print: 
• Product Name 
• Stock Value 
• Difference from Average 
"""

inventory = { 
"Laptop":[78000,4], 
"Mobile":[41000,11], 
"Tablet":[32000,6], 
"TV":[93000,2], 
"Monitor":[49000,8] 
}

pro_value = {}
value_total = 0
value_count = 0

for key, value in inventory.items():
    pro_value[key] = value[0] * value[1]
    value_total += value[0] * value[1]
    value_count += 1
avg = value_total/value_count

close_pro = None
close_value = float("inf")
s_value = 0

for key,value in pro_value.items():
    if abs(value - avg) < close_value:
        close_value = abs(value - avg)
        close_pro = key
        s_value = value

print(f"Product name : {close_pro}")
print(f"Stock Value : {s_value}")
print(f"Difference from Average : {close_value}")
```

```python
#10
"""
Find the department having the lowest average salary 
among departments whose total salary is above the company average department 
total salary.
"""

employees = [ 
["Rahul","IT",69000], 
["Neha","IT",74000], 
["Amit","HR",51000], 
["Riya","HR",56000], 
["Jay","Sales",62000], 
["Karan","Sales",70000], 
["Pooja","Finance",95000], 
["Raj","Finance",61000] 
]

dep_sal = {}
emp_count = {}

total_sal = 0
total_count = 0

for row in employees:
    department = row[1]
    salary = row[2]

    if department in dep_sal:
        dep_sal[department] += salary
        emp_count[department] += 1
    else:
        dep_sal[department] = salary
        emp_count[department] = 1

        total_count += 1
    total_sal += salary
c_avg = total_sal / total_count

lowest_dep = None
lowest_avg = float("inf")

for key, value in dep_sal.items():
    d_avg = dep_sal[key] / emp_count[key]

    if value > c_avg:
        if d_avg < lowest_avg:
            lowest_avg = d_avg
            lowest_dep = key

print(lowest_dep)
```
