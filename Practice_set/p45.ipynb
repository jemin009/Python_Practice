```python
#1 
""" 
A company stores employee monthly sales. 

Find: 
• Average sales 
• Employee whose sales are closest to the average 
• Difference from the average 
Do not use min(). 
""" 

employees = { 
"Rahul": 58000, 
"Neha": 63000, 
"Amit": 52000, 
"Riya": 61000, 
"Jay": 56000 
}

total_val = 0
total_count = 0

for value in employees.values():
    total_val += value
    total_count += 1
avg = total_val/total_count

close_val = float("inf")
close_emp = 0
var = 0

for key,value in employees.items():
    if abs(value - avg) < close_val:
        close_val = abs(value - avg)
        close_emp = key
        var = abs(value - avg)

print(f"Average sales : {avg}")
print(f"Employee whose sales are closest to the average  : {close_emp}")
print(f"Difference from the average  : {var}")
```

```python
#2 
""" 
Create a new dictionary. 
Key → Product Name 
Value → Profit (Selling Price − Cost Price) 

Include only products: 
• Profit is above average profit 
• Quantity is even 

Format: 
[ Selling Price , Cost Price , Quantity ] 
""" 

products = { 
"Laptop":[85000,70000,6], 
"Mobile":[42000,36000,11], 
"Tablet":[30000,25000,8], 
"TV":[97000,82000,4], 
"Monitor":[52000,47000,7] 
} 

pro_pro = {}
pro_qty = {}

total_pro = 0
total_count = 0

for key, value in products.items():
    pro_pro[key] = value[0] - value[1]
    pro_qty[key] = value[2]

    total_pro += value[0] - value[1]
    total_count += 1
avg = total_pro/total_count

new = {}
for key, value in pro_pro.items():
    if value > avg and pro_qty[key] % 2 == 0:
        new[key] = value

print(new)
```

```python
#3 
""" 
Create a dictionary. 
Key → Student 
Value → Lowest mark among three subjects 

Include only students whose lowest mark 
is above the average lowest mark. 
""" 

marks = { 
"Rahul":[88,79,92], 
"Neha":[95,91,89], 
"Amit":[70,68,75], 
"Riya":[99,96,98], 
"Jay":[84,82,86] 
} 

stu_mark = {}
total_mark = 0
total_count = 0

for key,value in marks.items():
    if value[0] < value[1] and value[0] < value[2]:
        stu_mark[key] = value[0]
        total_mark += value[0]
    elif value[1] < value[0] and value[1] < value[2]:
        stu_mark[key] = value[1]
        total_mark += value[1] 
    else:
        stu_mark[key] = value[2]
        total_mark += value[2]

    total_count += 1
avg = total_mark/total_count

new = {}
for key,value in stu_mark.items():
    if value > avg:
        new[key] = value

print(new)
```

```python
#4 
""" 
Create a new list containing only numbers: 
• First digit is even 
• Sum of digits is greater than 15 
• Number is above the average of the list 
""" 

numbers = [284,517,862,439,756,981,624]

total_sum = 0
total_count = 0

for i in numbers:
    total_sum += i
    total_count += 1
avg = total_sum/total_count

new = []
for i in numbers:
    txt = str(i)

    if (int(txt[0]) + int(txt[1]) + int(txt[2])) > 15 and i > avg and int(txt[0]) % 2 == 0:
        new.append(i)

print(new)
```

```python
#5 
""" 
Count customers who: 
• placed exactly three orders 
• total order value is greater than average customer total order value 
""" 

orders = [ 
["Rahul",8000], 
["Rahul",9000], 
["Rahul",10000], 
["Neha",18000], 
["Jay",7000], 
["Jay",9000], 
["Jay",11000], 
["Riya",12000], 
["Amit",6000], 
["Amit",5000] 
]

cus_value = {}
cus_count = {}

total_value = 0
total_count = 0

for row in orders:
    customer = row[0]
    values = row[1]

    if customer in cus_value:
        cus_value[customer] += values
        cus_count[customer] += 1
    else:
        cus_value[customer] = values
        cus_count[customer] = 1

        total_count += 1
    total_value += values

avg = total_value/total_count

count = 0
for key,value in cus_value.items():
    if value > avg and cus_count[key] == 3:
        count += 1

print(count)
```

```python
#6 
""" 
Create a dictionary. 
Key → Department 
Value → Total Salary 

Include only departments: 
• Average salary is above company average salary 
• Employee count is equal to average department employee count 
""" 

employees = [ 
["Rahul","IT",72000], 
["Neha","IT",68000], 
["Amit","HR",51000], 
["Riya","HR",56000], 
["Jay","Sales",64000], 
["Karan","Sales",66000], 
["Pooja","Finance",92000], 
["Raj","Finance",58000] 
]

dep_sal = {}
emp_count = {}

emp_total = 0
dep_count = 0

sal_total = 0
sal_count = 0

for row in employees:
    department = row[1]
    salary = row[2]

    if department in dep_sal:
        dep_sal[department] += salary
        emp_count[department] += 1
    else:
        dep_sal[department] = salary
        emp_count[department] = 1

        dep_count += 1
    sal_count += 1
    sal_total += salary
    emp_total += 1

c_avg = sal_total/sal_count
emp_avg = emp_total/dep_count

new = {}
for key,value in dep_sal.items():
    dep_avg = dep_sal[key] / emp_count[key]

    if dep_avg > c_avg and emp_count[key] == emp_avg:
       new[key] = value

print(new)
```
