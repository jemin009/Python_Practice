```python
#1
""" 
A company stores employee yearly bonuses. 

Find: 
• Average bonus 
• Employee whose bonus is farthest from the average 
• Difference from the average 

Do not use max() or min(). 
""" 

employees = { 
"Rahul": 185000, 
"Neha": 142000, 
"Amit": 97000, 
"Riya": 221000, 
"Jay": 168000 
}

total_value = 0
total_count = 0

for value in employees.values():
    total_value += value
    total_count += 1
avg = total_value/total_count

far_emp = None
far_bonus = 0
diff = 0

for key,value in employees.items():
    if abs(value - avg) > far_bonus:
        far_bonus = abs(value - avg)
        far_emp = key
        diff = abs(value - avg)

print(f"Average bonus : {avg}")
print(f"Employee whose bonus is farthest from the average : {far_emp}")
print(f"Difference from the average : {diff}")
```

```python
#2 
""" 
Create a new dictionary. 
Key → Product Name 
Value → Stock Value (Price × Quan ty) 
 
Include only products: 
• Stock Value is below average stock value 
• Price is greater than average price 

Format 
[Price, Quantity] 
""" 

products = { 
    "Laptop":[84000,4], 
    "Mobile":[39000,15], 
    "Tablet":[32000,7], 
    "TV":[102000,2], 
    "Monitor":[47000,6], 
    "Speaker":[18000,20] 
} 

total_val = 0
val_count = 0 

total_price = 0

for value in products.values():
    total_val += value[0] * value[1]
    total_price += value[0]

    val_count += 1
v_avg = total_val/val_count
p_avg = total_price/val_count

new = {}
for key,value in products.items():
    if (value[0] * value[1]) < v_avg and value[0] > p_avg:
        new[key] = value[0]*value[1]

print(new)
```

```python
#3
""" 
Create a dictionary. 
Key → Student 
Value → Average Marks 

Include only students whose average marks 
are greater than the class average marks. 
""" 

marks = { 
"Rahul":[84,79,88], 
"Neha":[96,92,91], 
"Amit":[69,72,70], 
"Riya":[99,97,98], 
"Jay":[82,85,81] 
}

stu_avg = {}

avg_total = 0
stu_count = 0

for key,value in marks.items():
    stu_avg[key] = (value[0] + value[1] + value[2]) / 3

    avg_total += (value[0] + value[1] + value[2]) / 3
    stu_count += 1
avg_avg = avg_total/stu_count

new = {}
for key,value in stu_avg.items():
    if value > avg_avg:
        new[key] = value

print(new)
```

```python
#4
""" 
Create a new list containing only numbers: 
• Middle digit is odd 
• Product of digits is greater than 60 
• Number is below the average of the list 
""" 

numbers = [357,824,691,573,248,935,462]

total_sum = 0
total_count = 0

for i in numbers:
    total_sum += i
    total_count += 1
avg = total_sum/total_count

new = []
for i in numbers:
    txt = str(i)

    if int(txt[1]) % 2 != 0 and (int(txt[0]) * int(txt[1]) * int(txt[2])) > 60 and i < avg:
        new.append(i)
print(new)
```

```python
#5
""" 
Count customers who: 
• placed exactly two orders 
• average order value is greater than 

average customer order value 
""" 

orders = [ 
["Rahul",12000], 
["Rahul",15000], 
["Neha",18000], 
["Neha",24000], 
["Amit",9000], 
["Jay",11000], 
["Jay",17000], 
["Riya",22000], 
["Karan",7000] 
]

order_value = {}
cus_count = {}

total_values = 0
values_count = 0

for row in orders:
    customer = row[0]
    values = row[1]

    if customer in order_value:
        order_value[customer] += values
        cus_count[customer] += 1
    else:
        order_value[customer] = values
        cus_count[customer] = 1

        values_count += 1
    total_values += values
avg = total_values/values_count

count = 0
for key,value in order_value.items():
    avg_order = order_value[key] / cus_count[key]

    if avg_order > avg and cus_count[key] == 2:
        count += 1

print(count)
```

```python
#6 
""" 
Create a dictionary. 
Key → Department 
Value → Employee Count 

Include only departments: 
• Total salary is above average department total salary 
• Average salary is below company average salary 
""" 

employees = [ 
["Rahul","IT",71000], 
["Neha","IT",67000], 
["Amit","HR",49000], 
["Riya","HR",54000], 
["Jay","Sales",69000], 
["Karan","Sales",65000], 
["Pooja","Finance",93000], 
["Raj","Finance",61000], 
["Rohan","Sales",64000] 
]

dep_sal = {}
emp_count = {}

total_sal = 0
dep_count = 0

total_emp = 0

for row in employees:
    department = row[1]
    salary = row[2]

    if department in dep_sal:
        dep_sal[department] += salary
        emp_count[department] += 1
    else:
        dep_sal[department] = salary
        emp_count[department] = 1

        dep_count += 1
    total_sal += salary
    total_emp += 1

d_avg = total_sal/dep_count
c_avg = total_sal/total_emp

new = {}
for key,value in dep_sal.items():
    avg_sal = dep_sal[key] / emp_count[key]

    if value > d_avg and avg_sal < c_avg:
        new[key] = emp_count[key]

print(new)
```

```python
#7 
""" 
Find the number whose 
(square × first digit) 
is closest to the average of all such values. 

Print: 
• Number 
• Calculated Value 
• Difference 
""" 

numbers = [27,48,63,75,39,84,56]

total_num = 0
count_num = 0

for i in numbers:
    txt = str(i)
    total_num += (i**2) * (int(txt[0])) 
    count_num += 1
avg = total_num/count_num

close_num = float("inf")
ans_num = None
cal_value = 0

for i in numbers:
    txt = str(i)
    value = (i**2) * int(txt[0])

    if abs(value - avg) < close_num:
        close_num = abs(value - avg)
        ans_num = i
        cal_value = value

print(f"Number : {ans_num}")
print(f"Calculated Value : {cal_value}")
print(f"Difference : {close_num}")
```

```python
#8 
""" 
Count students who scored: 
• above subject average in exactly two subjects 
• equal to or below subject average in exactly one subject 
""" 

students = [ 
["Rahul",88,79,91], 
["Neha",95,94,90], 
["Amit",71,68,74], 
["Riya",99,98,97], 
["Jay",83,86,84] 
]

m1_total = 0
m2_total = 0
m3_total = 0
marks_count = 0

for row in students:
    student = row[0]
    m1 = row[1]
    m2 = row[2]
    m3 = row[3]

    m1_total += m1
    m2_total += m2
    m3_total += m3
    marks_count += 1

m1_avg = m1_total/marks_count
m2_avg = m2_total/marks_count
m3_avg = m3_total/marks_count

count = 0
for row in students:
    student = row[0]
    m1 = row[1]
    m2 = row[2]
    m3 = row[3]

    above = 0
    below = 0

    if m1 > m1_avg:
        above += 1
    elif m1 <= m1_avg:
        below += 1

    if m2 > m2_avg:
        above += 1
    elif m2 <= m2_avg:
        below += 1

    if m3 > m3_avg:
        above += 1
    elif m3 <= m3_avg:
        below += 1

    if above == 2 and below == 1:
        count += 1

print(count)
    
```

```python
#9
""" 
Find the product whose 
(Price ÷ Quantity) 
is farthest from the average. 

Print: 
• Product Name 
• Price per Quantity 
• Difference 

Format: 
[Price, Quantity] 
""" 

inventory = { 
"Laptop":[80000,5], 
"Mobile":[42000,12], 
"Tablet":[33000,6], 
"TV":[96000,3], 
"Monitor":[51000,10] 
} 

pro_value = {}
value_total = 0
value_count = 0

for key,value in inventory.items():
    pro_value[key] = value[0] / value[1]

    value_total += value[0] / value[1]
    value_count += 1
avg = value_total/value_count

far_value = 0
far_pro = None
price_qty = 0
dif = 0

for key,value in pro_value.items():
    if abs(value - avg) > far_value:
        far_value = abs(value - avg)
        far_pro = key
        price_qty = value
        dif = abs(value - avg)


print(f"Product Name : {far_pro}")
print(f"Price per Quantity : {price_qty}")
print(f"Difference : {dif}")
```

```python
#10
""" 
Find the department having the lowest average salary 
 
among departments whose employee count 
is greater than the average department employee count. 
""" 

employees = [ 
["Rahul","IT",69000], 
["Neha","IT",71000], 
["Amit","HR",51000], 
["Riya","HR",56000], 
["Jay","Sales",62000], 
["Karan","Sales",66000], 
["Rohan","Sales",64000], 
["Pooja","Finance",91000], 
["Raj","Finance",59000] 
]

dep_sal = {}
emp_count = {}

total_emp = 0
dep_count = 0

for row in employees:
    department = row[1]
    salary = row[2]

    if department in dep_sal:
        dep_sal[department] += salary
        emp_count[department] += 1
    else:
        dep_sal[department] = salary
        emp_count[department] = 1
        dep_count += 1

    total_emp += 1

emp_avg = total_emp / dep_count

low_dep = None
low_avg = float("inf")

for key, value in emp_count.items():
    avg_sal = dep_sal[key] / emp_count[key]

    if emp_count[key] > emp_avg and avg_sal < low_avg:
        low_avg = avg_sal
        low_dep = key

print(low_dep)
print(low_avg)
```
