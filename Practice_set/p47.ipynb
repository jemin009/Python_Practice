```python
#1
""" 
A company stores employee yearly incentives. 

Find: 
• Average incentive 
• Employee whose incentive is closest to the average 
• Difference from the average 

Do not use max() or min(). 
""" 

employees = { 
"Rahul":178000, 
"Neha":149000, 
"Amit":103000, 
"Riya":216000, 
"Jay":171000 
}

total_inc = 0
inc_count = 0

for value in employees.values():
    total_inc += value
    inc_count += 1
avg = total_inc/inc_count

close_inc = float("inf")
close_emp = None
dif = 0

for key, value in employees.items():
    if abs(value - avg) < close_inc:
        close_inc = abs(value - avg)
        close_emp = key
        dif = abs(value - avg)

print(f"Average incentive : {avg}")
print(f"Employee whose incentive is closest to the average : {close_emp}")
print(f"Difference from the average  : {dif}")
```

```python
#2
""" 
Create a new dictionary. 
Key → Product Name 
Value → Profit Percentage 
Profit % = ((Selling Price - Cost Price) / Cost Price) × 100 

Include only products: 
• Profit percentage is above average profit percentage 
• Quantity is odd 

Format: 
[Selling Price, Cost Price, Quantity] 
""" 

products = { 
"Laptop":[86000,72000,5], 
"Mobile":[43000,38000,10], 
"Tablet":[31000,27000,9], 
"TV":[101000,83000,3], 
"Monitor":[54000,50000,8], 
"Speaker":[22000,18000,7] 
}

pro_pro = {}
pro_qty = {}


profit_total = 0
profit_count = 0

for key,value in products.items():
    pro_pro[key] = ((value[0]-value[1]) / value[1]) * 100
    pro_qty[key] = value[2]

    profit_total += ((value[0]-value[1]) / value[1]) * 100
    profit_count += 1
avg = profit_total/profit_count

new = {}
for key,value in pro_pro.items():
    if value > avg and pro_qty[key] % 2 != 0:
        new[key] = value

print(new)
```

```python
#3
""" 
Create a dictionary. 
Key → Student 
Value → Total Marks 

Include only students whose total marks 
are below the class average total marks. 
""" 

marks = { 
"Rahul":[87,82,91], 
"Neha":[96,93,95], 
"Amit":[68,72,70], 
"Riya":[99,98,97], 
"Jay":[81,84,83] 
}

stu_marks = {}
marks_total = 0
stu_count = 0

for key, value in marks.items():
    stu_marks[key] = value[0]+value[1]+value[2]
    marks_total += value[0]+value[1]+value[2]
    stu_count += 1
avg = marks_total/stu_count

new = {}
for key, value in stu_marks.items():
    if value < avg:
        new[key] = value

print(new)
```

```python
#4
""" 
Create a new list containing only numbers: 
• Last digit is even 
• Sum of digits is greater than 18 
• Number is above the average of the list 
""" 

numbers = [486,729,864,351,942,675,588]

total_num = 0
num_count = 0

for i in numbers:
    total_num += i
    num_count += 1
avg = total_num/num_count

new = []
for i in numbers:
    txt = str(i)

    if (int(txt[0]) + int(txt[1]) + int(txt[2])) > 18 and (int(txt[-1]) % 2 == 0) and i > avg:
        new.append(i)

print(new)
```

```python
#5
""" 
Count customers who: 
• placed exactly two orders 
• total order value is less than average customer total order value 
""" 

orders = [ 
["Rahul",11000], 
["Rahul",14000], 
["Neha",17000], 
["Neha",20000], 
["Amit",9000], 
["Jay",12000], 
["Jay",10000], 
["Riya",23000], 
["Karan",6000] 
]

cus_value = {}
cus_count = {}  

total_value = 0
cus_total = 0

for row in orders:
    customer = row[0]
    values = row[1]

    if customer in cus_value:
        cus_value[customer] += values
        cus_count[customer] += 1
    else:
        cus_value[customer] = values
        cus_count[customer] = 1

        cus_total += 1
    total_value += values
avg = total_value/cus_total

count = 0
for key, value in cus_value.items():
    if value < avg and cus_count[key] == 2:
        count += 1

print(count)
```

```python
#6
""" 
Create a dictionary. 
Key → Department 
Value → Average Salary 

Include only departments: 
• Employee count is greater than average department employee count 
• Average salary is above company average salary 
""" 

employees = [ 
["Rahul","IT",71000], 
["Neha","IT",73000], 
["Amit","HR",50000], 
["Riya","HR",54000], 
["Jay","Sales",67000], 
["Karan","Sales",65000], 
["Rohan","Sales",66000], 
["Pooja","Finance",94000], 
["Raj","Finance",62000] 
]

dep_sal = {}
emp_count = {}

total_sal = 0
sal_count = 0

dep_count = 0

for row in employees:
    department = row[1]
    salary = row[2]

    if department in dep_sal:
        dep_sal[department] += salary
        emp_count[department] += 1
    else:
        dep_sal[department] = salary
        emp_count[department] = 1
        dep_count += 1

    total_sal += salary
    sal_count += 1

avg_emp = sal_count/dep_count
avg_sal = total_sal/sal_count

new = {}
for key, value in dep_sal.items():
    avg_dep = dep_sal[key] / emp_count[key]

    if avg_dep > avg_sal and emp_count[key] > avg_emp:
        new[key] = dep_sal[key] / emp_count[key]

print(new)
```

```python
#7
""" 
Find the number whose 
(cube ÷ last digit) 
is farthest from the average of all such values. 

Print: 
• Number 
• Calculated Value 
• Difference 
""" 

numbers = [24,35,46,57,68,79,82]

total_num = 0
num_count = 0

for i in numbers:
    txt = str(i)
    total_num += (i**3) / int(txt[-1]) 
    num_count += 1
avg = total_num/num_count

far_value = 0
far_num = 0
cal_value = 0
dif = 0

for i in numbers:
    txt = str(i)

    if abs(((i**3) / int(txt[-1])) - avg) > far_value:
        far_value = abs(((i**3) / int(txt[-1])) - avg)
        cal_value = (i**3) / int(txt[-1])
        dif = abs(((i**3) / int(txt[-1])) - avg)
        far_num = i

print(f"Number : {far_num}")
print(f"Calculated Value : {cal_value}")
print(f"Difference : {dif}")
```

```python
#8
""" 
Count students who scored: 
• above subject average in exactly one subject 
• equal to or below subject average in exactly two subjects 
""" 

students = [ 
["Rahul",89,81,92], 
["Neha",97,94,95], 
["Amit",69,73,71], 
["Riya",99,98,97], 
["Jay",84,87,83] 
]

m1_total = 0
m2_total = 0
m3_total = 0
total_count = 0

for row in students:
    student = row[0]
    m1 = row[1]
    m2 = row[2]
    m3 = row[3]

    m1_total += m1
    m2_total += m2
    m3_total += m3
    total_count += 1

m1_avg = m1_total/total_count
m2_avg = m2_total/total_count
m3_avg = m3_total/total_count

count = 0
for row in students:
    student = row[0]
    m1 = row[1]
    m2 = row[2]
    m3 = row[3]

    above = 0
    below = 0

    if m1 > m1_avg:
        above += 1
    elif m1 <= m1_avg:
        below += 1

    if m2 > m2_avg:
        above += 1
    elif m2 <= m2_avg:
        below += 1

    if m3 > m3_avg:
        above += 1
    elif m3 <= m3_avg:
        below += 1

    if above == 1 and below == 2:
        count += 1

print(count)
```

```python
#9
""" 
Find the product whose 
(Price × Quantity) 
is farthest from the average stock value. 

Print: 
• Product Name 
• Stock Value 
• Difference 

Format: 
[Price, Quantity] 
""" 

inventory = { 
"Laptop":[82000,4], 
"Mobile":[43000,10], 
"Tablet":[34000,7], 
"TV":[98000,2], 
"Monitor":[50000,9] 
}

pro_val = {}

val_total = 0
val_count = 0

for key,value in inventory.items():
    pro_val[key] = value[0] * value[1]
    val_total += value[0] * value[1]
    val_count += 1
avg = val_total/val_count

far_value = 0
far_pro = None
stock = 0
dif = 0

for key,value in pro_val.items():
    if abs(value - avg) > far_value:
        far_value = abs(value - avg)
        far_pro = key
        stock = value
        dif = abs(value - avg)

print(f"Product Name : {far_pro}")
print(f"Stock Value : {stock}")
print(f"Difference : {dif}")
```

```python
#10
""" 
Find the department having the highest average salary 
 
among departments whose employee count 
is equal to the average department employee count. 
""" 
 
employees = [ 
["Rahul","IT",70000], 
["Neha","IT",72000], 
["Amit","HR",51000], 
["Riya","HR",55000], 
["Jay","Sales",63000], 
["Karan","Sales",67000], 
["Pooja","Finance",93000], 
["Raj","Finance",61000] 
]

dep_sal = {}
emp_num = {}

emp_total = 0
dep_count = 0

for row in employees:
    customer = row[1]
    salary = row[2]

    if customer in dep_sal:
        dep_sal[customer] += salary
        emp_num[customer] += 1
    else:
        dep_sal[customer] = salary
        emp_num[customer] = 1

        dep_count += 1
    emp_total += 1
avg_emp = emp_total/dep_count

high_sal = 0
high_dep = None

for key, value in dep_sal.items():
    avg_sal = dep_sal[key] / emp_num[key]

    if avg_sal > high_sal and emp_num[key] == avg_emp:
        high_sal = avg_sal
        high_dep = key

print(high_dep)
```
