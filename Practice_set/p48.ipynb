```python
#1
""" 
A company stores yearly performance bonuses. 

Find: 
• Average bonus 
• Employee whose bonus is second farthest from the average 
• Difference from the average 

Do not use max(), min(), sorted(). 
""" 

employees = { 
"Rahul":192000, 
"Neha":151000, 
"Amit":98000, 
"Riya":229000, 
"Jay":176000, 
"Pooja":135000 
}

total_bon = 0
bon_count = 0

for value in employees.values():
    total_bon += value
    bon_count += 1
avg = total_bon/bon_count

f_high_value = 0
f_high_pro = None

s_high_value = 0
s_high_pro = None

for key,value in employees.items():
    if abs(value - avg) > f_high_value:
        s_high_value = f_high_value
        s_high_pro = f_high_pro

        f_high_value = abs(value - avg)
        f_high_pro = key

    elif abs(value - avg) > s_high_value and abs(value - avg) < f_high_value:
        s_high_value = abs(value - avg)
        s_high_pro = key

print(f"Average bonus : {avg}")
print(f"Employee whose bonus is second farthest from the average  : {s_high_pro}")
print(f"Difference from the average : {s_high_value}")
```

```python
#2
""" 
Create a new dictionary. 
Key → Product Name 
Value → Profit Percentage 

Profit % = ((Selling Price - Cost Price) / Cost Price) × 100 

Include only products: 
• Profit % is above average profit % 
• Quantity is even 
• Selling Price is greater than average selling price 

Format: 
[Selling Price, Cost Price, Quantity] 
""" 

products = { 
"Laptop":[91000,76000,6], 
"Mobile":[46000,41000,9], 
"Tablet":[34000,29000,8], 
"TV":[108000,89000,4], 
"Monitor":[56000,51000,10], 
"Speaker":[24000,20000,5] 
}

pro_sell = {}
pro_qty = {}
pro_per = {}

sell_total = 0
sell_count = 0
per_total = 0

for key,value in products.items():
    pro_per[key] = (((value[0] - value[1]) / value[1])) * 100
    pro_sell[key] = value[0]
    pro_qty[key] = value[2]

    sell_total += value[0]
    sell_count += 1
    per_total += (((value[0] - value[1]) / value[1])) * 100

avg_sell = sell_total/sell_count
avg_per = per_total/sell_count

new = {}
for key,value in products.items():
    if value[0] > avg_sell and pro_qty[key] % 2 == 0 and pro_per[key] > avg_per:
        new[key] = (((value[0] - value[1]) / value[1])) * 100

print(new)
```

```python
#3
""" 
Create a dictionary. 
Key → Student 
Value → Highest Marks 

Include only students whose highest marks 
are above the average highest marks. 
""" 

marks = { 
"Rahul":[88,84,91], 
"Neha":[97,94,96], 
"Amit":[71,69,73], 
"Riya":[99,98,100], 
"Jay":[85,87,83] 
}

stu_mark = {}
stu_total = 0
stu_count = 0

for key,value in marks.items():
    if value[0] > value[1] and value[0] > value[2]:
        stu_mark[key] = value[0]
        stu_total += value[0]
    elif value[1] > value[0] and value[1] > value[2]:
        stu_mark[key] = value[1]
        stu_total += value[1]
    else:
        stu_mark[key] = value[2]
        stu_total += value[2]

    stu_count += 1
avg = stu_total/stu_count

new = {}
for key,value in stu_mark.items():
    if value > avg:
        new[key] = value

print(new)
```

```python
#4 
""" 
Create a new list containing only numbers: 
• First digit is odd 
• Sum of digits is divisible by 3 
• Product of digits is greater than 80 
• Number is below the average of the list 
""" 

numbers = [357,684,915,572,489,963,741] 

total_num = 0
num_count = 0

for i in numbers:
    total_num += i
    num_count += 1
avg = total_num/num_count

new = []
for i in numbers:
    txt = str(i)

    if int(txt[0]) % 2 != 0 and ((int(txt[0]) + int(txt[1]) + int(txt[2]))) % 3 == 0 and ((int(txt[0]) * int(txt[1]) * int(txt[2]))) > 80 and i < avg:
        new.append(i)

print(new)
```

```python
#5
""" 
Count customers who: 
• placed exactly three orders 
• average order value is greater than 
average customer average order value 
""" 

orders = [ 
["Rahul",12000], 
["Rahul",14000], 
["Rahul",15000], 
["Neha",18000], 
["Neha",22000], 
["Jay",11000], 
["Jay",13000], 
["Jay",16000], 
["Amit",9000], 
["Riya",24000], 
["Karan",7000] 
]

cus_value = {}
order_count = {}

values_total = 0
cus_count = 0 

for row in orders:
    customer = row[0]
    values = row[1]

    if customer in cus_value:
        cus_value[customer] += values
        order_count[customer] += 1
    else:
        cus_value[customer] = values
        order_count[customer] = 1

        cus_count += 1
    values_total += values
avg = values_total/cus_count

count = 0
for key,value in cus_value.items():
    avg_order = cus_value[key] / order_count[key]

    if value < avg_order and order_count[key] == 3:
        count += 1

print(count)
```
