```python
#1
""" 
A company stores yearly performance bonuses. 

Find: 
• Average bonus 
• Employee whose bonus is second closest to the average 
• Difference from the average 

Do not use max(), min(), sorted(). 
""" 

employees = { 
"Rahul":186000, 
"Neha":154000, 
"Amit":101000, 
"Riya":218000, 
"Jay":172000, 
"Pooja":139000 
}

sum_value = 0
value_count = 0
for value in employees.values():
    sum_value += value
    value_count += 1
avg = sum_value/value_count

f_close_sal = float("inf")
f_close_emp = None

s_close_sal = float("inf")
s_close_emp = None

for key, value in employees.items():
    if abs(value - avg) < f_close_sal:
        s_close_sal = f_close_sal
        s_close_emp = f_close_emp

        f_close_sal = abs(value - avg)
        f_close_emp = key

    elif abs(value - avg) < s_close_sal and abs(value - avg) > f_close_sal:
        s_close_sal = abs(value - avg)
        s_close_emp = key

print(f"Average bonus : {avg}")
print(f"Employee whose bonus is second closest to the average : {s_close_emp}")
print(f"Difference from the average : {s_close_sal}")
```

```python
#2
""" 
Create a new dictionary. 
Key → Product Name 
Value → Stock Value (Selling Price × Quan ty) 
 
Include only products: 
• Stock value is above average stock value 
• Profit % is above average profit % 
• Quantity is odd 
 
Profit % = ((Selling Price - Cost Price) / Cost Price) × 100 
 
Format: 
[Selling Price, Cost Price, Quantity] 
""" 
 
products = { 
"Laptop":[93000,77000,5], 
"Mobile":[47000,42000,8], 
"Tablet":[36000,31000,7], 
"TV":[112000,92000,3], 
"Monitor":[58000,52000,10], 
"Speaker":[25000,21000,9] 
}

pro_stock = {}
pro_per = {}
pro_qty = {}

stock_sum = 0
pro_count = 0

per_sum = 0

for key,value in products.items():
    pro_stock[key] = value[0] * value[2]
    pro_per[key] = ((value[0] - value[1]) / value[1]) * 100 
    pro_qty[key] = value[2]

    stock_sum += value[0] * value[2] 
    per_sum += ((value[0] - value[1]) / value[1]) * 100 
    pro_count += 1

avg_stock = stock_sum/pro_count
avg_per = per_sum/pro_count

new = {}
for key, value in pro_stock.items():
    if value > avg_stock and pro_per[key] > avg_per and pro_qty[key] % 2 != 0:
        new[key] = value 

print(new)
```

```python
#3
""" 
Create a dictionary. 
Key → Student 
Value → Lowest Marks 

Include only students whose lowest marks 
are below the average lowest marks. 
""" 

marks = { 
"Rahul":[91,83,89], 
"Neha":[98,95,96], 
"Amit":[69,72,70], 
"Riya":[100,99,98], 
"Jay":[84,87,82] 
}

stu_mark = {}
mark_sum = 0
stu_count = 0

for key,value in marks.items():
    if value[0] < value[1] and value[0] < value[2]:
        stu_mark[key] = value[0]
        mark_sum += value[0]
    elif value[1] < value[0] and value[1] < value[2]:
        stu_mark[key] = value[1]
        mark_sum += value[1]
    else:
        stu_mark[key] = value[2]
        mark_sum += value[2]

    stu_count += 1
avg_mark = mark_sum/stu_count

new = {}
for key, value in stu_mark.items():
    if value < avg_mark:
        new[key] = value

print(new)
```

```python
#4 
""" 
Create a new list containing only numbers: 
• Last digit is odd 
• Sum of digits is greater than 16 
• Product of digits is divisible by 5 
• Number is above the average of the list 
""" 

numbers = [485,726,951,573,864,395,642]

num_sum = 0
num_count = 0
for i in numbers:
    num_sum += i
    num_count += 1
avg_num = num_sum/num_count

new = []
for i in numbers:
    txt = str(i)

    if int(txt[-1]) % 2 != 0 and (int(txt[0]) + int(txt[1]) + int(txt[2])) > 16 and (int(txt[0]) * int(txt[1]) * int(txt[2])) % 5 == 0 and i > avg_num:
        new.append(i)

print(new)
```

```python
#5 
""" 
Count customers who: 
• placed exactly two orders 
• total order value is greater than 
average customer total order value 
• average order value is below 
average customer average order value 
""" 

orders = [ 
["Rahul",13000], 
["Rahul",14000], 
["Neha",21000], 
["Neha",18000], 
["Amit",8000], 
["Jay",9000], 
["Jay",10000], 
["Riya",25000], 
["Karan",7000] 
]

cus_value = {}
odr_count = {}

value_sum = 0
cus_count = 0

for row in orders:
    customer = row[0]
    values = row[1]

    if customer in cus_value:
        cus_value[customer] += values
        odr_count[customer] += 1
    else:
        cus_value[customer] = values 
        odr_count[customer] = 1

        cus_count += 1
    value_sum += values
avg_value = value_sum/cus_count

avg_sum = 0
avg_count = 0
for key, value in cus_value.items():
    avg_avg = cus_value[key] / odr_count[key]
    avg_count += 1
    avg_sum += avg_avg
avg = avg_sum/avg_count

count = 0
for key, value in cus_value.items():
    avg_avg = cus_value[key] / odr_count[key]

    if value > avg_value and avg_avg < avg and odr_count[key] == 2:
        count += 1

print(count)
```

```python
#6
""" 
Create a dictionary. 
Key → Department 
Value → Total Salary 

Include only departments: 
• Average salary is above company average salary 
• Employee count is equal to average department employee count 
• Total salary is below average department total salary 
""" 

employees = [ 
["Rahul","IT",73000], 
["Neha","IT",71000], 
["Amit","HR",52000], 
["Riya","HR",55000], 
["Jay","Sales",66000], 
["Karan","Sales",68000], 
["Pooja","Finance",95000], 
["Raj","Finance",61000] 
]

dep_sal = {}
emp_count = {}

sal_sum = 0
emp_sum = 0 
dep_count = 0

for row in employees:
    department = row[1]
    salary = row[2]

    if department in dep_sal:
        dep_sal[department] += salary
        emp_count[department] += 1
    else:
        dep_sal[department] = salary
        emp_count[department] = 1
        dep_count += 1

    sal_sum += salary
    emp_sum += 1

c_avg = sal_sum/emp_sum
d_avg = sal_sum/dep_count
e_avg = emp_sum/dep_count

new = {}
for key, value in dep_sal.items():
    avg = dep_sal[key] / emp_count[key]

    if avg > c_avg and emp_count[key] == e_avg and value < d_avg:
        new[key] = value

print(new)
```

```python
#8
""" 
Count students who scored: 
• above subject average in exactly one subject 
• below or equal subject average in exactly two subjects 
• total marks below class average total marks 
""" 

students = [ 
["Rahul",90,84,91], 
["Neha",98,95,96], 
["Amit",71,69,72], 
["Riya",100,99,98], 
["Jay",86,88,84] 
]

m1_total = 0
m2_total = 0
m3_total = 0

total_marks = 0
student_count = 0

for row in students:
    m1_total += row[1]
    m2_total += row[2]
    m3_total += row[3]

    total_marks += row[1] + row[2] + row[3]
    student_count += 1

m1_avg = m1_total / student_count
m2_avg = m2_total / student_count
m3_avg = m3_total / student_count

class_avg_total = total_marks / student_count


count = 0
for row in students:
    m1 = row[1]
    m2 = row[2]
    m3 = row[3]

    above = 0
    below = 0

    if m1 > m1_avg:
        above += 1
    else:
        below += 1

    if m2 > m2_avg:
        above += 1
    else:
        below += 1

    if m3 > m3_avg:
        above += 1
    else:
        below += 1

    total = m1 + m2 + m3

    if above == 1 and below == 2 and total < class_avg_total:
        count += 1

print(count)
```

```python
#9  
""" 
Find the product whose 
(Price × Quantity ÷ Cost Price) 
is second closest to the average 
of all calculated values. 

Print: 
• Product Name 
• Calculated Value 
• Difference 

Do not use sorted(), max(), min(). 

Format: 
[Price, Cost Price, Quantity] 
""" 

inventory = { 
"Laptop":[87000,73000,5], 
"Mobile":[45000,39000,10], 
"Tablet":[34000,30000,8], 
"TV":[102000,86000,3], 
"Monitor":[54000,48000,9] 
}

pro_value = {}
value_sum = 0
value_count = 0

for key,value in inventory.items():
    pro_value[key] = (value[0] * value[2]) / value[1]
    value_sum += (value[0] * value[2]) / value[1]
    value_count += 1
avg = value_sum/value_count

f_close_value = float("inf")
f_close_pro = None
f_cal = 0
f_dif = 0

s_close_value = float("inf")
s_close_pro = None
s_cal = 0
s_dif = 0

for key,value in pro_value.items():
    if abs(value - avg) < f_close_value:
        s_close_value = f_close_value
        s_close_pro = f_close_pro
        s_cal = f_cal
        s_dif = f_dif

        f_close_value = abs(value - avg)
        f_close_pro = key
        f_cal = value
        f_dif = abs(value - avg)

    elif abs(value - avg) < s_close_value and abs(value - avg) > f_close_value:
        s_close_value = abs(value - avg)
        s_close_pro = key
        s_cal = value
        s_dif = abs(value - avg)


print(f"Product Name : {s_close_pro}")
print(f"Calculated Value : {s_cal}")
print(f"Difference : {s_dif}")
```

```python
#10
""" 
Find the department having the second lowest average salary 
among departments whose employee count 
is greater than or equal to 
the average department employee count. 
 
Do not use sorted(), max(), min(). 
""" 
 
employees = [ 
["Rahul","IT",72000], 
["Neha","IT",70000], 
["Amit","HR",51000], 
["Riya","HR",56000], 
["Jay","Sales",65000], 
["Karan","Sales",67000], 
["Rohan","Sales",66000], 
["Pooja","Finance",94000], 
["Raj","Finance",62000] 
]

dep_sal = {}
emp_count = {}

emp_sum = 0
dep_sum = 0

for row in employees:
    department = row[1]
    salary = row[2]

    if department in dep_sal:
        dep_sal[department] += salary
        emp_count[department] += 1
    else:
        dep_sal[department] = salary
        emp_count[department] = 1

        dep_sum += 1
    emp_sum += 1
avg = emp_sum/dep_sum

f_low_value = float("inf")
f_low_dep = None

s_low_value = float("inf")
s_low_dep = None

for key,value in dep_sal.items():
    d_avg = dep_sal[key] / emp_count[key]

    if d_avg < f_low_value and emp_count[key] >= avg:
        s_low_value = f_low_value
        s_low_dep = f_low_dep

        f_low_value = value
        f_low_dep = key

    elif d_avg < s_low_value and d_avg > f_low_value and emp_count[key] >= avg:
        s_low_value = value
        s_low_dep = key

print(s_low_dep,":",s_low_value)
```
