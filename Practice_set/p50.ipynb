```python
#1
""" 
A company stores yearly performance bonuses. 

Find: 
• Average bonus 
• Employee whose bonus is third closest to the average 
• Difference from the average 

Do not use sorted(), max(), min(). 
""" 

employees = { 
"Rahul":184000, 
"Neha":156000, 
"Amit":102000, 
"Riya":221000, 
"Jay":173000, 
"Pooja":141000, 
"Karan":165000 
}

value_sum = 0
value_count = 0
for value in employees.values():
    value_sum += value
    value_count += 1
avg = value_sum/value_count


f_close_val = float("inf")
f_close_emp = None

s_close_val = float("inf")
s_close_emp = None

t_close_val = float("inf")
t_close_emp = None

for key, value in employees.items():
    if abs(value - avg) < f_close_val:
        t_close_val = s_close_val
        t_close_emp = s_close_emp

        s_close_val = f_close_val
        s_close_emp = f_close_emp

        f_close_val = abs(value - avg)
        f_close_emp = key

    elif abs(value - avg) < s_close_val and abs(value - avg) > f_close_val:
        t_close_val = s_close_val
        t_close_emp = s_close_emp

        s_close_val = abs(value - avg)
        s_close_emp = key

    elif abs(value - avg) < t_close_val and abs(value - avg) > s_close_val:
        t_close_val = abs(value - avg)
        t_close_emp = key

print(f"Average bonus : {avg}")
print(f"Employee whose bonus is third closest to the average : {t_close_emp}")
print(f"Average bonus : {t_close_val}")
```

```python
#2
""" 
Create a new dictionary. 
Key → Product Name 
Value → Profit % 

Profit % = ((Selling Price - Cost Price) / Cost Price) × 100 

Include only products: 
• Profit % is above average profit % 
• Selling Price is above average selling price 
• Quantity is even 
• Stock Value (Selling Price × Quantity) is above average stock value 

Format: 
[Selling Price, Cost Price, Quantity] 
""" 

products = { 
"Laptop":[95000,79000,6], 
"Mobile":[49000,43000,9], 
"Tablet":[38000,32000,8], 
"TV":[115000,94000,4], 
"Monitor":[60000,54000,10], 
"Speaker":[27000,22000,7] 
}

sell_sum = 0
sell_count = 0

per_sum = 0
per_count = 0

stock_sum = 0
stock_count = 0

for value in products.values():
    sell_sum += value[0]
    sell_count += 1

    per_sum += ((value[0]-value[1]) / value[1]) * 100
    per_count += 1

    stock_sum += value[0] * value[2]
    stock_count += 1

avg_sell = sell_sum/sell_count
avg_per = per_sum/per_count
avg_stock = stock_sum/stock_count


new = {}
for key,value in products.items():
    if ((value[0]-value[1]) / value[1]) * 100 > avg_per and value[0] > avg_sell and value[2] % 2 == 0 and (value[0] * value[2]) > avg_stock:

        new[key] = ((value[0]-value[1]) / value[1]) * 100

print(new)
```

```python
#3
""" 
Create a dictionary. 
Key → Student 
Value → Total Marks 

Include only students whose: 
• Highest mark is above average highest mark 
• Lowest mark is below average lowest mark 
• Total marks are above class average total marks 
""" 

marks = { 
"Rahul":[92,85,90], 
"Neha":[99,96,97], 
"Amit":[72,70,74], 
"Riya":[100,99,98], 
"Jay":[87,89,84], 
"Pooja":[95,88,91] 
}

stu_high = {}
stu_low = {}

high_sum = 0
low_sum = 0
both_count = 0
total_mark = {}
total = 0


for key,value in marks.items():
    if value[0] > value[1] and value[0] > value[2]:
        stu_high[key] = value[0]
        high_sum += value[0]
    elif value[0] < value[1] and value[0] < value[2]:
        stu_low[key] = value[0]
        low_sum += value[0]

    if value[1] > value[0] and value[1] > value[2]:
        stu_high[key] = value[1]
        high_sum += value[1]
    elif value[1] < value[0] and value[1] < value[2]:
        stu_low[key] = value[1]
        low_sum += value[1]


    if value[2] > value[0] and value[2] > value[1]:
        stu_high[key] = value[2]
        high_sum += value[2]
    elif value[2] < value[0] and value[2] < value[1]:
        stu_low[key] = value[2]
        low_sum += value[2]

    total_mark[key] = value[0] + value[1] + value[2]
    total += value[0] + value[1] + value[2]
    both_count += 1

avg_high = high_sum/both_count
avg_low = low_sum/both_count 
avg_total = total/both_count

new = {}
for key,value in stu_high.items():
    if value > avg_high and stu_low[key] < avg_low and total_mark[key] > avg_total:
        new[key] = total_mark[key]

print(new)
```

```python
#4
""" 
Create a new list containing only numbers: 
• First digit is even 
• Last digit is odd 
• Sum of digits is divisible by 4 
• Product of digits is greater than 100 
• Number is below the average of the list 
""" 

numbers = [861,472,985,643,289,847,561]

num_sum = 0
num_count = 0 
for i in numbers:
    num_sum += i
    num_count += 1
avg = num_sum/num_count

new = []
for i in numbers:
    txt = str(i)
    
    if int(txt[0]) % 2 == 0 and int(txt[-1]) % 2 != 0 and (int(txt[0]) + int(txt[1]) + int(txt[2])) % 4 == 0 and (int(txt[0]) * int(txt[1]) * int(txt[2])) > 100 and i < avg:
        new.append(i)

print(new)
```

```python
#5
""" 
Count customers who: 
• placed exactly two orders 
• total order value is above average customer total order value 
• average order value is below average customer average order value 
• highest order value is above average highest order value 
""" 

orders = [ 
["Rahul",14000], 
["Rahul",15000], 
["Neha",23000], 
["Neha",18000], 
["Amit",9000], 
["Jay",10000], 
["Jay",11000], 
["Riya",26000], 
["Karan",8000] 
]

cus_values = {}
for row in orders:
    customer = row[0]
    values = row[1]

    if customer in cus_values:
        cus_values[customer] += values
    else:
        cus_values[customer] = values

cus_high = {}

for row in orders:
    customer = row[0]
    value = row[1]

    if customer in cus_high:
        if value > cus_high[customer]:
            cus_high[customer] = value
    else:
        cus_high[customer] = value

print(cus_high)
```

```python
#6 
""" 
Create a dictionary. 
Key → Department 
Value → Average Salary 

Include only departments: 
• Employee count is equal to average department employee count 
• Average salary is above company average salary 
• Total salary is below average department total salary 
• Highest salary is above company average salary 
""" 
employees = [ 
["Rahul","IT",74000], 
["Neha","IT",72000], 
["Amit","HR",53000], 
["Riya","HR",57000], 
["Jay","Sales",68000], 
["Karan","Sales",69000], 
["Pooja","Finance",96000], 
["Raj","Finance",63000] 
]

dep_high = {}
dep_sal = {}
dep_count = {}

sal_sum = 0
emp_count = 0

dep_sum = 0

for row in employees:
    department = row[1]
    salary = row[2]

    if department in dep_high:
        if salary > dep_high[department]:
            dep_high[department] = salary
        dep_sal[department] += salary
        dep_count[department] += 1
    else:
        dep_high[department] = salary
        dep_sal[department] = salary
        dep_count[department] = 1

        dep_sum += 1
    sal_sum += salary
    emp_count += 1

avg_emp = sal_sum/emp_count
avg_dep = sal_sum/dep_sum
avg_e = emp_count/dep_sum

new = {}
for key,value in dep_sal.items():
    avg = dep_sal[key] / dep_count[key]

    if dep_count[key] == avg_e and avg > avg_emp and value < avg_dep and dep_high[key] > avg_emp:
        new[key] = value

print(new)
```

```python
#7
""" 
Find the number whose 
(cube ÷ first digit) 
is third farthest from the average 
of all calculated values. 

Print: 
• Number 
• Calculated Value 
• Difference 

Do not use sorted(), max(), min(). 
""" 

numbers = [27,46,58,73,84,39,65]

val_sum = 0
val_count = 0
for i in numbers:
    txt = str(i)

    val_sum += ((i**3) / int(txt[0])) 
    val_count += 1
avg = val_sum/val_count

f_far_value = 0
f_far_num = 0
f_val = 0

s_far_value = 0
s_far_num = 0
s_val = 0

t_far_value = 0
t_far_num = 0
t_val = 0

for i in numbers:
    txt = str(i)

    if abs((i**3) / int(txt[0]) - avg) > f_far_value:
        t_far_value = s_far_value
        t_far_num = s_far_num
        t_val = s_val

        s_far_value = f_far_value
        s_far_num = f_far_num
        s_val = f_val

        f_far_value = abs((i**3) / int(txt[0]) - avg)
        f_far_num = i
        f_val = (i**3) / int(txt[0])

    elif abs((i**3) / int(txt[0]) - avg) > s_far_value and abs((i**3) / int(txt[0]) - avg) < f_far_value:
        t_far_value = s_far_value
        t_far_num = s_far_num
        t_val = s_val

        s_far_value = abs((i**3) / int(txt[0]) - avg)
        s_far_num = i
        s_val = (i**3) / int(txt[0])

    elif abs((i**3) / int(txt[0]) - avg) > t_far_value and abs((i**3) / int(txt[0]) - avg) < s_far_value:
        t_far_value = abs((i**3) / int(txt[0]) - avg)
        t_far_num = i
        t_val = (i**3)/int(txt[0])

print(f"Number : {t_far_num}")
print(f"Calculated Value : {t_val}")
print(f"Difference : {t_far_value}")
```

```python
#8 
""" 
Count students who scored: 
• above subject average in exactly two subjects 
• below or equal subject average in exactly one subject 
• total marks above class average total marks 
• highest mark above average highest mark 
""" 

students = [ 
["Rahul",91,86,92], 
["Neha",99,96,98], 
["Amit",73,71,75], 
["Riya",100,99,98], 
["Jay",88,90,85], 
["Pooja",95,89,93] 
]

stu_total = {}
stu_high = {}

m1_sum = 0
m2_sum = 0
m3_sum = 0

high_sum = 0
total_sum = 0
count = 0

# First pass
for row in students:
    student = row[0]
    m1 = row[1]
    m2 = row[2]
    m3 = row[3]

    m1_sum += m1
    m2_sum += m2
    m3_sum += m3

    total = m1 + m2 + m3
    stu_total[student] = total
    total_sum += total

    if m1 > m2 and m1 > m3:
        stu_high[student] = m1
        high_sum += m1
    elif m2 > m1 and m2 > m3:
        stu_high[student] = m2
        high_sum += m2
    else:
        stu_high[student] = m3
        high_sum += m3

    count += 1

m1_avg = m1_sum / count
m2_avg = m2_sum / count
m3_avg = m3_sum / count

avg_total = total_sum / count
avg_high = high_sum / count

count = 0
for row in students:
    student = row[0]
    m1 = row[1]
    m2 = row[2]
    m3 = row[3]

    above = 0
    below = 0

    if m1 > m1_avg:
        above += 1
    else:
        below += 1

    if m2 > m2_avg:
        above += 1
    else:
        below += 1

    if m3 > m3_avg:
        above += 1
    else:
        below += 1

    if (above == 2 and below == 1 and stu_total[student] > avg_total and stu_high[student] > avg_high):
        count += 1

print(count)
```

```python
#9
""" 
Find the product whose 
((Selling Price × Quantity) ÷ Cost Price) 
is third closest to the average 
of all calculated values. 

Print: 
• Product Name 
• Calculated Value 
• Difference 

Do not use sorted(), max(), min(). 

Format: 
[Selling Price, Cost Price, Quantity] 
""" 

inventory = { 
"Laptop":[89000,74000,6], 
"Mobile":[47000,41000,9], 
"Tablet":[36000,31000,8], 
"TV":[104000,87000,4], 
"Monitor":[56000,50000,10], 
"Speaker":[26000,22000,7] 
}

pro_cal = {}
cal_sum = 0
cal_count = 0

for key,value in inventory.items():
    pro_cal[key] = (value[0] * value[2]) / value[1]
    cal_sum += (value[0] * value[2]) / value[1]
    cal_count += 1
avg = cal_sum/cal_count

f_close_val = float("inf")
f_close_pro = None
f_cal = 0

s_close_val = float("inf")
s_close_pro = None
s_cal = 0

t_close_val = float("inf")
t_close_pro = None
t_cal = 0

for key,value in pro_cal.items():
    if abs(value - avg) < f_close_val:
        t_close_val = s_close_val
        t_close_pro = s_close_pro
        t_cal = s_cal

        s_close_val = f_close_val
        s_close_pro = f_close_pro
        s_cal = f_cal

        f_close_val = abs(value - avg)
        f_close_pro = key
        f_cal = value

    elif abs(value - avg) < s_close_val and abs(value - avg) > f_close_val:
        t_close_val = s_close_val
        t_close_pro = s_close_pro
        t_cal = s_cal
        
        s_close_val = abs(value - avg)
        s_close_pro = key
        s_cal = value

    elif abs(value - avg) < t_close_val and abs(value - avg) > s_close_val:
        t_close_val = abs(value - avg)
        t_close_pro = key
        t_cal = value

print(f"Product Name : {t_close_pro}")
print(f"Calculated Value : {t_cal}")
print(f"Difference : {t_close_val}")
```
