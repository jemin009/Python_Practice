```python
#1
""" 
A company stores yearly performance bonuses. 

Find: 
• Average bonus 
• Employee whose second farthest bonus below the average 
• Difference from the average 

Do not use sorted(), max(), min(). 
""" 

employees = { 
"Rahul":188000, 
"Neha":162000, 
"Amit":104000, 
"Riya":226000, 
"Jay":171000, 
"Pooja":145000, 
"Karan":158000 
}

bonus_sum = 0
bonus_count = 0
for value in employees.values():
    bonus_sum += value
    bonus_count += 1
avg = bonus_sum/bonus_count

f_far_value = 0
f_far_emp = None

s_far_value = 0
s_far_emp = None

for key,value in employees.items():
    if value < avg and abs(value-avg) > f_far_value:
        s_far_value = f_far_value
        s_far_emp = f_far_emp

        f_far_value = abs(value-avg)
        f_far_emp = key

    if value < avg and abs(value-avg) > s_far_value and abs(value-avg) < f_far_value:
        s_far_value = abs(value-avg)
        s_far_emp = key

print(f"Average bonus : {avg}")
print(f"Employee whose second farthest bonus below the average : {s_far_emp}")
print(f"Difference from the average : {s_far_value}")
```

```python
#2
""" 
Create a new dictionary. 
Key → Product Name 
Value → Stock Value 

Stock Value = Selling Price × Quantity 
 
Include only products: 
• Stock value is above average stock value 
• Profit % is above average profit % 
• Quantity is odd 
• Selling Price is below average selling price 
• Cost Price is above average cost price 
 
Profit % = ((Selling Price - Cost Price) / Cost Price) × 100 
 
Format: 
[Selling Price, Cost Price, Quantity] 
""" 
 
products = { 
"Laptop":[98000,81000,5], 
"Mobile":[52000,45000,8], 
"Tablet":[39000,33000,7], 
"TV":[118000,96000,3], 
"Monitor":[62000,56000,9], 
"Speaker":[29000,24000,11] 
}

sell_sum = 0
sell_count = 0

cost_sum = 0
cost_count = 0

s_value_sum = 0
s_value_count = 0

per_sum = 0
per_count = 0

for key,value in products.items():
    sell_sum += value[0] 
    sell_count += 1

    cost_sum += value[1]
    cost_count += 1

    s_value_sum += value[0] * value[2]
    s_value_count += 1

    per_sum += ((value[0] - value[1]) / value[1]) * 100
    per_count += 1

avg_sell = sell_sum/sell_count
avg_cost = cost_sum/cost_count
avg_s_value = s_value_sum/s_value_count
avg_per = per_sum/per_count

new = {}
for key,value in products.items():
    if value[0] * value[2] > avg_s_value and ((value[0] - value[1]) / value[1]) * 100 > avg_per and value[2] % 2 != 0 and value[0] < avg_sell and value[1] > avg_cost:
        new[key] = value[0] * value[2]

print(new)
```

```python
#3
""" 
Create a dictionary. 
Key → Student 
Value → Average Marks 

Include only students whose: 
• Highest mark is above average highest mark 
• Lowest mark is below average lowest mark 
• Total marks are above class average total marks 
• Average marks are above class average marks 
""" 

marks = { 
"Rahul":[93,87,91], 
"Neha":[100,97,98], 
"Amit":[74,72,76], 
"Riya":[99,100,98], 
"Jay":[89,91,86], 
"Pooja":[96,90,94] 
}

stu_total = {}
stu_high = {}
stu_low = {}

high_sum = 0
low_sum = 0

total_sum = 0

m1_total = 0
m2_total = 0
m3_total = 0
total_count = 0

for key,value in marks.items():
    m1_total += value[0]
    m2_total += value[1]
    m3_total += value[2]
    total_count += 1

    stu_total[key] = value[0] + value[1] + value[2]
    total_sum += value[0] + value[1] + value[2]

    if value[0] > value[1] and value[0] > value[2]:
        stu_high[key] = value[0]
        high_sum += value[0]
    elif value[0] < value[1] and value[0] < value[2]:
        stu_low[key] = value[0]
        low_sum += value[0]

    if value[1] > value[0] and value[1] > value[2]:
        stu_high[key] = value[1]
        high_sum += value[1]
    elif value[1] < value[0] and value[1] < value[2]:
        stu_low[key] = value[1]
        low_sum += value[1]

    if value[2] > value[1] and value[2] > value[0]:
        stu_high[key] = value[2]
        high_sum += value[2]
    elif value[2] < value[1] and value[2] < value[0]:
        stu_low[key] = value[2]
        low_sum += value[2]

m1_avg = m1_total/total_count
m2_avg = m2_total/total_count
m3_avg = m3_total/total_count

avg_total = total_sum/total_count
avg_high = high_sum/total_count
avg_low = low_sum/total_sum

new = {}

for key in marks:
    above = 0

    if marks[key][0] > m1_avg:
        above += 1

    if marks[key][1] > m2_avg:
        above += 1

    if marks[key][2] > m3_avg:
        above += 1

    if (
        stu_high[key] > avg_high
        and stu_low[key] < avg_low
        and stu_total[key] > avg_total
        and (stu_total[key] / 3) > ((m1_avg + m2_avg + m3_avg) / 3)
    ):
        new[key] = stu_total[key] / 3

print(new)
```

```python
#4
""" 
Create a new list containing only numbers: 
• First digit is odd 
• Last digit is even 
• Sum of digits is divisible by 5 
• Product of digits is greater than 120 
• Number is above the average of the list 
""" 

numbers = [954,681,572,846,395,784,963]

num_sum = 0
num_count = 0
for i in numbers:
    num_sum  += i
    num_count += 1
avg = num_sum/num_count

new = []
for i in numbers:
    txt = str(i)

    if int(txt[0]) % 2 != 0 and int(txt[-1]) % 2 == 0 and (int(txt[0]) + int(txt[1]) + int(txt[2])) % 5 == 0 and (int(txt[0]) * int(txt[1]) * int(txt[2])) > 120 and i > avg:
        new.append(i)

print(new)
```

```python
#5
""" 
Count customers who: 
• placed exactly two orders 
• total order value is below average customer total order value 
• average order value is above average customer average order value 
• highest order value is below average highest order value 
""" 

orders = [ 
["Rahul",17000], 
["Rahul",12000], 
["Neha",26000], 
["Neha",15000], 
["Amit",10000], 
["Jay",9000], 
["Jay",14000], 
["Riya",28000], 
["Karan",7000] 
]

emp_high = {}
emp_count = {}
emp_value = {}

value_sum = 0
value_count = 0
emp_sum = 0

for row in orders:
    customer = row[0]
    values = row[1]

    if customer in emp_high:
        if values > emp_high[customer]:
            emp_high[customer] = values
        emp_count[customer] += 1
        emp_value[customer] += values

    else:
        emp_high[customer] = values
        emp_count[customer] = 1
        emp_value[customer] = values

        value_count += 1
    value_sum += values
    emp_sum += 1

avg_val = value_sum/value_count
avg_emp = value_sum/emp_sum

high_sum = 0
high_count = 0

for value in emp_high.values():
    high_sum += value
    high_count += 1

avg_high = high_sum / high_count

count = 0

for key, value in emp_value.items():
    avg = emp_value[key] / emp_count[key]

    if (
        emp_count[key] == 2
        and emp_value[key] < avg_val
        and avg > avg_emp
        and emp_high[key] < avg_high
    ):
        count += 1

print(count)
```

```python
#6
""" 
Create a dictionary. 
Key → Department 
Value → Highest Salary 

Include only departments: 
• Employee count is equal to average department employee count 
• Average salary is above company average salary 
• Total salary is above average department total salary 
• Highest salary is above company average salary 
""" 

employees = [ 
["Rahul","IT",76000], 
["Neha","IT",74000], 
["Amit","HR",55000], 
["Riya","HR",59000], 
["Jay","Sales",70000], 
["Karan","Sales",71000], 
["Pooja","Finance",98000], 
["Raj","Finance",65000] 
]

dep_high = {}
dep_emp = {}
dep_salary = {}

sal_sum = 0
emp_count = 0
dep_count = 0

for row in employees:
    department = row[1]
    salary = row[2]

    if department in dep_high:
        if salary > dep_high[department]:
            dep_high[department] = salary
        dep_emp[department] += 1
        dep_salary[department] += salary
    else: 
        dep_high[department] = salary
        dep_emp[department] = 1
        dep_salary[department] = salary

        dep_count += 1
    sal_sum += salary
    emp_count += 1

avg_c = sal_sum/emp_count
avg_d = sal_sum/dep_count
avg_emp = emp_count/dep_count

new = {}
for key,value in dep_salary.items():
    avg = dep_salary[key] / dep_emp[key]

    if dep_emp[key] == avg_emp and avg > avg_c and value > avg_d and dep_high[key] > avg_c:
        new[key] = dep_high[key]

print(new)
```

```python
#8
""" 
Count students who scored: 
• above subject average in exactly one subject 
• below or equal subject average in exactly two subjects 
• total marks below class average total marks 
• highest mark below average highest mark 
""" 

students = [ 
["Rahul",92,85,90], 
["Neha",99,97,98], 
["Amit",75,73,71], 
["Riya",100,99,98], 
["Jay",88,90,86], 
["Pooja",95,91,93] 
]

stu_total = {}
stu_high = {}

high_sum = 0
mark_sum = 0
m1_total = 0
m2_total = 0
m3_total = 0
total_count = 0

for row in students:
    student = row[0]
    m1 = row[1]
    m2 = row[2]
    m3 = row[3]

    m1_total += m1
    m2_total += m2
    m3_total += m3
    total_count += 1

    stu_total[student] = m1+m2+m3
    mark_sum += stu_total[student]

    if m1>m2 and m1>m3:
        stu_high[student] = m1 
        high_sum += m1
    elif m2>m1 and m2>m3:
        stu_high[student] = m2
        high_sum += m2
    elif m3>m1 and m3>m2:
        stu_high[student] = m3
        high_sum += m3

m1_avg = m1_total/total_count
m2_avg = m2_total/total_count
m3_avg = m3_total/total_count
avg_mark = mark_sum/total_count
avg_high = high_sum/total_count

count = 0
for row in students:
    student = row[0]
    m1 = row[1]
    m2 = row[2]
    m3 = row[3]

    above = 0
    below = 0

    if m1 > m1_avg:
        above += 1
    elif m1 <= m1_avg:   
        below += 1

    if m2 > m2_avg:
        above += 1
    elif m2 <= m2_avg:   
        below += 1

    if m3 > m3_avg:
        above += 1
    elif m3 <= m3_avg:   
        below += 1

    if above == 1 and below == 2 and stu_total[student] < avg_mark and stu_high[student] < avg_high:
        count += 1

print(count)
```

```python
#9
""" 
Find the product whose 
(Price × Quantity ÷ Cost Price) 
is third closest to the average 
of all calculated values. 

Print: 
• Product Name 
• Calculated Value 
• Difference 

Do not use sorted(), max(), min(). 

Format: 
[Price, Cost Price, Quantity] 
""" 

inventory = { 
"Laptop":[89000,74000,6], 
"Mobile":[47000,41000,9], 
"Tablet":[36000,31000,7], 
"TV":[105000,89000,4], 
"Monitor":[56000,50000,8] 
}

pro_cal = {}

cal_sum = 0
cal_count = 0

for key,value in inventory.items():
    pro_cal[key] = (value[0] * value[2]) / value[1]

    cal_sum += (value[0] * value[2]) / value[1]
    cal_count += 1
avg_cal = cal_sum/cal_count

f_close_val = float("inf")
f_close_pro = None
f_cal = 0

s_close_val = float("inf")
s_close_pro = None
s_cal = 0

t_close_val = float("inf")
t_close_pro = None
t_cal = 0

for key,value in pro_cal.items():
    if abs(value - avg_cal) < f_close_val:
        t_close_val = s_close_val
        t_close_pro = s_close_pro
        t_cal = s_cal

        s_close_val = f_close_val
        s_close_pro = f_close_pro
        s_cal = f_cal

        f_close_val = abs(value - avg_cal)
        f_close_pro = key
        f_cal = value

    elif abs(value - avg_cal) < s_close_val and abs(value - avg_cal) > f_close_val:
        t_close_val = s_close_val
        t_close_pro = s_close_pro
        t_cal = s_cal

        s_close_val = abs(value - avg_cal)
        s_close_pro = key
        s_cal = value

    elif abs(value - avg_cal) < t_close_val and abs(value - avg_cal) > s_close_val:
        t_close_val = abs(value - avg_cal)
        t_close_pro = key
        t_cal = value

print(f"Product Name : {t_close_pro}")
print(f"Calculated Value : {t_cal}")
print(f"Difference : {t_close_val}")
```
