```python
#1
""" 
A company stores yearly bonuses. 
Find: 
• Employee with the third highest bonus. 
• Print employee name and bonus. 

Do not use sorted(), max(), min(). 
""" 

employees = { 
"Rahul":191000, 
"Neha":163000, 
"Amit":108000, 
"Riya":228000, 
"Jay":176000, 
"Pooja":149000, 
"Karan":158000 
}

f_high_bonus = 0
f_high_emp = None

s_high_bonus = 0
s_high_emp = None

t_high_bonus = 0
t_high_emp = None

for key,value in employees.items():
    if value > f_high_bonus:
        t_high_bonus = s_high_bonus
        t_high_emp = s_high_emp

        s_high_bonus = f_high_bonus
        s_high_emp = f_high_emp

        f_high_bonus = value
        f_high_emp = key

    elif value > s_high_bonus and value < f_high_bonus:
        t_high_bonus = s_high_bonus
        t_high_emp = s_high_emp

        s_high_bonus = value
        s_high_emp = key

    elif value > t_high_bonus and value < s_high_bonus:
        t_high_bonus = value
        t_high_emp = key

print(f"Employee with the third highest bonus : {t_high_emp}:{t_high_bonus}")
```

```python
#2
""" 
Create a new dictionary. 
Key   → Product Name 
Value → Profit Amount 

Profit Amount = 
(Selling Price - Cost Price) × Quantity

Include only products whose 
profit amount is positive. 

Format: 
[Selling Price, Cost Price, Quantity] 
""" 

products = { 
"Laptop":[98000,81000,6], 
"Mobile":[52000,45000,8], 
"Tablet":[39000,33000,7], 
"TV":[118000,96000,3], 
"Monitor":[62000,56000,9], 
"Speaker":[29000,24000,11] 
}

pro_amt = {}
for key,value in products.items():
    if (value[0] - value[1]) * value[2] > 0:
        pro_amt[key] = (value[0] - value[1]) * value[2]

print(pro_amt)
```

```python
#3
""" 
Create a dictionary. 
Key   → Student 
Value → Highest Mark 

Only include students whose 
highest mark is greater than 95. 
""" 

marks = { 
"Rahul":[93,87,91], 
"Neha":[100,97,98], 
"Amit":[74,72,76], 
"Riya":[99,100,98], 
"Jay":[89,91,86], 
"Pooja":[96,90,94] 
}

stu_high = {}
for key,value in marks.items():
    if value[0] > value[1] and value[0] > value[2] and value[0] > 95:
        stu_high[key] = value[0]

    elif value[1] > value[0] and value[1] > value[2] and value[1] > 95:
        stu_high[key] = value[1]

    elif value[2] > value[0] and value[2] > value[1] and value[2] > 95:
        stu_high[key] = value[2]

print(stu_high)
```

```python
#4
""" 
Create a new list containing numbers whose: 
• first digit is greater than last digit 
• sum of digits is even 
""" 

numbers = [954,681,572,846,395,784,963]

new = []
for i in numbers:
    txt = str(i)

    if int(txt[0]) > int(txt[-1]) and (int(txt[0]) + int(txt[1]) + int(txt[2])) % 2 == 0:
        new.append(i)

print(new)
```

```python
#5
""" 
Count customers who placed 
exactly two orders. 
""" 

orders = [ 
["Rahul",17000], 
["Rahul",12000], 
["Neha",26000], 
["Neha",15000], 
["Amit",10000], 
["Jay",9000], 
["Jay",14000], 
["Riya",28000], 
["Karan",7000] 
]

cus_odr = {}
for row in orders:
    customer = row[0]

    if customer in cus_odr:
        cus_odr[customer] += 1
    else:
        cus_odr[customer] = 1

count = 0
for key,value in cus_odr.items():
    if value == 2:
        count += 1

print(count)
```

```python
#6
""" 
Create a dictionary. 
Key   → Department 
Value → Employee Count 

Include only departments 
having exactly two employees. 
""" 

employees = [ 
["Rahul","IT",76000], 
["Neha","IT",74000], 
["Amit","HR",55000], 
["Riya","HR",59000], 
["Jay","Sales",70000], 
["Karan","Sales",71000], 
["Pooja","Finance",98000], 
["Raj","Finance",65000] 
]

dep_emp = {}
for row in employees:
    department = row[1]

    if department in dep_emp:
        dep_emp[department] += 1
    else:
        dep_emp[department] = 1

new = {}
for key,value in dep_emp.items():
    if value == 2:
        new[key] = value

print(new)
```

```python
#7
""" 
Find the number whose 
(square ÷ last digit) 
is second highest. 

Print: 
• Number 
• Calculated Value 

Do not use sorted(), max(), min(). 
""" 

numbers = [28,47,59,74,83,96,65]

f_high_cal = 0
f_high_num = 0

s_high_cal = 0
s_high_num = 0

for i in numbers:
    txt = str(i)

    if (i**2) / int(txt[-1]) > f_high_cal:
        s_high_cal = f_high_cal
        s_high_num = f_high_num

        f_high_cal = (i**2) / int(txt[-1])
        f_high_num = i

    elif (i**2) / int(txt[-1]) > s_high_cal and (i**2) / int(txt[-1]) < f_high_cal:
        s_high_cal = (i**2) / int(txt[-1])
        s_high_num = i

print(f"Number : {s_high_num}")
print(f"Calculated Value : {s_high_cal}")
```

```python
#8
""" 
Find the student having 
the highest total marks. 

Print: 
• Student 
• Total Marks 

Do not use sorted(), max(), min(). 
""" 

students = [ 
["Rahul",92,85,90], 
["Neha",99,97,98], 
["Amit",75,73,71], 
["Riya",100,99,98], 
["Jay",88,90,86], 
["Pooja",95,91,93] 
] 

high_mark = 0
high_stu = None

for data in students:
    if data[1] + data[2] + data[3] > high_mark:
        high_mark = data[1] + data[2] + data[3]
        high_stu = data[0]

print(f"Student : {high_stu}")
print(f"Total Marks : {high_mark}")
```

```python
#9
""" 
Find the product having 
the second highest stock value. 

Stock Value = Price × Quantity 

Print: 
• Product Name 
• Stock Value 

Do not use sorted(), max(), min(). 

Format: 
[Price, Cost Price, Quantity] 
""" 

inventory = { 
"Laptop":[89000,74000,6], 
"Mobile":[47000,41000,9], 
"Tablet":[36000,31000,7], 
"TV":[105000,89000,4], 
"Monitor":[56000,50000,8] 
}

f_high_stock = 0
f_high_pro = None

s_high_stock = 0
s_high_pro = None

for key,value in inventory.items():
    if value[0] * value[2] > f_high_stock:
        s_high_stock = f_high_stock
        s_high_pro = f_high_pro

        f_high_stock = value[0] * value[2]
        f_high_pro = key

    elif value[0] * value[2] > s_high_stock and value[0] * value[2] < f_high_stock:
        s_high_stock = value[0] * value[2]
        s_high_pro = key

print(f"Product Name : {s_high_pro}")
print(f"Stock Value : {s_high_stock}")
```
