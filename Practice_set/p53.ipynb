```python
#1
""" 
A company stores yearly bonuses. 

Find: 
• Employee with the second lowest bonus. 

Print: 
• Employee Name 
• Bonus 

Do not use sorted(), max(), min(). 
""" 

employees = { 
"Rahul":194000, 
"Neha":166000, 
"Amit":111000, 
"Riya":231000, 
"Jay":178000, 
"Pooja":151000, 
"Karan":160000 
}

f_low_bonus = float("inf")
f_low_emp = None

s_low_bonus = float("inf")
s_low_emp = None

for key,value in employees.items():
    if value < f_low_bonus:
        s_low_bonus = f_low_bonus
        s_low_emp = f_low_emp

        f_low_bonus = value
        f_low_emp = key

    elif value < s_low_bonus and value > f_low_bonus:
        s_low_bonus = value
        s_low_emp = key

print(s_low_emp,":",s_low_bonus)
    
```

```python
#2
""" 
Create a dictionary. 
Key   → Product Name 
Value → Stock Value 

Stock Value = Selling Price × Quantity 

Include only products 
whose stock value is greater than 400000. 

Format: 
[Selling Price, Cost Price, Quantity] 
""" 

products = { 
"Laptop":[99000,82000,5], 
"Mobile":[53000,46000,8], 
"Tablet":[41000,35000,7], 
"TV":[121000,98000,4], 
"Monitor":[64000,57000,9], 
"Speaker":[31000,26000,11] 
}

new = {}
for key,value in products.items():
    if value[0] * value[2] > 400000:
        new[key] = value[0] * value[2]

print(new)
```

```python
#4
""" 
Create a new list containing numbers whose: 
• first digit is less than last digit 
• product of digits is divisible by 3 
""" 

numbers = [382,761,594,428,635,873,249]

new = []
for i in numbers:
    txt = str(i)

    if int(txt[0]) < int(txt[-1]) and (int(txt[0]) * int(txt[1]) * int(txt[2])) % 3 == 0:
        new.append(i) 

print(new)
```

```python
#5
""" 
Count customers 
who placed exactly one order. 
""" 

orders = [ 
["Rahul",18000], 
["Rahul",12000], 
["Neha",24000], 
["Jay",13000], 
["Amit",9000], 
["Riya",27000], 
["Karan",8000] 
]

cus_order = {}
for row in orders:
    if row[0] in cus_order:
        cus_order[row[0]] += 1
    else:
        cus_order[row[0]] = 1

count = 0
for key,value in cus_order.items():
    if value == 1:
        count += 1

print(count)
```

```python
#6
""" 
Create a dictionary. 
Key   → Department 
Value → Total Salary 

Include only departments 
whose employee count is 2. 
""" 

employees = [ 
["Rahul","IT",77000], 
["Neha","IT",75000], 
["Amit","HR",56000], 
["Riya","HR",60000], 
["Jay","Sales",71000], 
["Karan","Sales",72000], 
["Pooja","Finance",99000], 
["Raj","Finance",67000] 
]

emp_count = {}
dep_sal = {}
for row in employees:
    if row[1] in emp_count:
        emp_count[row[1]] += 1
        dep_sal[row[1]] += row[2]
    else:
        emp_count[row[1]] = 1
        dep_sal[row[1]] = row[2]

new = {}
for key,value in emp_count.items():
    if value == 2:
        new[key] = dep_sal[key]

print(new)
```

```python
#9
""" 
Find the product 
having the highest profit amount. 
 
Profit Amount = 
(Selling Price − Cost Price) × Quan ty 
 
Print: 
• Product Name 
• Profit Amount 
 
Do not use sorted(), max(), min(). 
 
Format: 
[Price, Cost Price, Quantity] 
""" 
 
inventory = { 
"Laptop":[91000,76000,6], 
"Mobile":[49000,42000,9], 
"Tablet":[37000,32000,7], 
"TV":[108000,91000,4], 
"Monitor":[58000,51000,8] 
}

f_high_profit = 0
f_high_product = None

for key,value in inventory.items():
    if (value[0]-value[1]) * value[2] > f_high_profit:
        f_high_profit = (value[0]-value[1]) * value[2]
        f_high_product = key

print(f"Product Name : {f_high_product}")
print(f"Profit Amount : {f_high_profit}")
    
```

```python
#10
""" 
Find the department 
having the highest average salary. 
 
Do not use sorted(), max(), min(). 
""" 
 
employees = [ 
["Rahul","IT",74000], 
["Neha","IT",76000], 
["Amit","HR",55000], 
["Riya","HR",59000], 
["Jay","Sales",70000], 
["Karan","Sales",72000], 
["Rohan","Sales",71000], 
["Pooja","Finance",98000], 
["Raj","Finance",66000] 
]

dep_sal = {}
dep_emp = {}
for row in employees:
    if row[1] in dep_sal:
        dep_sal[row[1]] += row[2]
        dep_emp[row[1]] += 1
    else:
        dep_sal[row[1]] = row[2]
        dep_emp[row[1]] = 1

high_avg = 0
high_dep = None
for key,value in dep_sal.items():
    avg = dep_sal[key] / dep_emp[key]

    if avg > high_avg:
        high_avg = avg
        high_dep = key

print(high_dep,":",high_avg)
```
