```python
#1
""" 
Take two numbers as input. 

Print: 
• Sum 
• Difference 
• Product 
• Division 
• Remainder 
""" 

num1 = int(input("Enter a first number :"))
num2 = int(input("Enter a second number :"))

print(f"Sum {num1+num2}")
print(f"Difference {abs(num1-num2)}")
print(f"Product {num1*num2}")
print(f"Division {num1/num2}")
print(f"Remainder {num1%num2}")
```

```python
#2
""" 
Take a string as input. 

Print: 
• Length 
• First character 
• Last character 
• Second character 
• Second-last character 
"""

string = input("Enter a string ")

print(f"Length : {len(string)}")
print(f"First character : {string[0]}")
print(f"Last character : {string[-1]}")
print(f"Second character : {string[1]}")
print(f"Second-last character : {string[-2]}")
```

```python
#3
""" 
Take a word as input. 
 
Print: 
• First 3 characters 
• Last 3 characters 
• Characters with step 2 
• Reverse of the word 
"""

word = input("Enter a word :")

print(f"First 3 characters : {word[0:3]}")
print(f"Last 3 characters : {word[-3:]}")
print(f"Characters with step 2 : {word[::2]}")
print(f"Reverse of the word : {word[::-1]}")
```

```python
#4
"""
Take a sentence as input. 
 
Print: 
• Uppercase 
• Lowercase 
• Capitalized version 
• Number of characters 
"""

sentence = input("Enter a sentence :")

print(f"Uppercase : {sentence.upper()}")
print(f"Lowercase : {sentence.lower()}")
print(f"Capitalized version : {sentence.capitalize()}")
print(f"Number of characters : {len(sentence)}")
```

```python
#5
""" 
Create a list of 6 numbers. 
 
Print: 
• First element 
• Third element 
• Last element 
• First four elements 
• Reverse of the list 
"""

list1 = [23,56,12,22,67,34]

print(f"First element : {list1[0]}")
print(f"Third element : {list1[2]}")
print(f"Last element : {list1[-1]}")
print(f"First four elements : {list1[:4]}")
print(f"Reverse of the list : {list1[::-1]}")
```

```python
#7
""" 
Create two sets. 
 
Print: 
• Union 
• Intersection 
• Difference of first set 
• Difference of second set 
"""

num1 = {10,20,30,40,50}
num2 = {40,50,60,70,80}

print(f"Union : {num1.union(num2)}")
print(f"Intersection : {num1.intersection(num2)}")
print(f"Difference of first set : {num1-num2}")
print(f"Difference of second set : {num2-num1}")
```

```python
#8
""" 
Create a dictionary containing 5 students 
and their marks. 
 
Print: 
• Complete dictionary 
• All student names 
• All marks 
• Marks of one student 
"""

marks = {
    "Jemin" : 45,
    "Krish" : 49,
    "Dhruvin" : 55,
    "Sujal" : 69,
    "Denish" : 78
}

print(f"Complete dictionary : {marks}")
print(f"All student names : {marks.keys()}")
print(f"All marks : {marks.values()}")
print(f"Marks of one student : {marks['Jemin']}")
```

```python
#9
""" 
Take a decimal number as input. 
 
Convert it into: 
• int 
• float 
• Decimal 
• Fraction 
 
Print each value along with its type. 
"""

from decimal import Decimal
from fractions import Fraction

num = input("Enter a Decimal number :")

print(f"Int : {int(num)}, Type : {type(int(num))}")
print(f"Float : {float(num)}, Type : {type(float(num))}")
print(f"Decimal : {Decimal(num)}, Type : {type(Decimal(num))}")
print(f"Fraction : {Fraction(num)}, Type : {type(Fraction(num))}")
```

```python
#10
""" 
Use the math and random modules. 
 
Print: 
• Square root of 64 
• 5 raised to the power 3 
• A random integer between 1 and 50 
• A random choice from this list: 
["Python", "SQL", "Excel", "Power BI", "Pandas"] 
"""

import math
import random

course = ["Python", "SQL", "Excel", "Power BI", "Pandas"] 

print(math.sqrt(64))
print(math.pow(5,3))
print(random.randint(1,50))
print(random.choice(course))
```
