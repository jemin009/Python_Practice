```python
#1
""" 
Take two numbers as input. 

Print: 
• Larger number 
• Smaller number 
• Difference between them 
"""

num1 = int(input("Entre a first number :"))
num2 = int(input("Entre a second number :"))

if num1 > num2:
    print(f"num1 {num1} is largest number")
else:
    print(f"num2 {num2} is largest number")

if num1 < num2:
    print(f"num1 {num1} is smaller number")
else:
    print(f"num2 {num2} is smaller number")

print(f"Difference between them is {abs(num1-num2)}")
```

```python
#2
""" 
Take a string as input. 

Print: 
• First 2 characters 
• Last 2 characters 
• Middle part 
• Reverse of the string 
"""

string = input("Entre a string :")

print(string[0:2])
print(string[-2:])
print(string[1:len(string)-1])
print(string[::-1])
```

```python
#3
""" 
Take a word as input. 
 
Check and print: 
• Whether the word starts with "P" 
• Whether the word ends with "n" 
• Uppercase version 
• Lowercase version 
"""

word = input("Enter a word :")

print(word.startswith("P"))
print(word.endswith("n"))
print(word.upper())
print(word.lower())
```

```python
#6
""" 
Create two sets. 
 
Print: 
• Union 
• Intersection 
• Elements only in the first set 
• Elements only in the second set 
"""

set1 = {10,20,30,40,50}
set2 = {40,50,60,70,80}

print(set1.union(set2))
print(set1.intersection(set2))
print(set1-set2)
print(set2-set1)
```

```python
#9
""" 
Create a list of numbers. 
 
Find and print: 
• Total of all numbers 
• Average of the numbers 
• Number of elements 
 
Do not use sum(). 
"""

num = [12,78,42,34,106,42,96,108]

total = 0
count = 0
for i in num:
    total += i
    count += 1
avg = total/count

print(f"Total of all numbers : {total}")
print(f"Average of all numbers : {avg}")
print(f"Number of elements : {count}")
```

```python
#10
""" 
Use the math and random modules. 
 
Print: 
• Absolute value of -45 
• Square root of 144 
• 2 raised to the power 5 
• A random integer between 10 and 100 
• A random choice from: 
  ["Python", "SQL", "Excel", "Pandas"] 
"""

import math
import random

print(abs(-45))
print(math.sqrt(144))
print(math.pow(2,5))
print(random.randint(10,100))

course = ["Python", "SQL", "Excel", "Pandas"] 

print(random.choice(course))
```
