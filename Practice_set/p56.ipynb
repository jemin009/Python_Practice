```python
#1
""" 
Create a list of numbers. 

Find: 
• Total of all numbers 
• Average 
• Count of even numbers 
• Count of odd numbers 

Do not use sum(). 
"""

num = [12,34,56,3,106,96]

total = 0
count = 0
even = 0
odd = 0

for i in num:
    if i % 2 == 0:
        even += 1
    else:
        odd += 1
    total += i
    count += 1
avg = total/count

print(f"Total : {total}")
print(f"Average : {avg}")
print(f"Count of even numbers : {even}")
print(f"Count of odd numbers : {odd}")
```

```python
#2
""" 
Take a string as input. 

Find: 
• Total number of characters 
• Number of characters at even indexes 
• Number of characters at odd indexes 
• Reverse of the string 

Use indexing/slicing where appropriate. 
"""

string = input("Enter a string :")

print(f"Total number of characters : {len(string)}")
print(f"Number of characters at even indexes  : {len(string[::2])}")
print(f"Number of characters at odd indexes  : {len(string[1::2])}")
print(f"Reverse of the string  : {string[::-1]}")
```

```python
#3
""" 
Create a list of 8 numbers. 
 
Create a new list containing the square 
of every number. 
 
Example: 
[2, 3, 4] → [4, 9, 16] 
 
Do not use any function that directly calculates 
all squares at once. 
"""

num = [12,42,34,23,67,45,77,33]

new = []
for i in num:
    new.append(i**2)

print(new)
```

```python
#4
""" 
Create a list of 5 three-digit numbers. 
 
For every number, calculate: 
• First digit 
• Middle digit 
• Last digit 
• Sum of its digits 
 
Print the information for every number. 
"""

num = [123,106,142,196,108]

for i in num:
    print(i)
    print(f"First digit : {str(i)[0]}")
    print(f"Middle digit : {str(i)[1]}")
    print(f"Last digit : {str(i)[-1]}")
    print(f"Sum of its digits : {int(str(i)[0]) + int(str(i)[1]) + int(str(i)[-1])}")
    print()
```

```python
#5
""" 
Create a list of 5 students with their marks. 
 
Convert it into a dictionary: 
Key   → Student Name 
Value → Marks 
 
Then create another dictionary containing 
only the students whose marks are greater than 70. 
"""

students = [
    ["Rahul", 85],
    ["Neha", 92],
    ["Amit", 50],
    ["Riya", 95],
    ["Jay", 88]
]

new = {}
great = {}
for i in students:
    new[i[0]] = i[1]

    if i[1] > 70:
        great[i[0]] = i[1]

print(new)
print(great)
```

```python
#7
""" 
Create a list of numbers. 
Find the largest number manually. 
 
Do not use: 
• max() 
• sorted() 
 
Use a variable to keep track of the largest number 
while going through the list. 
"""

num = [12,42,34,23,67,45,77,33]

largest = num[0]

for i in num:
    if i > largest:
        largest = i

print(largest)
```

```python
#8
""" 
Create a list of numbers. 
Find the smallest number manually. 
 
Do not use: 
• min() 
• sorted() 
 
Use a variable to keep track of the smallest number. 
"""

num = [12,42,34,23,67,45,77,33]

smallest = num[0]

for i in num:
    if i < smallest:
        smallest = i

print(smallest)
```

```python
#9
""" 
Create a list containing repeated numbers. 
Count how many times each number appears. 
 
Create a dictionary: 
Key   → Number 
Value → Frequency 
 
Example: 
[2, 3, 2, 4, 3, 2] 
 
Output: 
{2: 3, 3: 2, 4: 1} 
"""

num = [2, 5, 2, 8, 5, 2, 9, 8]

new = {}
for i in num:
    if i in new:
        new[i] += 1
    else:
        new[i] = 1

print(new)
```

```python
#10
""" 
Use the random module to generate 
10 random integers between 1 and 100. 
 
Store them in a list. 
 
Then find and print: 
• Generated list 
• Total 
• Average 
• Largest number 
• Smallest number 
• Count of even numbers 
 
Do not use: 
• sum() 
• max() 
• min() 
• sorted() 
"""

import random

number = []
total = 0
count = 0
largest = 0
smallest = float("inf")
even = 0

for i in range(10):
    gen = random.randint(1, 100)
    number.append(gen)

    total += gen
    count += 1

    if gen > largest:
        largest = gen

    if gen < smallest:
        smallest = gen

    if gen % 2 == 0:
        even += 1

print(f"Generated list : {number}")
print(f"Total : {total}")
print(f"Average : {total / count}")
print(f"Largest : {largest}")
print(f"Smallest : {smallest}")
print(f"Even numbers : {even}")
```
