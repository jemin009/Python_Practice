```python
#1
""" 
Create a list of 7 numbers. 
Create a new list containing the cube 
of every number. 

Example: 
[2, 3, 4] → [8, 27, 64] 
"""

import math 

num = [12, 45, 67, 23, 89, 34, 56]

new = []
for i in num:
    new.append(pow(i,3))

print(new)
```

```python
#2
""" 
Create a list of numbers. 

Find and print: 
• Total numbers 
• Count of positive numbers 
• Count of negative numbers 
• Count of zero 

Do not use any built-in function for counting. 
"""

num = [-12, 0, 67, -23, 89, -34, 0]

count = 0
count_positive = 0
count_negative = 0
count_zero = 0

for i in num:
    if i > 0:
        count_positive += 1
    elif i < 0:
        count_negative += 1
    else:
        count_zero += 1
    count += 1

print(f"Total numbers : {count}")
print(f"Count of positive numbers : {count_positive}")
print(f"Count of negative numbers : {count_negative}")
print(f"Count of zero : {count_zero}")
```

```python
#3
""" 
Create a list of 6 three-digit numbers. 
 
For every number, print: 
• Number 
• First digit 
• Middle digit 
• Last digit 
• Product of its digits 
"""

num = [123, 456, 789, 321, 654, 987]

for i in num:
    txt = str(i)
    print(f"Number : {i}")
    print(f"First digit : {(int(txt[0]))}")
    print(f"Middle digit : {(int(txt[1]))}")
    print(f"Last digit : {(int(txt[-1]))}")
    print(f"Product of its digits : {int(txt[0]) * int(txt[1]) * int(txt[2])}")
    print()
```

```python
#4
""" 
Take a string as input. 
 
Print: 
• First 2 characters 
• Last 2 characters 
• Characters from index 1 to 4 
• Characters with step 2 
• Reverse of the string 
"""

string = input("Enter a string :")

print(f"First 2 characters : {string[:2]}")
print(f"Last 2 characters : {string[-2:]}")
print(f"Characters from index 1 to 4 : {string[1:4]}")
print(f"Characters with step 2 : {string[::2]}")
print(f"Reverse of the string : {string[::-1]}")
```

```python
#5
""" 
Create a list of 8 numbers. 
 
Find: 
• Total 
• Average 
• Largest number 
• Smallest number 
 
Do not use: 
• sum() 
• max() 
• min() 
• sorted() 
"""

num = [12, 45, 67, 23, 89, 34, 56, 78]

total = 0
count = 0
largest = num[0]
smallest = num[0]

for i in num:
    total += i
    count += 1

    if i > largest:
        largest = i
    elif i < smallest:
        smallest = i

avg = total/count

print(f"Total : {total}")
print(f"Average : {avg}")
print(f"Largest number : {largest}")
print(f"Smallest number : {smallest}")
```

```python
#6
""" 
Create a list containing 6 products 
and their prices. 
 
Convert the list into a dictionary: 
Key   → Product Name 
Value → Price 
 
Then calculate the total price of all products 
without using sum(). 
""" 

products = [
    ["Laptop", 65000],
    ["Mobile", 25000],
    ["Tablet", 32000],
    ["TV", 45000],
    ["Speaker", 5000],
    ["Monitor", 18000]
]

new = {}
total = 0
for i in products:
    new[i[0]] = i[1]
    total += i[1]

print(new)
print(total)
```

```python
#7
""" 
Create a list containing repeated numbers. 
 
Create a dictionary: 
Key   → Number 
Value → Frequency 
 
Then print: 
• The complete frequency dictionary 
• Frequency of one specific number 
"""

num = [2, 5, 2, 8, 5, 2, 9, 8]

new = {}
for i in num:
    if i in new:
        new[i] += 1
    else:
        new[i] = 1

print(new)
print(new[9])
```

```python
#8
""" 
Create two sets of numbers. 
Create a new set containing only the elements 
that are present in both sets. 
 
Then print: 
• First set 
• Second set 
• New set 
"""

set1 = {10, 20, 30, 40, 50}
set2 = {30, 40, 50, 60, 70}

new = set()
for i in set1:
    if i in set2:
        new.add(i)

print(f"First set : {set1}")
print(f"Second set : {set2}")
print(f"New set : {new}")
```

```python
#9
""" 
Create a list of numbers. 
Take a number as input. 
 
Check whether that number exists 
in the list. 
 
Print: 
• "Found" if it exists 
• "Not Found" if it does not exist 
 
Do not use: 
• in 
• count() 
"""

num = [12, 45, 67, 23, 89, 34, 56, 78]

number = int(input("Enter a number :"))

found = False

for i in num:
    if number == i:
        found = True
        break

if found:
    print("Found")
else:
    print("Not Found")
```

```python
#10
""" 
Use random module to generate 8 random integers 
between 1 and 100. 
 
Store them in a list. 
 
Then find and print: 
• Generated list 
• Total 
• Average 
• Largest number 
• Smallest number 
 
Do not use: 
• sum() 
• max() 
• min() 
• sorted() 
 
For largest and smallest, use variables 
to keep track of the values. 
"""

import random   

num = []
total = 0
count = 0
largest = 0
smallest = float("inf")

for i in range(8):
    gen = random.randint(1,100)
    num.append(gen)
    total += gen
    count += 1

    if gen > largest:
        largest = gen
    if gen < smallest:
        smallest = gen

avg = total/count

print(f"Generated list : {num}")
print(f"Total : {total}")
print(f"Average : {avg}")
print(f"Largest number : {largest}")
print(f"Smallest number : {smallest}")
```
