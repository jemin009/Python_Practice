```python
#1
""" 
Create a list of 8 numbers. 
 
Create a new list containing only the numbers 
whose square is greater than 1000. 
 
Do not use: 
• list comprehension 
• filter() 
"""

numbers = [12, 25, 8000, 4500, 17, 31, 6, 22]

new = []
for i in numbers:
    if i > 1000:
        new.append(i)

print(new)
```

```python
#2
""" 
Create a list of numbers. 
 
Find: 
• Total 
• Average 
• Count of numbers greater than the average 
 
Do not use: 
• sum() 
""" 

numbers = [12, 25, 8000, 4500, 17, 31, 6, 22]

total = 0
count = 0
for i in numbers:
    total += i
    count += 1

avg = total/count

avg_count = 0
for i in numbers:
    if i > avg:
        avg_count += 1

print(f"Total : {total}")
print(f"Average : {avg}")
print(f"Count of numbers greater than the average : {avg_count}")
```

```python
#3
""" 
Create a list of 6 three-digit numbers. 
 
For every number, calculate: 
• Sum of digits 
• Product of digits 
 
Create a new list containing the numbers 
whose sum of digits is greater than 12. 
"""

numbers = [123, 456, 789, 234, 567, 891]

new = []
for i in numbers:
    txt = str(i)

    total = int(txt[0]) + int(txt[1]) + int(txt[2])
    product = int(txt[0]) * int(txt[1]) * int(txt[2])

    print(f"Sum of digits : {total}")
    print(f"Product of digits : {product}")
    print()

    if total > 12:
        new.append(i)

print(new)
```

```python
#4
""" 
Take a string as input. 
 
Create a new string containing 
characters at even indexes. 
 
Then print: 
• Original string 
• New string 
• Length of new string 
• Reverse of new string 
"""

string = input("Enter a string: ")

new = string[::2]

print(f"Original string : {string}")
print(f"New string : {new}")
print(f"Length of new string : {len(new)}")
print(f"Reverse of new string : {new[::-1]}")
```

```python
#5
""" 
Create a list of 7 numbers. 
 
Find the second largest number manually. 
 
Do not use: 
• max() 
• sorted() 
""" 

numbers = [15, 42, 8, 73, 29, 56, 91]

f_large = numbers[0]
s_large = numbers[0]

for i in numbers:
    if i > f_large:
        s_large = f_large
        f_large = i

    elif i > s_large and i < f_large:
        s_large = i 

print(f"Second largest number : {s_large}")
```

```python
#6
""" 
Create a list containing 6 students 
with their marks. 
 
Convert it into a dictionary: 
Key   → Student Name 
Value → Marks 
 
Then create another dictionary containing 
only students whose marks are greater than 
the average marks. 
"""

students = [
    ["Rahul", 78],
    ["Priya", 85],
    ["Amit", 67],
    ["Neha", 92],
    ["Karan", 74],
    ["Riya", 88]
]

dictionary = {}
great_avg = {}

total = 0
count = 0
for i in students:
    dictionary[i[0]] = i[1]

    total += i[1]
    count += 1
avg = total/count

for key,value in dictionary.items():
    if value > avg:
        great_avg[key] = value

print(dictionary)
print(great_avg)
```

```python
#7
""" 
Create two sets of numbers. 
 
Create a new set containing elements 
that are present in the first set 
but not in the second set. 
 
Do not use: 
• difference() 
"""

set1 = {10, 20, 30, 40, 50}
set2 = {30, 40, 50, 60, 70}

new = set()
for i in set1:
    if i not in set2:
        new.add(i)

print(new)
```

```python
#8
""" 
Create a list of products: 
[Product Name, Selling Price, Cost Price, Quantity] 
 
Calculate the profit amount for every product: 
 
Profit = (Selling Price - Cost Price) × Quantity 
 
Create a dictionary: 
Key   → Product Name 
Value → Profit Amount 
"""

products = [
    ["Laptop", 55000, 45000, 5],
    ["Phone", 30000, 24000, 8],
    ["Headphones", 2500, 1800, 12],
    ["Keyboard", 1500, 1000, 10],
    ["Mouse", 800, 500, 15]
]

new = {}
for i in products:
    new[i[0]] = (i[1] - i[2]) * i[3]

print(new)
```

```python
#9
""" 
Create a list containing repeated numbers. 
 
Create a frequency dictionary. 
 
Then find the number having 
the highest frequency. 
 
Do not use: 
• max() 
• sorted() 
"""

numbers = [10, 20, 10, 30, 20, 10, 40, 30, 20, 10]

freq = {}
highest = 0
for i in numbers:
    if i in freq:
        freq[i] += 1
    else:
        freq[i] = 1

if freq[i] > highest:
    highest = freq[i]

print(freq)
print(f"Highest frequency : {highest}")
```

```python
#10
""" 
Use random module to generate 10 random integers 
between 1 and 100. 
 
Store them in a list. 
 
Find manually: 
• Total 
• Average 
• Largest 
• Smallest 
• Count of even numbers 
• Count of numbers greater than the average 
 
Do not use: 
• sum() 
• max() 
• min() 
• sorted() 
 
Print all results. 
""" 

import random

num_list = []
total = 0
count = 0
largest = 0
smallest = float("inf")
even_count = 0
great_avg = 0

for i in range(10):
    num = random.randint(1,100)
    num_list.append(num)

    if num > largest:
        largest = num

    if num < smallest:
        smallest = num

    if num % 2 == 0:
        even_count += 1

    total += num
    count += 1
avg = total/count

for i in num_list:
    if i > avg:
        great_avg += 1

print(f"Generated list : {num_list}")
print(f"Total : {total}")
print(f"Average : {avg}")
print(f"Largest : {largest}")
print(f"Smallest : {smallest}")
print(f"Count of even numbers : {even_count}")
print(f"Count of numbers greater than the average : {great_avg}")
```
