```python
#1
""" 
Create a list of 10 numbers. 
 
Find the: 
• Largest number 
• Second largest number 
• Difference between them 
 
Do not use: 
• max() 
• sorted() 
"""

numbers = [15, 42, 8, 73, 29, 56, 91, 34, 67, 12]

f_largest = numbers[0]
s_largest = numbers[1]

for i in numbers:
    if i > f_largest:
        s_largest = f_largest
        f_largest = i

    elif i > s_largest and i < f_largest:
        s_largest = i

    dif = abs(s_largest - f_largest)

print(f"Largest number : {f_largest}")
print(f"Second largest number : {s_largest}")
print(f"Difference between them : {dif}")
```

```python
#2
""" 
Create a list of 8 numbers. 
 
Create a new list containing numbers 
whose value is greater than the average 
but less than the largest number. 
 
Do not use: 
• sum() 
• max() 
• sorted() 
"""

numbers = [18, 45, 7, 62, 31, 89, 24, 53]

largest = numbers[0]

new = []

total = 0
count = 0
for i in numbers:
    total += i
    count += 1

    if i > largest:
        largest = i
avg = total/count

for i in numbers:
    if i > avg and i < largest:
        new.append(i)

print(new)
```

```python
#3
""" 
Create a list of 7 three-digit numbers. 
 
For every number, calculate: 
• Sum of digits 
• Product of digits 
 
Count how many numbers have: 
• Digit sum greater than 12 
• Digit product greater than 20 
"""

numbers = [123, 456, 789, 234, 567, 891, 345]

great_12 = 0
great_20 = 0

for i in numbers:
    txt = str(i)

    total = int(txt[0]) + int(txt[1]) + int(txt[2])
    product = int(txt[0]) * int(txt[1]) * int(txt[2])

    print(f"Sum of digits : {total}")
    print(f"Product of digits : {product}")
    print()

    if total > 12:
        great_12 += 1
    if product > 20:
        great_20 += 1

print(f"Digit sum greater than 12 total {great_12} numbers")
print(f"Digit product greater than 20 total {great_20} numbers")
```

```python
#5
""" 
Create a list containing repeated numbers. 
 
Create a frequency dictionary. 
 
Then manually find: 
• Number with the highest frequency 
• Its frequency 
 
Do not use: 
• max() 
• sorted() 
"""

numbers = [10, 20, 10, 30, 20, 10, 40, 30, 20, 10]

freq = {}
for i in numbers:
    if i in freq:
        freq[i] += 1
    else:
        freq[i] = 1

highest_freq = 0
highest_num = 0
for key,value in freq.items():
    if value > highest_freq:
        highest_freq = value
        highest_num = key

print(freq)
print(f"Highest frequency number is {highest_num} and its frequency is {highest_freq}")
```

```python
#6
""" 
Create a list containing 6 students 
with their marks. 
 
Convert it into a dictionary: 
Key   → Student Name 
Value → Marks 
 
Find: 
• Average marks 
• Student with the highest marks 
• Students who scored above average 
 
Do not use: 
• max() 
""" 

students = [
    ["Rahul", 78],
    ["Priya", 85],
    ["Amit", 67],
    ["Neha", 92],
    ["Karan", 74],
    ["Riya", 88]
]

new = {}
great_avg = {}

total = 0
count = 0
highest_mark = 0
highest_stu = None

for i in students:
    new[i[0]] = i[1]
    total += i[1]
    count += 1

    if i[1] > highest_mark:
        highest_mark = i[1]
        highest_stu = i[0]
        
avg = total/count

for key,value in new.items():
    if value > avg:
        great_avg[key] = value

print(f"Average marks : {avg}")
print(f"Student with the highest marks : {highest_stu} : {highest_mark}")
print(f"Students who scored above average : {great_avg}")
```

```python
#7
""" 
Create two sets of numbers. 
Create a new set containing numbers 

that are present in either set, 
but not present in both sets. 
 
Do not use: 
• symmetric_difference() 
"""

set1 = {10, 20, 30, 40, 50}
set2 = {30, 40, 50, 60, 70}

new = set()

for i in set1:
    if i not in set2:
        new.add(i)

for i in set2:
    if i not in set1:
        new.add(i)

print(new) 
```

```python
#8
""" 
Create a list of 5 products: 
[Product Name, Selling Price, Cost Price, Quantity] 
 
Calculate for every product: 
Profit = (Selling Price - Cost Price) × Quantity 
 
Create a dictionary: 
Key   → Product Name 
Value → Profit 
 
Then find the product with the 
highest total profit. 
 
Do not use: 
• max() 
"""

products = [
    ["Laptop", 55000, 42000, 3],
    ["Mouse", 1200, 700, 10],
    ["Keyboard", 2500, 1500, 6],
    ["Monitor", 18000, 13000, 4],
    ["Headphones", 3500, 2200, 8]
]

new = {}
profit = {}
highest_profit = 0
highest_product = None

for i in products:
    new[i[0]] = i[1]
    profit[i[0]] = (i[1] - i[2]) * i[3]

    if (i[1] - i[2]) * i[3] > highest_profit:
        highest_profit = (i[1] - i[2]) * i[3]
        highest_product = i[0]

print(new)
print(f"Product with the highest total profit :- {highest_product} : {highest_profit}")
```

```python
#9
""" 
Create a list of numbers. 
 
Create a frequency dictionary. 
 
Then create a new dictionary containing 
only numbers whose frequency is greater than 1. 
 
Example: 
 
Input: 
[10, 20, 10, 30, 20, 10] 
 
Output: 
{10: 3, 20: 2} 
"""

numbers = [10, 20, 10, 30, 20, 10, 40, 30, 50, 20]

freq = {}
new = {}
for i in numbers:
    if i in freq:
        freq[i] += 1
    else:
        freq[i] = 1

for key,value in freq.items():
    if value > 1:
        new[key] = value

print(new)
```

```python
#10
""" 
Use random module to generate 12 random integers 
between 1 and 100. 
 
Store them in a list. 
 
Find manually: 
• Total 
• Average 
• Largest 
• Smallest 
• Second largest 
• Count of even numbers 
• Count of odd numbers 
• Count of numbers greater than average 
 
Do not use: 
• sum() 
• max() 
• min() 
• sorted() 
"""

import random

num_list = []
total = 0
count = 0 
largest = 0
smallest = float("inf")
s_largest = 0
count_even = 0
count_odd = 0
great_avg = 0

for i in range(12):
    gen = random.randint(1,100)
    num_list.append(gen)

    total += gen
    count += 1

    if gen > largest:
        largest = gen

    if gen < smallest:
        smallest = gen

    if gen > smallest and gen < largest:
        s_largest = gen

    if gen % 2 == 0:
        count_even += 1

    if gen % 2 != 0:
        count_odd += 1

avg = total/count

for i in num_list:
    if i > avg:
        great_avg += 1


print(f"Total : {total}")
print(f"Average : {avg}")
print(f"Largest : {largest}")
print(f"Smallest : {smallest}")
print(f"Second largest : {s_largest}")
print(f"Count of even numbers : {count_even}")
print(f"Count of odd numbers : {count_odd}")
print(f"Count of numbers greater than average : {great_avg}")
```
