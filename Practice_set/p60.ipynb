```python
#1
""" 
Create a list of 10 numbers. 

Find manually: 
• Largest number 
• Second largest number 
• Smallest number 
• Second smallest number 

Do not use: 
• max() 
• min() 
• sorted() 
"""

numbers = [15, 42, 8, 73, 29, 56, 91, 34, 67, 12]

f_largest = numbers[0]
s_largest = numbers[0]

f_smallest = numbers[0]
s_smallest = numbers[0]

for i in numbers:
    if i > f_largest:
        s_largest = f_largest
        f_largest = i
    
    elif i > s_largest and i < f_largest:
        s_largest = i

for i in numbers:
    if i < f_smallest:
        s_smallest = f_smallest
        f_smallest = i
    
    elif i < s_smallest and i > f_smallest:
        s_smallest = i

print(f"Largest number : {f_largest}")
print(f"Second largest number : {s_largest}")
print(f"Smallest number : {f_smallest}")
print(f"Second smallest number : {s_smallest}")
```

```python
#2
""" 
Create a list of 10 numbers. 
 
Calculate: 
• Total 
• Average 
 
Then create a new list containing numbers 
that are: 
• Greater than average 
• Even 
 
Do not use: 
• sum() 
"""

numbers = [24, 57, 13, 89, 41, 6, 72, 35, 98, 19]

total = 0
count = 0
for i in numbers:
    total += i
    count += 1
avg = total/count

new = []
for i in numbers:
    if i > avg and i % 2 == 0:
        new.append(i)

print(new)
```

```python
#3
""" 
Create a list of 8 three-digit numbers. 
 
For every number: 
• Calculate digit sum 
• Calculate digit product 
 
Then count how many numbers have: 
• Digit sum greater than 10 
• Digit product greater than 15 
"""

numbers = [123, 456, 789, 234, 567, 891, 345, 678]

count1 = 0
count2 = 0
for i in numbers:
    txt = str(i)

    total = int(txt[0]) + int(txt[1]) + int(txt[2])
    product = int(txt[0]) * int(txt[1]) * int(txt[2])

    if total > 10:
        count1 += 1
    if product > 15:
        count2 += 1

print(count1)
print(count2)
```

```python
#4
""" 
Take a string as input. 
 
Create a new string containing: 
• Characters at even indexes 
• Characters at odd indexes 
 
Print: 
• Original string 
• Even-index string 
• Odd-index string 
• Length of both new strings 
""" 

string = input("Enter a string :")

even = string[::2]
odd = string[1::2]

print(f"Original string : {string}")
print(f"Even-index string : {even}")
print(f"Odd-index string : {odd}")
print(f"Length of even string : {len(even)}")
print(f"Length of odd string : {len(odd)}")
```

```python
#5
""" 
Create a list containing repeated numbers. 
 
Create a frequency dictionary. 
 
Then manually find: 
• Number with the highest frequency 
• Number with the lowest frequency 
"""

numbers = [10, 20, 10, 30, 20, 10, 40, 30, 50, 20]

freq = {}

highest_freq = 0
highest_num = 0

lowest_freq = 0
lowest_num = 0

for i in numbers:
    if i in freq:
        freq[i] += 1
        freq[i] > highest_freq
        highest_freq = i
        highest_num = freq[i]
    else:
        freq[i] = 1
        freq[i] < lowest_freq
        lowest_freq = i
        lowest_num = freq[i]

print(f"Number with the highest frequency :- {highest_num} : {highest_freq}")
print(f"Number with the lowest frequency :- {lowest_num} : {lowest_freq}")
```

```python
#6
""" 
Create a list of 6 students: 
[Name, Marks] 
 
Convert it into a dictionary: 
Key   → Student Name 
Value → Marks 
 
Find: 
• Average marks 
• Highest-scoring student 
• Students who scored below average 
 
Do not use: 
• max() 
"""

students = [
    ["Rahul", 78],
    ["Priya", 85],
    ["Amit", 72],
    ["Neha", 92],
    ["Riya", 81],
    ["Karan", 68]
]

new = {}
total = 0
count = 0

for i in students:
    new[i[0]] = i[1]
    total += i[1]
    count += 1
avg = total/count

below_avg = {}
highest_mark = 0
highest_stu = None

for key,value in new.items():
    if value > highest_mark:
        highest_mark = value
        highest_stu = key

for key,value in new.items():
    if value < avg:
        below_avg[key] = value

print(new)
print(f"Average marks : {avg}")
print(f"Highest scoring student : {highest_stu}")
print(f"Students who scored below average : {below_avg}")
```

```python
#8
""" 
Create a list of 5 products: 
[Product Name, Selling Price, Cost Price, Quantity] 
 
Calculate: 
Profit = (Selling Price - Cost Price) × Quantity 
 
Create a dictionary: 
Key   → Product Name 
Value → Profit 
 
Then manually find: 
• Total profit 
• Product with the lowest profit 
 
Do not use: 
• min() 
"""

products = [
    ["Laptop", 55000, 42000, 3],
    ["Mouse", 1200, 700, 10],
    ["Keyboard", 2500, 1500, 6],
    ["Monitor", 18000, 13000, 4],
    ["Headphones", 3500, 2200, 8]
]

new = {}
total = 0
lowest_profit = float("inf")
lowest_product = None

for i in products:
    profit = (i[1] - i[2]) * i[3]

    new[i[0]] = profit
    total += profit

    if profit < lowest_profit:
        lowest_profit = profit
        lowest_product = i[0]
        
print(f"Total profit : {total}")
print(f"Product with the lowest profit : {lowest_product} : {lowest_profit}")
```

```python
#9
""" 
Create a list of numbers. 
 
Create a frequency dictionary. 
 
Then create two new dictionaries: 
1. Numbers appearing exactly once 
2. Numbers appearing more than twice 
 
Example: 
Input: 
[10, 20, 10, 30, 20, 10, 40] 
 
Output: 
Once       → {30: 1, 40: 1} 
More than 2 → {10: 3} 
"""

numbers = [10, 20, 10, 30, 20, 10, 40, 30, 50, 20]

freq = {}
for i in numbers:
    if i in freq:
        freq[i] += 1
    else:
        freq[i] = 1

once = {}
twise = {}

for key,value in freq.items():
    if value == 1:
        once[key] = value
    if value > 2:
        twise[key] = value

print(f"Numbers appearing exactly once : {once}")
print(f"Numbers appearing more than twice : {twise}")
```

```python
#10
""" 
Use random module to generate 15 random integers 
between 1 and 100. 
 
Store them in a list. 
 
Find manually: 
• Total 
• Average 
• Largest 
• Second largest 
• Smallest 
• Count of even numbers 
• Count of odd numbers 
• Count of numbers greater than average 
 
Do not use: 
• sum() 
• max() 
• min() 
• sorted() 
"""  

import random

num_list = []
total = 0
count = 0
largest = 0
s_largest = 0
smallest = float("inf")
count_even = 0
count_odd = 0
great_avg = 0

for i in range(15):
    gen = random.randint(1,100)
    num_list.append(gen)

    total += gen
    count += 1

    if gen > largest:
        s_largest = largest
        largest = gen

    elif gen > s_largest and gen < largest:
        s_largest = gen

    if gen < smallest:
        smallest = gen

    if gen % 2 == 0:
        count_even += 1

    if gen % 2 != 0:
        count_odd += 1

avg = total/count

for i in num_list:
    if i > avg:
        great_avg += 1

print(f"Total : {total}")
print(f"Average : {avg}")
print(f"Largest : {largest}")
print(f"Second largest : {s_largest}")
print(f"Smallest : {smallest}")
print(f"Count of even numbers : {count_even}")
print(f"Count of odd numbers : {count_odd}")
print(f"Count of numbers greater than average : {great_avg}")
```
