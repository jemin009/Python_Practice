```python
#1
""" 
Create a list of 10 numbers. 

Find manually: 
• Largest number 
• Second largest number 
• Third largest number 

Do not use: 
• max() 
• sorted() 
"""

numbers = [15, 42, 8, 73, 29, 56, 91, 34, 67, 12]

f_largest = numbers[0]
s_largest = numbers[0]
t_largest = numbers[0]

for i in numbers:
    if i > f_largest:
        t_largest = s_largest
        s_largest = f_largest
        f_largest = i

    elif i > s_largest and i < f_largest:
        t_largest = s_largest
        s_largest = i

    elif i > t_largest and i < s_largest:
        t_largest = i

print(f"First largest number : {f_largest}")
print(f"Second largest number : {s_largest}")
print(f"Third largest number : {t_largest}")
```

```python
#2
""" 
Create a list of 10 numbers. 

Find: 
• Total 
• Average 
• Count of numbers below average 
• Count of numbers above average 

Do not use: 
• sum() 
"""

numbers = [24, 57, 13, 89, 41, 6, 72, 35, 98, 19]

total = 0
count = 0
below_avg = 0
above_avg = 0

for i in numbers:
    total += i
    count += 1
avg = total/count

for i in numbers:
    if i > avg:
        above_avg += 1
    if i < avg:
        below_avg += 1

print(f"Total : {total}")
print(f"Average : {avg}")
print(f"Count of numbers below average : {below_avg}")
print(f"Count of numbers above average : {above_avg}")
```

```python
#3
""" 
Create a list of 8 three-digit numbers. 

For every number: 
• Calculate digit sum 
• Calculate digit product 

Create two new lists: 
1. Numbers with digit sum greater than 10 
2. Numbers with digit product greater than 20 
"""

numbers = [123, 456, 789, 234, 567, 891, 345, 678]

total_10 = []
product_20 = []

for i in numbers:
    txt = str(i)

    total = int(txt[0]) + int(txt[1]) + int(txt[2])
    product = int(txt[0]) * int(txt[1]) * int(txt[2])

    if total > 10:
        total_10.append(i)

    if product > 20:
        product_20.append(i)

    print(f"Total of {i}: {total}")
    print(f"Product of {i}: {product}")
    print()

print(f"Digit sum greater than 10 : {total_10}")
print(f"Digit product greater than 20 : {product_20}")

```

```python
#4
""" 
Take a string as input. 

Create: 
• A string containing characters at even indexes 
• A string containing characters at odd indexes 

Then find: 
• Which new string is longer 
• Length difference between them 
"""

string = "jeminramani"
#string = input("Enter a string :")

even = string[::2]
odd = string[1::2]

print(f"A string containing characters at even indexes : {even}")
print(f"A string containing characters at odd indexes : {odd}")

if len(even) > len(odd):
    print("Even indexes string is longer")
else:
    print("Odd indexes string is longer")

print(f"Length difference between them : {abs(len(even)-len(odd))}")
```

```python
#5
""" 
Create a list containing repeated numbers. 

Create a frequency dictionary. 

Then manually find: 
• Number with the highest frequency 
• Its frequency 
• Number with the lowest frequency 
• Its frequency 

Do not use: 
• max() 
• min() 
"""

numbers = [10, 20, 10, 30, 20, 10, 40, 30, 50, 20]

freq = {}
highest_freq = 0
highest_num = 0
lowest_freq = float("inf")
lowest_num = 0

for i in numbers:
    if i in freq:
        freq[i] += 1
    else:
        freq[i] = 1

for key, value in freq.items():
    if value > highest_freq:
        highest_freq = value
        highest_num = key

    if value < lowest_freq:
        lowest_freq = value 
        lowest_num = key

print(freq)
print(f"Number with the highest frequency is {highest_num} and Its frequency is {highest_freq}")
print(f"Number with the lowest frequency is {lowest_num} and Its frequency is {lowest_freq}")
```

```python
#6
""" 
Create a list of 6 students: 
[Name, Marks] 

Convert it into a dictionary. 

Then create a new dictionary containing 
only students who scored 80 or more. 

Also calculate: 
• Average marks 
• Number of students above average 
"""

students = [
    ["Rahul", 78],
    ["Priya", 85],
    ["Amit", 72],
    ["Neha", 92],
    ["Riya", 81],
    ["Karan", 68]
]

dictionary = {}
scored_80 = {}
total = 0
above_avg = 0

for i in students:
    dictionary[i[0]] = i[1]

    if i[1] >= 80:
        scored_80[i[0]] = i[1]

    total += i[1]
avg = total/len(dictionary)

for key,value in dictionary.items():
    if value > avg:
        above_avg += 1

print(dictionary)
print(f"students who scored 80 or more : {scored_80}")
print(f"Average marks : {avg}")
print(f"Number of students above average : {above_avg}")
```

```python
#7
""" 
Create two sets of numbers. 

Create a new set containing numbers 
that are present in set1 but NOT in set2. 

Then create another set containing numbers 
that are present in set2 but NOT in set1. 

Do not use: 
• difference() 
• - 
"""

set1 = {10, 20, 30, 40, 50}
set2 = {30, 40, 50, 60, 70}

new1 = set()
new2 = set()

for i in set1:
    if i not in set2:
        new1.add(i)

for i in set2:
    if i not in set1:
        new2.add(i)

print(new1)
print(new2)
```

```python
#8
""" 
Create a list of 5 products: 
[Product Name, Selling Price, Cost Price, Quantity] 

Calculate total profit for every product: 
Profit = (Selling Price - Cost Price) × Quantity 

Create a dictionary: 
Product Name → Profit 

Then manually find: 
• Highest-profit product 
• Lowest-profit product 
• Difference between highest and lowest profit 

Do not use: 
• max() 
• min() 
"""

products = [
    ["Laptop", 55000, 42000, 3],
    ["Mouse", 1200, 700, 10],
    ["Keyboard", 2500, 1600, 6],
    ["Monitor", 18000, 13000, 4],
    ["Headphones", 3500, 2200, 8]
]

dictionary = {}
highest_product = None
highest_profit = 0

lowest_product = None
lowest_profit = float("inf")

for i in products:
    profit = (i[1] - i[2]) * i[3]
    dictionary[i[0]] = profit

    if profit > highest_profit:
        highest_profit = profit
        highest_product = i[0]

    if profit < lowest_profit:
        lowest_profit = profit
        lowest_product = i[0]

print(dictionary)
print(f"Highest-profit product is {highest_product} : {highest_profit}")
print(f"Lowest-profit product is {lowest_product} : {lowest_profit}")
print(f"Difference between highest and lowest profit : {abs(highest_profit - lowest_profit)}")
```

```python
#9
""" 
Create a list of numbers. 

Create a frequency dictionary. 

Then create three new dictionaries: 
1. Numbers appearing exactly once 
2. Numbers appearing exactly twice 
3. Numbers appearing more than twice 

Do not use Counter(). 
"""

numbers = [10, 20, 10, 30, 20, 10, 40, 30, 50, 20]

freq = {}
once = {}
twice = {}
more_twice = {}

for i in numbers:
    if i in freq:
        freq[i] += 1
    else:
        freq[i] = 1

for key,value in freq.items():
    if value == 1:
        once[key] = value
    if value == 2:
        twice[key] = value
    if value > 2:
        more_twice[key] = value

print(f"Numbers appearing exactly once : {once}")
print(f"Numbers appearing exactly twice : {twice}")
print(f"Numbers appearing more than twice : {more_twice}")
```

```python
#10
""" 
Use random module to generate 15 random integers 
between 1 and 100. 

Store them in a list. 

Find manually: 
• Total 
• Average 
• Largest 
• Second largest 
• Smallest 
• Second smallest 
• Count of even numbers 
• Count of odd numbers 
• Count of numbers greater than average 

Do not use: 
• sum() 
• max() 
• min() 
• sorted() 
"""

import random

num_list = []
total = 0
largest = 0
s_largest = 0
smallest = float("inf")
s_smallest = float("inf")
count_even = 0
count_odd = 0
great_avg = 0


for i in range(15):
    gen = random.randint(1,100)
    num_list.append(gen)

    total += gen

    if gen > largest:
        s_largest = largest
        largest = gen

    if gen > s_largest and gen < largest:
        s_largest = gen

    if gen < smallest:
        s_smallest = smallest
        smallest = gen

    if gen < s_smallest and gen > smallest:
        s_smallest = gen

    if gen % 2 == 0:
        count_even += 1

    if gen % 2 != 0:
        count_odd += 1

avg = total/len(num_list)

for i in num_list:
    if i > avg:
        great_avg += 1

print(f"Total : {total}")
print(f"Average : {avg}")
print(f"Largest : {largest}")
print(f"Second largest : {s_largest}")
print(f"Smallest : {smallest}")
print(f"Second smallest : {s_smallest}")
print(f"Count of even numbers : {count_even}")
print(f"Count of odd numbers : {count_odd}")
print(f"Count of numbers greater than average : {great_avg}")
```
