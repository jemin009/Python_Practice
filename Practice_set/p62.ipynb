```python
#1
"""
For every number, determine whether it is: 
 Positive  
 Negative  
 Zero  
Create three separate lists for these categories.
"""

numbers = [15, -8, 23, 0, -17, 42, 6, -3, 31, 0, -25, 14]

positive = []
negative = []
zero = []

for i in numbers:
    if i > 0:
        positive.append(i)
    elif i < 0:
        negative.append(i)
    else:
        zero.append(i)

print(f"Positive number : {positive}")
print(f"Negative number : {negative}")
print(f"Zero number : {zero}")
```

```python
#2
"""
Starting from the first number, calculate the running total after each number. 
 
Example: 
Numbers: [5, 8, 3] 
Running totals: [5, 13, 16] 
 
Do not use sum().
"""

numbers = [12, 47, 8, 63, 25, 91, 34, 56, 19, 72]

new = []
total = 0

for i in numbers:
    total += i

    new.append(total) 

print(new)
```

```python
#3
"""
Create a new list where: 
 Even numbers → mul ply by 2  
 Odd numbers → mul ply by 3  

Print both lists. 
"""

numbers = [12, 47, 8, 63, 25, 91, 34, 56, 19, 72]

Even = []
Odd = []

for i in numbers:
    if i % 2 == 0:
        Even.append(i * 2)
    elif i % 2 != 0:
        Odd.append(i * 3)

print(f"Even numbers : {Even}")
print(f"Odd numbers : {Odd}")
```

```python
#4
"""
Find manually: 
 Number of vowels  
 Number of consonants  
 Number of digits  
 Number of spaces  

Do not use any advanced library. 
"""

string = input("Enter a string :")

vowels = ["a","e","i","o","u","A","E","I","O","U"]
num_vowels = 0

consonants = [
    "B", "C", "D", "F", "G", "H", "J", "K", "L", "M", "N",
    "P", "Q", "R", "S", "T", "V", "W", "X", "Y", "Z",
    "b", "c", "d", "f", "g", "h", "j", "k", "l", "m", "n",
    "p", "q", "r", "s", "t", "v", "w", "x", "y", "z"
]
num_consonants = 0

num_digit = 0

num_spaces = 0

for i in string:
    if i in vowels:
        num_vowels += 1

    if i in consonants:
        num_consonants += 1

    if i.isdigit():
        num_digit += 1

    if i == " ":
        num_spaces += 1

print(num_vowels)
print(num_consonants)
print(num_digit)
print(num_spaces)
```

```python
#5
"""
Create a new list containing each number only once, while maintaining the original 
order. 
 
Do not use: 
 set()  
 dict.fromkeys()  
 
Example: 
Input:  [10, 20, 10, 30, 20] 
Output: [10, 20, 30]
"""

numbers = [10, 20, 10, 30, 20, 10, 40, 30, 50, 20]

new = []

for i in numbers:
    if i not in new:
        new.append(i)

print(new)
```

```python
#6
"""
Find the first number that appears more than once when scanning the list from left 
to right. 
 
Example: 
[5, 8, 3, 8, 2, 5] 
 
Answer: 
8 
Do not use count(). 
"""

numbers = [7, 12, 7, 25, 18, 12, 7, 30, 18, 25]

seen = set()

for i in numbers:
    if i in seen:
        print(f"First repeated number : {i}")
        break
    seen.add(i)
```

```python
#7
"""
Convert it into a dictionary. 
 
Then create: 
 Dictionary of students scoring 90 or more  
 Dictionary of students scoring between 70 and 89  
 Count of students scoring below 70  
Also find the highest marks manually.
"""

students = [
    ["Rahul", 78],
    ["Priya", 85],
    ["Amit", 72],
    ["Neha", 92],
    ["Riya", 81],
    ["Karan", 68]
]

dictionary = {}
dict_90 = {}
dict_70 = {}
count_70 = 0
highest_mark = 0

for i in students:
    dictionary[i[0]] = i[1]

    if i[1] > highest_mark:
        highest_mark = i[1]

    if i[1] >= 90:
        dict_90[i[0]] = i[1]

    elif i[1] >= 70 and i[1] < 90:
        dict_70[i[0]] = i[1]

    elif i[1] < 70:
        count_70 += 1

print(f"Dictionary : {dictionary}")
print(f"Dictionary of students scoring 90 or more : {dict_90}")
print(f"Dictionary of students scoring between 70 and 89 : {dict_70}")
print(f"Count of students scoring below 70 : {count_70}")
print(f"Highest mark : {highest_mark}")
```

```python
#8
"""
Calculate for every product: 
Stock Value = Price × Quantity 
 
Create: 
Product Name → Stock Value 
 
Then manually find: 
 Product with highest stock value  
 Product with lowest stock value  
 Difference between them  
Do not use max() or min().
"""

products = [
    ["Laptop", 55000, 3],
    ["Mouse", 800, 10],
    ["Keyboard", 1500, 6],
    ["Monitor", 12000, 4],
    ["Headphones", 2500, 8]
]

pro_stock = {}
highest_pro = None
highest_value = 0

lowest_pro = None
lowest_value = float("inf")


for i in products:
    stock_value = i[1] * i[2]
    pro_stock[i[0]] = stock_value

    if stock_value > highest_value:
        highest_value = stock_value
        highest_pro = i[0]

    if stock_value < lowest_value:
        lowest_value = stock_value
        lowest_pro = i[0]

print(f"Product-Stock_value : {pro_stock}")
print(f"Product with highest stock value : {highest_pro} : {highest_value}")
print(f"Product with lowest stock value :- {lowest_pro} : {lowest_value}")
print(f"Difference between them :- {abs(highest_value - lowest_value)}")
```

```python
#9
"""
Create a new list containing numbers that appear in both lists. 
 
Rules: 
 Do not use set()  
 Do not use intersection()  
 Do not use &  

Also make sure a number is added to the new list only once, even if it is repeated.
"""

list1 = [10, 20, 30, 40, 50, 20, 60, 30]
list2 = [20, 30, 70, 80, 40, 20, 90]

new = []

for i in list1:
    if i in list2 and i not in new:
        new.append(i)

print(new)
```

```python
#10
"""
Find manually: 
 Total  
 Average  
 Count of numbers below average  
 Count of numbers above average  
 Count of even numbers  
 Count of odd numbers  
 Largest number  
 Smallest number  
 Difference between largest and smallest  
 Create a new list containing numbers greater than average and even  
 
Do not use: 
 sum()  
 max()  
 min()  
 sorted()
"""

import random

num_list = []
total = 0
below_avg = 0
count_even = 0
count_odd = 0
largest_num = 0
smallest_num = float("inf")

for i in range(15):
    gen = random.randint(1,100)

    num_list.append(gen)

    total += gen

    if gen % 2 == 0:
        count_even += 1

    if gen % 2 != 0:
        count_odd += 1

    if gen > largest_num:
        largest_num = gen

    if gen < smallest_num:
        smallest_num = gen

avg = total/len(num_list)

new = []
for i in num_list:
    if i < avg:
        below_avg += 1

    if i > avg:
        above_avg += 1

    if i > avg and i % 2 == 0:
        new.append(i)

print(num_list)
print(f"Total : {total}")
print(f"Average : {avg}")
print(f"Count of numbers below average : {below_avg}")
print(f"Count of numbers above average : {above_avg}")
print(f"Count of even numbers : {count_even}")
print(f"Count of odd numbers : {count_odd}")
print(f"Largest number : {largest_num}")
print(f"Smallest number : {smallest_num}")
print(f"Difference between largest and smallest : {abs(largest_num - smallest_num)}")
print(f"list containing numbers greater than average and even : {new}")
```
