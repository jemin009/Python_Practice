```python
#1
"""
Perform: 
1. Create three new lists: positive, negative, and zero.  
2. Count how many positive numbers.  
3. Count how many negative numbers.  
4. Count how many zeros.  
5. Find the largest positive number manually. 
6. Find the smallest negative number manually.  
"""

numbers = [15, -8, 23, 0, -17, 42, 6, -3, 31, 0] 

positive = []
count_positive = 0

negative = []
count_negative = 0

zero = []
count_zero = 0

largest_positive = numbers[0]
smallest_negative = float("inf")

for i in numbers:
    if i > 0:
        positive.append(i)
        count_positive += 1

    elif i < 0:
        negative.append(i)
        count_negative += 1

    else:
        zero.append(i)
        count_zero += 1

    if i > largest_positive:
        largest_positive = i

    if i < smallest_negative:
        smallest_negative = i

print(f"Count of positive numbers : {count_positive}")
print(f"Count of negative numbers : {count_negative}")
print(f"Count of zeros : {count_zero}")
print(f"largest positive number : {largest_positive}")
print(f"smallest negative number : {smallest_negative}")
```

```python
#2
"""
Create a new list containing the running total after each number. 
 
Example: 
[5, 8, 3] → [5, 13, 16] 
Do not use sum().
"""

numbers = [12, 8, 15, 5, 20, 10, 7, 18] 

new = []
total = 0

for i in numbers:
    total += i
    new.append(total)

print(new)
```

```python
#3
"""
Create a new list where: 
 Even number → mul ply by 2  
 Odd number → mul ply by 3  
Also count how many even and odd numbers there are.
"""

numbers = [12, 7, 18, 25, 30, 11, 42, 9, 16, 21] 

even = []
count_even = 0

odd = []
count_odd = 0

for i in numbers:
    if i % 2 == 0:
        even.append(i*2)
        count_even += 1

    if i % 2 != 0:
        odd.append(i*3)
        count_odd += 1

print(f"Even list : {even}")
print(f"Count of even : {count_even}")

print(f"Odd list : {odd}")
print(f"Count of odd : {count_odd}")
```

```python
#4
"""
Take a string as input. 

Find: 
 Number of vowels  
 Number of consonants  
 Number of digits  
 Number of spaces  
 
Example: 
Input: Jemin Ramani 123 
Do not use advanced libraries.
"""

# string = input("Enter a string :")
string = "Jemin Ramani 123"

vowels = [
    "a", "e", "i", "o", "u",
    "A", "E", "I", "O", "U"
]

consonants = [
    "b", "c", "d", "f", "g", "h", "j", "k", "l", "m",
    "n", "p", "q", "r", "s", "t", "v", "w", "x", "y", "z",
    "B", "C", "D", "F", "G", "H", "J", "K", "L", "M",
    "N", "P", "Q", "R", "S", "T", "V", "W", "X", "Y", "Z"
]

count_vowels = 0
count_consonants = 0
count_digit = 0
count_space = 0

for i in string:
    if i in vowels:
        count_vowels += 1

    if i in consonants:
        count_consonants += 1

    if i.isdigit():
        count_digit += 1

    if i == " ":
        count_space += 1


print(f"Numbers of vowels : {count_vowels}")
print(f"Numbers of consonants : {count_consonants}")
print(f"Numbers of digit : {count_digit}")
print(f"Numbers of space : {count_space}")
```

```python
#5
"""
Create a new list where every number appears only once, while maintaining the 
original order. 
 
Do not use: 
 set()  
 dict.fromkeys() 
"""

numbers = [10, 20, 10, 30, 20, 40, 30, 50, 10, 40] 

new = []
for i in numbers:
    if i not in new:
        new.append(i)

print(new)
```

```python
#7
"""
Perform: 
1. Convert it into a dictionary.  
2. Create a dictionary of students scoring 90 or more.  
3. Create a dictionary of students scoring between 70 and 89.  
4. Count students scoring below 70.  
5. Find the highest marks manually. 
"""

students = [ 
    ["Rahul", 78], 
    ["Priya", 85], 
    ["Amit", 72], 
    ["Neha", 92], 
    ["Riya", 81], 
    ["Karan", 68] 
]

dictionary = {}
more_90 = {}
between_70_89 = {}
below_70 = {}
highest_mark = 0

for i in students:
    dictionary[i[0]] = i[1]

    if i[1] >= 90:
        more_90[i[0]] = i[1]

    if i[1] >= 70 and i[1] < 90:
        between_70_89[i[0]] = i[1]

    if i[1] < 70:
        below_70[i[0]] = i[1]

    if i[1] > highest_mark:
        highest_mark = i[1]

print(f"Dictionary : {dictionary}")
print(f"students scoring 90 or more : {more_90}")
print(f"students scoring between 70 and 89 : {between_70_89}")
print(f"students scoring below 70 : {below_70}")
print(f"highest marks : {highest_mark}")
```

```python
#8
"""
Then: 
1. Create Product Name → Stock Value dic onary.  
2. Find the product with the highest stock value manually.  
3. Find the product with the lowest stock value manually.  
4. Find the difference between them.  
 
Do not use max() or min(). 

For every product: 
Stock Value = Price × Quantity
"""

products = [ 
    ["Laptop", 55000, 3], 
    ["Mouse", 800, 10], 
    ["Keyboard", 1500, 6], 
    ["Monitor", 12000, 4], 
    ["Headphones", 2500, 8] 
]

pro_stock = {}

highest_pro = None
highest_stock = 0

lowest_pro = None
lowest_stock = float("inf")

for i in products:
    stock_value = i[1] * i[2]
    pro_stock[i[0]] = stock_value

    if stock_value > highest_stock:
        highest_stock = stock_value
        highest_pro = i[0]

    if stock_value < lowest_stock:
        lowest_stock = stock_value
        lowest_pro = i[0]

print(f"Product - Stock value : {pro_stock}")
print(f"product with the highest stock value :- {highest_pro} : {highest_stock}")
print(f"product with the lowest stock value :- {lowest_pro} : {lowest_stock}")
print(f"difference between them :- {abs(highest_stock - lowest_stock)}")
```

```python
#9
"""
Create a new list containing numbers that appear in both lists. 
 
Rules: 
 No set()  
 No intersection()  
 No &  
 Each number must appear only once in the new list.  
 
Expected result: 
[20, 30, 40] 
"""

list1 = [10, 20, 30, 40, 50, 20, 60, 30] 
list2 = [20, 30, 70, 80, 40, 20, 90] 

new = []

for i in list1:
    if i in list2 and i not in new:
        new.append(i)

print(new)
```

```python
#10
"""
Generate 15 random integers between 1 and 100. 
 
Find manually: 
1. Total  
2. Average  
3. Count below average  
4. Count above average  
5. Count even  
6. Count odd  
7. Largest  
8. Smallest  
9. Difference between largest and smallest  
10. Create a new list containing numbers greater than average and even  
 
Do not use: 
 sum()  
 max()  
 min()  
 sorted()
"""

import random


num_list = []
total = 0
count = 0
below_avg = 0
above_avg = 0
even = 0
odd = 0 
largest = 0
smallest = float("inf")
new = []

for i in range(15):
    gen = random.randint(1,100)
    num_list.append(gen)

    total += gen
    count += 1

avg = total / count

for i in num_list:
    if i < avg:
        below_avg += 1

    if i > avg:
        above_avg += 1

    if i > largest:
        largest = i

    if i < smallest:
        smallest = i

    if i > avg and i % 2 == 0:
        new.append(i)

    if i % 2 == 0:
            even += 1
    
    if i % 2 != 0:
        odd += 1

print(f"List : {num_list}")
print(F"Total : {total}")
print(f"Average : {avg}")
print(f"below average : {below_avg}")
print(f"above average : {above_avg}")
print(f"even : {even}")
print(f"odd : {odd}")
print(f"largest : {largest}")
print(f"smallest : {smallest}")
print(f"Difference between largest and smallest : {abs(largest - smallest)}")
print(f"list containing numbers greater than average and even : {new}")
```
