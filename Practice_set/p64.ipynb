```python
#1
"""
Number Classifica on 

numbers = [24, -15, 0, 37, -8, 42, 0, -29, 16, 53, -4, 11] 

1. Create separate lists for positive, negative, and zero values. 
2. Count positive, negative, and zero values. 
3. Find the largest positive number manually. 
4. Find the smallest negative number manually.
"""

positive = []
negative = []
zero = []

count_positive = 0
count_negative = 0
count_zero = 0

largest_positive = 0
smallest_negative = float('inf')

numbers = [24, -15, 0, 37, -8, 42, 0, -29, 16, 53, -4, 11] 

for i in numbers:
    if i > 0:
        positive.append(i)
        count_positive += 1
        if i > largest_positive:
            largest_positive = i

    elif i < 0:
        negative.append(i)
        count_negative += 1
        if i < smallest_negative:
            smallest_negative = i

    else:
        zero.append(i)
        count_zero += 1

print(f"Positive numbers : {positive}")
print(f"Negative numbers : {negative}")
print(f"Zero values : {zero}")

print(f"\nCount of positive numbers : {count_positive}")
print(f"Count of negative numbers : {count_negative}")
print(f"Count of Zero values : {count_zero}")

print(f"\nLargest Positive number : {largest_positive}")
print(f"Smallest Negative number : {smallest_negative}")
```

```python
#2
"""
Running Account Balance 

transactions = [800, -250, 400, -150, 600, -350, 200, -100] 
 
Treat positive values as deposits and negative values as withdrawals. 
1. Create a list of the running balance after each transaction. 
2. Calculate total deposits and total withdrawals manually. 
3. Find the final balance. 
 
Restriction: No sum().
"""

transactions = [800, -250, 400, -150, 600, -350, 200, -100]

balance = 0
deposits = 0
withdrawals = 0
running_balances = []

for i in transactions:
    balance += i

    if i > 0:
        deposits += i
    elif i < 0:
        withdrawals += abs(i)

    running_balances.append(balance)

print(f"Running balances: {running_balances}")
print(f"Total deposits: {deposits}")
print(f"Total withdrawals: {withdrawals}")
print(f"Final balance: {balance}")
```

```python
#3 
"""
Transform Numbers 

numbers = [12, 19, 24, 31, 46, 53, 68, 75, 82, 97] 

1. Put even numbers multiplied by 2 into a new list. 
2. Put odd numbers multiplied by 3 into another list. 
3. Count even and odd numbers separately. 
4. Find the difference between the two counts.
"""

by2 = []
by3 = []

count_odd = 0
count_even = 0


numbers = [12, 19, 24, 31, 46, 53, 68, 75, 82, 97] 

for i in numbers:

    if i%2 == 0:
        by2.append(i*2)
        count_even += 1

    elif i%2 != 0:
        by3.append(i*3)
        count_odd += 1

print(f"Even numbers list : {by2}") 
print(f"Odd numbers list : {by3}") 

print(f"\nCount of even numbers : {count_even}")
print(f"Count of odd numbers : {count_odd}")

print(f"\nDifference between them : {abs(count_odd - count_even)}")
```

```python
#4
"""
String Analysis 

text = "Jemin Data Analyst 2026" 
 
Count: 
1. Vowels 
2. Consonants 
3. Digits 
4. Spaces 
5. Uppercase letters 
6. Lowercase ltters 

Scan the string character by character. Do not use external libraries.
"""

vowels = 0
consonants = 0
Digits = 0
Spaces = 0
uppercase_letters = 0
lowercase_letters = 0

vowel = ['I','O','U','A','E','i','o','u','a','e']

text = "Jemin Data Analyst 2026" 

for i in text:

    if i in vowel:
        vowels += 1

    if i.isalpha() and i not in vowel:
        consonants += 1

    if i.isdigit():
        Digits += 1

    if i == " ":
        Spaces += 1

    if i.isupper():
        uppercase_letters += 1

    if i.islower():
        lowercase_letters += 1

print(f"Vowels : {vowels}")
print(f"Consonants : {consonants}")
print(f"Digits : {Digits}")
print(f"Spaces : {Spaces}")
print(f"Uppercase Letters : {uppercase_letters}")
print(f"Lowercase Letters : {lowercase_letters}")
```

```python
#5
"""
Frequency Analysis 

numbers = [14, 7, 14, 22, 7, 9, 14, 22, 7, 30, 9, 7] 

1. Create a new list containing each number only once, preserving order. 
2. Create a dictionary showing each number’s frequency. 
3. Find the number with the highest frequency manually. 
4. Find how many numbers appear exactly once. 

Restrictions: No count(), set(), or dict.fromkeys().
"""

unique = []
freq = {}

highest_num = 0
highest_freq = 0

unique_count = 0

numbers = [14, 7, 14, 22, 7, 9, 14, 22, 7, 30, 9, 7] 

for i in numbers:
    if i not in unique:
        unique.append(i)

    if i in freq:
        freq[i] += 1
    else:
        freq[i] = 1

for key,value in freq.items():
    if value > highest_freq:
        highest_freq = value
        highest_num = key

    if value == 1:
        unique_count += 1


print(f"Unique : {unique}")
print(f"Frequency : {freq}")
print(f"Highest frequency with its number : {highest_num}:{highest_freq}")
print(f"Exactly once number : {unique_count}")
```

```python
#6
"""
First Non-Repeated Number

numbers = [8, 12, 8, 25, 18, 12, 30, 25, 40, 18, 50] 

Find the first number, scanning from left to right, that appears exactly once. 

Restriction: No count().
"""

numbers = [8, 12, 8, 25, 18, 12, 30, 25, 40, 18, 50]

ones = {}

for i in numbers:
    if i in ones:
        ones[i] += 1
    else:
        ones[i] = 1

for i in numbers:
    if ones[i] == 1:
        print(f"First non-repeated number: {i}")
        break

print(f"Frequency dictionary: {ones}")
```

```python
#7
"""
Student Performance 

students = [ 
    ["Rahul", 78], 
    ["Priya", 92], 
    ["Amit", 65], 
    ["Neha", 88], 
    ["Riya", 95], 
    ["Karan", 72], 
    ["Mehul", 84] 
] 

1. Convert the data into a student-to-marks dictionary. 
2. Create dictionaries for marks of 90 or above, 70–89, and below 70. 
3. Count students scoring below 70. 
4. Find the highest scorer manually. 
5. Calculate average marks manually. 

Restrictions: No sum() or max().
"""

marks_dict = {}

above_90 = {}
between = {}
below_70 = {}

Highest_score = 0
Highest_student = None

total = 0
count = 0

count_below_70 = 0

students = [ 
    ["Rahul", 78], 
    ["Priya", 92], 
    ["Amit", 65], 
    ["Neha", 88], 
    ["Riya", 95], 
    ["Karan", 72], 
    ["Mehul", 84] 
]

for i in students:
    marks_dict[i[0]] = i[1] 

    if i[1] >= 90:
        above_90[i[0]] = i[1]

    elif i[1] >= 70 and i[1] < 90:
        between[i[0]] = i[1]

    elif i[1] < 70:
        below_70[i[0]] = i[1]

    if i[1] < 70:
        count_below_70 += 1

    if i[1] > Highest_score:
        Highest_score = i[1]
        Highest_student = i[0]

    total += i[1]
    count +=1
avg = total/count

print(f"Student marks Dictionary : {marks_dict}")

print(f"\nAbove 90 or more marks Dictionary : {above_90}")
print(f"Between 70-89 marks Dictionary : {between}")
print(f"Below 70 marks Dictionary : {below_70}")

print(f"\nHighest score with their name : {Highest_student}:{Highest_score}")

print(f"\nCount of below 70 is {count_below_70}")

print(f"\nAverage marks is {avg}")
```

```python
#8
"""
Inventory Value 

products = [ 
    ["Laptop", 55000, 4], 
    ["Mouse", 800, 15], 
    ["Keyboard", 1500, 8], 
    ["Monitor", 12000, 5], 
    ["Headphones", 2500, 10], 
    ["Webcam", 3000, 6] 
] 

Calculate each product’s stock value: 
Stock Value = Price × Quantity

Then: 
1. Create a product-to-stock-value dictionary. 
2. Find the product with the highest stock value. 
3. Find the product with the lowest stock value. 
4. Calculate the difference between them. 

Restrictions: No max() or min().
"""

dictionary = {}

highest_stock_value = 0
highest_stock_product = None

lowest_stock_value = float('inf')
lowest_stock_product = None

products = [ 
    ["Laptop", 55000, 4], 
    ["Mouse", 800, 15], 
    ["Keyboard", 1500, 8], 
    ["Monitor", 12000, 5], 
    ["Headphones", 2500, 10], 
    ["Webcam", 3000, 6] 
] 

for i in products:
    dictionary[i[0]] = i[1] * i[2]

    if i[1] * i[2] > highest_stock_value:
        highest_stock_value = i[1] * i[2]
        highest_stock_product = i[0]

    if i[1] * i[2] < lowest_stock_value:
        lowest_stock_value = i[1] * i[2]
        lowest_stock_product = i[0]

dif = abs(highest_stock_value - lowest_stock_value)

print(dictionary)

print(f"\nHighest stock product with stock :- {highest_stock_product} : {highest_stock_value}")
print(f"Lowest stock product with stock :- {lowest_stock_product} : {lowest_stock_value}")

print(f"Difference between them :- {dif}")
```
